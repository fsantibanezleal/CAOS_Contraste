## Method

The notebook states each claim, shows the evidence of the committed files with the engine, and shows a control or a counterfactual that makes the claim testable: for the UF claim the cost of not treating the UF as a currency, for the breach-test claim histories where the test must be green and where it must be red. Each claim ends with its limit. The claims and their evidence are stated in the SDD's list of claims; the notebook does not state more than the SDD does.

- **The UF claim.** The reajustabilidad risk of RAN 21-13 Annex 1 as a separate currency (the UR) with its own shocks, floor, the indexation charge (2.5 percent of the absolute net UF position) and the 50 percent offset of the peso delta against the UR delta. Evidence: `data/derived/C15/long-uf.json`, `floors-active.json`, `offset-on-off.json`.
- **The breach-test claim.** The test `irrbb.behavioural_change_at_breach`, which turns a finding of the Federal Reserve's review of the failed bank into a validation test. Evidence: the committed parameters of `data/derived/C14/models-svb-2021-2022.json` and the breach flags of the three dates.

## Comparison

**For the UF claim.** The Basel standard has 21 currencies in its shock table and a rule for aggregating them; it has no unit like the UF, an indexed unit of account whose value in the local currency follows inflation, and no currency of the local jurisdiction that is missing from the table gets a size (SRP31.90, Table 2; BCBS d578, https://www.bis.org/publications/202407-standards-recalibration-shocks-interest-rate-risk-banking-book.pdf). The Chilean regulator's Annex 1 of chapter 21-13 (https://www.cmfchile.cl/portal/principal/623/articles-29808_doc_pdf.pdf) fills that gap with the UR. The research found the Chilean approach described as "a genuine differentiator" (dossier 03, F.2) and found no public implementation of the Annex; that is a finding of the dossier's search and stays **UNVERIFIED** beyond it. The claim is regional, not global.

**For the breach-test claim.** The Federal Reserve's review of Silicon Valley Bank states the practice as a finding: the bank "had breached its long-term IRR limits on and off since 2017" and "in April 2022 ... made counterintuitive modeling assumptions about the duration of deposits to address the limit breach rather than managing the actual risk" (https://www.federalreserve.gov/publications/files/svb-review-20230428.pdf; dossier 05, D.7). The research proposes the test ("a Contraste validation test should therefore flag any change in NMD behavioural parameters that coincides with a limit breach") and cites no tool that implements it.

The first cell below shows the first fact for the UF claim with the engine: the Basel table has no entry for the peso or for the UF, and the engine refuses a size for them instead of inventing one.

````python
# The Basel table has no entry for the peso or the UF: the engine refuses a size; the Chilean table has both
for calibration in ("2016", "2024"):
    table = shocks.shock_table(calibration)
    has = [("no", "yes")[c in table] for c in ("CLP", "CLF")]
    print(calibration, "| currencies:", len(table), "| CLP:", has[0], "| CLF:", has[1])
for currency in ("CLP", "CLF"):
    try:
        shocks.shock_size(currency, "2024")
    except ValueError as error:
        print(currency, "the engine refuses a size:", type(error).__name__)
for currency in ("CLP", "CLF"):
    z = shocks.shock_size(currency, "cmf-annex1")
    print(f"{currency} cmf-annex1: {z.parallel_bp:.0f}/{z.short_bp:.0f}/{z.long_bp:.0f} bp | origin {z.origin}")
````

## Evidence

### The UF as its own currency

**What the claim adds.** Annex 1 of chapter 21-13 treats the reajustabilidad risk as a separate currency with its own shocks, its own floor, the indexation charge and the 50 percent offset; the engine implements it, and the committed C15 variants recompute from it. The first cell shows the separation in the committed documents: the sizes and the floor of each currency and the curve each one is discounted on. The second shows why the floor must be the currency's own: the UR curve is a real-rate curve, around 1.2 to 2.8 percent, and a downward shock of 200 bp pushes its short end below zero, where the peso floor of 0 percent would bind in nine bands and the UR floor of -2 percent in none.

````python
# The sizes, the floor and the curve of each currency in the committed documents, and the engine's size for each
out = variants["short-uf"]["outputs"]
rates = {"CLP": np.array(out["curves"]["clp"]["zero_rates_at_midpoints"])}
rates["UR"] = np.array(out["curves"]["ur"]["zero_rates_at_midpoints"])
floors = {"CLP": out["floors"]["other_floor"], "UR": out["floors"]["ur_floor"]}
print("EVE sizes (par/short/long) | NII size | floor | rate band 1 | rate band 19")
for committed in out["eve_shock_sizes"]:
    leg = "CLP" if committed["currency"] == "CLP" else "UR"
    nii = next(x["parallel_bp"] for x in out["nii_shock_sizes"] if x["currency"] == committed["currency"])
    z = shocks.shock_size(committed["currency"], "cmf-annex1")
    print(f"{leg:4s} {z.parallel_bp:.0f}/{z.short_bp:.0f}/{z.long_bp:.0f} | {nii:.0f} | {floors[leg]:.2f}")
    print(f"    {rates[leg][0]:.5f} | {rates[leg][18]:.5f}")
fig, ax = plt.subplots(figsize=(8, 4.2))
for leg, colour in (("CLP", "tab:blue"), ("UR", "tab:green")):
    ax.plot(MID, rates[leg] * 100, color=colour, label=leg)
    ax.axhline(floors[leg] * 100, color=colour, linestyle=":", label="floor " + leg)
ax.set_xlabel("midpoint of the band (years)")
ax.set_ylabel("zero rate (%)")
ax.set_title("The two currencies: curve and floor")
ax.legend(fontsize=8)
plt.show()
````

The two solid lines are the curves each currency is discounted on, the peso curve between about 4.5 and 5.9 percent and the real UF curve between 1.2 and 2.8 percent, and the dotted lines are the floors of each currency: 0 percent for the peso and -2 percent for the UR. The sizes printed above differ too, 300/400/200 bp for the peso and 200/250/150 bp for the UR in the EVE, and the same 300 bp in the NII. Everything that Annex 1 separates is visible here: a curve, a set of sizes and a floor per currency, which a single-currency treatment would replace by the peso's. The figure shows the inputs of the claim; the next step shows what they do to a result.

Now the floor. The next cell applies a 200 bp fall to the UR curve of the first variant with the peso floor and then with the UR floor, and lists the bands where each floor binds.

````python
# Apply a 200 bp fall to the UR curve with the peso floor and with the UR floor and find where each binds
shape = np.full(19, -200.0)
raw = shocks.apply_floor(rates["UR"], shape, None)
fig, ax = plt.subplots(figsize=(8, 4.2))
ax.plot(MID, rates["UR"] * 100, label="UR curve")
ax.plot(MID, raw * 100, linestyle="--", label="UR curve after -200 bp, no floor")
for label, floor in (("peso floor (0)", 0.0), ("UR floor (-2 percent)", -0.02)):
    floored = shocks.apply_floor(rates["UR"], shape, floor)
    bound = [int(k) + 1 for k in np.flatnonzero(np.abs(floored - raw) > 1e-12)]
    print(f"{label:28s} bands where it binds: {bound}")
    ax.plot(MID, floored * 100, label=label)
print("lowest rate of the UR curve after -200 bp, before the floor:", round(float(raw.min()), 5))
ax.set_xlabel("midpoint of the band (years)")
ax.set_ylabel("zero rate (%)")
ax.set_title("A 200 bp fall on the UR curve")
ax.legend(fontsize=8)
plt.show()
````

The dashed line is the real curve after a 200 bp fall, with its short end at -0.795 percent. The peso floor lifts the first nine bands back to zero, the line that sits on the axis, and so would change a real-rate curve that has no reason to be non-negative; the UR floor of -2 percent lies below the whole shocked curve and binds in no band. That is why the floor has to be the currency's own: the peso floor applied to the UF would manufacture a loss in a currency whose rates were low but legitimately positive to begin with and negative after the shock. The figure shows one shock on one curve at one date and does not show that either floor binds in the committed variants, which the limit of the UF claim below says plainly.

**The offset and the charge.** The offset rule gives half credit to the gain leg when the two deltas have opposite signs. The `offset-on-off` variant holds the pair with and without the offset, and the same engine functions give the Annex measures both ways for the four variants. The charge is added once to the delta NII whatever the offset. The next cell prints both measures with and without the offset, the share of the delta NII that is the charge, and the closest approach of any post-shock rate to its floor, and draws the effect of the offset.

````python
# The Annex measures with and without the offset, the charge as a share of the delta NII, and the closest floor
annex = {}
for v, d in variants.items():
    o = d["outputs"]
    d_clp, d_ur = (np.array([r["pair"][k] for r in o["eve"]["rows"]]) for k in ("delta_clp", "delta_ur"))
    n_clp, n_ur = (np.array([r["pair"][k] for r in o["nii"]["rows"]]) for k in ("delta_clp", "delta_ur"))
    eve_on, eve_off = (eve_nii.annex1_delta_eve(d_clp, d_ur, flag) for flag in (True, False))
    pn = o["indexation_charge"]["pn_ur"]
    nii_on, nii_off = (eve_nii.annex1_delta_nii(n_clp, n_ur, pn, flag) for flag in (True, False))
    closest = min(x["distance_bp"] for x in o["floors"]["distances"])
    annex[v] = (eve_on.delta_eve, nii_on.delta_nii, eve_off.delta_eve, nii_off.delta_nii)
    print(f"{v:14s} ΔEVE {eve_on.delta_eve:10,.0f} {eve_off.delta_eve:10,.0f}")
    print(f"    ΔNII {nii_on.delta_nii:10,.0f} {nii_off.delta_nii:10,.0f}")
    print(f"    charge / ΔNII {nii_on.indexation_charge / nii_on.delta_nii:.4f} | closest floor (bp) {closest:.2f}")
x = np.arange(len(annex))
fig, (left, right) = plt.subplots(1, 2, figsize=(11, 4))
panels = ((left, (0, 2), "delta EVE (CLP millions)"), (right, (1, 3), "delta NII (CLP millions)"))
for ax, (k_on, k_off), ylabel in panels:
    ax.bar(x - 0.2, [annex[v][k_on] for v in annex], 0.4, label="with offset")
    ax.bar(x + 0.2, [annex[v][k_off] for v in annex], 0.4, label="without offset")
    ax.set_xticks(x, list(annex), rotation=20, ha="right", fontsize=8)
    ax.set_ylabel(ylabel)
left.set_title("Offset effect on the Annex 1 measures")
left.legend(fontsize=8)
fig.tight_layout()
plt.show()
````

The offset acts mostly on the NII. In the short-UF variant it lowers the delta NII from 14,597 to 11,067, 24 percent, and in the offset-on-off variant it lowers the delta EVE from 94,721 to 81,173, 14 percent; in the short-UF variant it does not change the delta EVE because the worst shock has both legs losing. The charge is 55.9, 55.1 and 56.1 percent of the delta NII in the three variants with a net UF position of 30 percent of Tier 1, and 15.6 percent in the variant with 5 percent. The closest approach of any post-shock rate to its floor is 20.48 bp in all four. The figure shows the effect of one rule on four synthetic variants, not a statement about any real bank.

**The counterfactual: what a single-currency treatment loses.** Without the UR, a bank would put its UF flows, converted to pesos, with the peso flows, discount them all on the peso curve, shock them with the peso sizes and the zero floor, and add no charge: one currency. The cell does exactly that with the engine for the four variants, using the committed net flows of each leg, and compares the two measures and the ratios to Tier 1. It is a counterfactual, not a rival method: it shows what is lost when the UF is not a currency of its own. The difference has four parts that act together, the curve (nominal instead of real), the sizes (peso sizes for the UF flows), the full netting of the legs inside a single sum, and the missing charge.

````python
# Annex 1 measures against a single-currency treatment of the same flows: peso curve, sizes and floor, no charge
peso = shocks.shock_size("CLP", "cmf-annex1")
single = {}
for v, d in variants.items():
    o = d["outputs"]
    uf = o["ur_currency"]["uf_value"]
    rows = {leg["leg"]: [r for item in leg["items"] for r in item["cash_flows"]] for leg in o["legs"]}
    vec = {k: np.bincount([r["bucket"] - 1 for r in rs], [r["amount"] for r in rs], 19) for k, rs in rows.items()}
    total = vec["CLP"] + uf * vec["UR"]
    r_peso = np.array(o["curves"]["clp"]["zero_rates_at_midpoints"])
    loss_eve = []
    for row in o["eve"]["rows"]:
        net = np.array(row["clp"]["net_cash_flows"]) + np.array(row["ur"]["net_cash_flows"])
        shape = shocks.shock_curve(row["shock"], peso.parallel_bp, peso.short_bp, peso.long_bp, MID)
        shocked = shocks.apply_floor(r_peso, shape, 0.0)
        loss_eve.append(eve_nii.delta_eve(total, net, r_peso, shocked, MID))
    flat = [shocks.apply_floor(r_peso, np.full(19, bp), 0.0) for bp in (300.0, -300.0)]
    loss_nii = [eve_nii.delta_nii(total, r_peso, shocked, MID) for shocked in flat]
    single[v] = (max(0.0, max(loss_eve)), max(0.0, max(loss_nii)))
    print(f"{v:14s} ΔEVE Annex 1 {annex[v][0]:10,.0f} | single currency {single[v][0]:10,.0f}")
    print(f"    ΔNII Annex 1 {annex[v][1]:10,.0f} | single currency {single[v][1]:10,.0f}")
````

The next cell draws the four measures of each variant as a share of Tier 1, with the 15 percent line of the EVE and the 5 percent line of the NII.

````python
# Draw the Annex 1 measures and the single-currency counterfactual as a share of Tier 1
tier = {v: d["outputs"]["thresholds"]["tier1"] for v, d in variants.items()}
series = []
specs = [("ΔEVE Annex 1", annex, 0), ("ΔEVE single currency", single, 0)]
specs += [("ΔNII Annex 1", annex, 1), ("ΔNII single currency", single, 1)]
for label, source, k in specs:
    series.append((label, [source[v][k] / tier[v] for v in variants]))
x = np.arange(len(variants))
fig, ax = plt.subplots(figsize=(9, 4.2))
for k, (label, values) in enumerate(series):
    ax.bar(x + (k - 1.5) * 0.2, [100 * value for value in values], 0.2, label=label)
ax.axhline(15, color="tab:red", linestyle="--", linewidth=0.8)
ax.axhline(5, color="tab:orange", linestyle="--", linewidth=0.8)
ax.set_xticks(x, list(variants), rotation=20, ha="right", fontsize=8)
ax.set_ylabel("percent of Tier 1")
ax.set_title("The UF as a currency against one currency")
ax.legend(fontsize=7)
plt.show()
````

Treating the UF flows as pesos gives a delta NII of 1,356, 1,648 and 1,254 for the short-UF, long-UF and offset-on-off variants against the Annex 1 values of 11,067, 11,213 and 11,016: 6.8 to 8.8 times smaller on this committed book (one seed; over the 30 synthetic books per variant of the UF paper, `manuscripts/uf-as-its-own-currency-in-irrbb/`, the ratio of the mean measures is 4.9, 5.5 and 5.7 and the delta EVE is 5.7, 15.4 and 24.2 percent smaller), because the single-currency sum has no charge and lets the legs net out fully. The delta EVE is 64,213, 84,710 and 59,422 against 79,643, 93,816 and 81,173: 10 to 27 percent smaller. In the floors-active variant, with a net UF position of 5 percent of Tier 1, the gap is smaller, 4,653 against 6,606 for the NII. In these four variants no ratio crosses its line under either treatment, so the claim is not that a bank would pass or fail differently but that the measure the regulator defines is not the one a single-currency calculation gives.

**The limit of the UF claim.** The evidence shows that the engine implements Annex 1 as a separate currency and that the committed variants recompute from it. It does not show that the Annex is applied the way the CMF applies it: there is no second implementation to reconcile with, and the dossier's search found none (UNVERIFIED beyond that search). The floors are implemented and tested but never bind on the committed curves, the closest being 20.48 bp away, so their effect on a number is not demonstrated by this data. The bank is synthetic. And the claim is regional: the standard has no UF-like unit, and only Chile's Annex needs one here.

### A test that flags a behavioural assumption changed at a limit breach

**What the claim adds.** `irrbb.behavioural_change_at_breach` takes the series of a behavioural parameter, the dates and a breach flag per date, and counts the dates where the parameter changed and the flag is true. It is red at one or more, green at none, and `not_evaluated` with fewer than two dates. The first cell shows that the test discriminates, on five constructed histories whose answer is known: a parameter held fixed through a breach, a parameter that changes with no breach, a parameter changed at a breach, a parameter changed before the breach but not at it, and a single date. These are constructed numbers that exercise the rule, not data of any bank.

````python
# Five constructed histories with a known answer: fixed, changed with no breach, changed at or before a breach
days = np.array(["2021-12-31", "2022-06-30", "2022-12-31"], dtype="datetime64[D]")
cases = [
    ("fixed through a breach", [4.0, 4.0, 4.0], [True, True, True], days),
    ("changed, no breach", [4.0, 4.5, 5.0], [False, False, False], days),
    ("changed at a breach", [4.0, 4.0, 4.5], [False, False, True], days),
    ("changed before the breach", [4.0, 4.5, 4.5], [False, False, True], days),
    ("a single date", [4.0], [True], days[:1]),
]
results = []
for label, values, flags, when in cases:
    run = rv_irrbb.behavioural_change_at_breach(np.array(values), when, np.array(flags, dtype=bool))
    results.append(run)
    print(f"{label:28s} {values} {[NAME[f] for f in flags]}")
    print(f"    {LIGHT[run.light.value]} | flagged dates: {run.extras.get('flagged_dates', [])}")
tint = {"green": "tab:green", "red": "tab:red", "not_evaluated": "lightgrey"}
fig, ax = plt.subplots(figsize=(9, 4))
counts = [np.nan_to_num(run.statistic) for run in results]
ax.bar([c[0] for c in cases], counts, color=[tint[run.light.value] for run in results])
ax.set_ylabel("flagged changes")
ax.set_title("The breach test on five constructed histories")
ax.tick_params(axis="x", labelsize=8)
plt.show()
````

Only the third history is red, the parameter that moves from 4.0 to 4.5 at the date where the limit is breached, and the test names that date, 2022-12-31. The parameter held fixed through a breach is green, so a breach alone does not trigger it; the parameter that changes with no breach is green, so a change alone does not either; and the parameter that changed before the breach and then held is green, because the change did not coincide with it. A single date cannot be compared and the test says so instead of returning a number. These are constructed numbers that exercise the rule: they show the test discriminates, not that anyone has done what the third history does.

**The evidence on the committed bank.** The committed model document lists, for two consecutive pairs of dates, the behavioural parameters of the wholesale deposits and whether each changed: the core share and the core maturity that Contraste requests (stated, synthetic) and the weighted average maturity of the fitted replicating portfolio. The cell runs the engine test on the three series of the three dates with the breach flag of each date from the committed scenarios, and prints the flagged dates. Then it runs the same test on the average maturity with the breach flags all set to false, as a control: the verdict is decided by the flag, not by the movement alone.

````python
# Run the breach test on three parameters of the wholesale deposits at the three dates, and a control
dates = np.array([d["outputs"]["valuation_date"] for d in docs.values()], dtype="datetime64[D]")
breach = np.array([bool(d["outputs"]["scenarios"][0]["outlier"]["breached"]) for d in docs.values()])
wholesale = [next(s for s in d["outputs"]["nmd_segments"] if s["category"] == "wholesale") for d in docs.values()]
by_date = {m["valuation_date"]: m["weighted_average_maturity_years"] for m in fit["nmd_models"]}
wam = [by_date[str(day)] for day in dates]
series = {"requested core share": [s["requested_core_share"] for s in wholesale]}
series["requested core maturity (y)"] = [s["requested_core_maturity_years"] for s in wholesale]
series["fitted average maturity (y)"] = wam
print("dates:", [str(x) for x in dates], "| breach flags:", [NAME[bool(b)] for b in breach])
for name, values in series.items():
    run = rv_irrbb.behavioural_change_at_breach(np.array(values), dates, breach)
    print(f"{name:34s} {[round(v, 4) for v in values]}")
    print(f"    {LIGHT[run.light.value]} | flagged dates: {run.extras['flagged_dates']}")
control = rv_irrbb.behavioural_change_at_breach(np.array(wam), dates, np.zeros(3, dtype=bool))
print("control: the fitted maturity with no breach ->", LIGHT[control.light.value])
fig, ax = plt.subplots(figsize=(7, 4))
ax.plot(dates, wam, marker="o", color="tab:blue", label="fitted average maturity")
for day in run.extras["flagged_dates"]:
    ax.axvline(np.datetime64(day), color="tab:red", linestyle="--", linewidth=1)
ax.axvline(dates[0], color="tab:red", linestyle="--", linewidth=1, label="flagged change")
ax.set_ylabel("average maturity (years)")
ax.set_title("A change that coincides with a breach")
ax.legend(fontsize=8)
plt.show()
````

Two of the three parameters do not move: the share of the stable deposits that Contraste requests stays at 0.6 and its maturity at 5 years, and the test is green for both although the limit is breached at every date. The fitted average maturity does move, 4.7595, 4.2049 and 5.0 years, and the test is red at 2022-06-30 and 2022-12-31, the dates of the dashed lines; with the breach flags set to false the same series is green, so the flag decides the verdict. The figure shows that the test operates on the committed parameters. It does not show that anyone tuned the fitted maturity: it is re-estimated at each date and so differs by construction.

**The limit of the breach-test claim.** The red row is not a finding about the bank. In C14 the bank's outlier test is breached at the three dates, so the breach flag never varies; the parameter that moves is the fitted replicating-portfolio maturity, which is re-estimated at each date and therefore differs by construction, while the two parameters Contraste states are held fixed and are not flagged. What the evidence shows is that the test separates a stated assumption held through a breach from a series that moves while the limit is breached, and that its verdict is decided by the breach flag; the same series with no breach is green. It does not show that anyone chose the fitted maturity to relieve a breach, and the bank's behaviour is synthetic. A red is a prompt for the validator to ask who moved the parameter, when, why, and with what approval, and whether the new value is supported by the data; the first answers are in the model-governance record, not in the test.

### Where the method changes or breaks

The test that turns the Federal Reserve's finding into a count compares each value of a behavioural parameter with the previous one for exact inequality (Board of Governors of the Federal Reserve System, 2023, https://www.federalreserve.gov/publications/files/svb-review-20230428.pdf). With $v_t$ the value of the parameter at date $t$ and $b_t$ equal to 1 when a limit was breached in the window ending at $t$, it counts

$$N = \sum_{t=2}^{T} \mathbf{1}\,[\,v_t \ne v_{t-1}\,]\;b_t$$

and the light is red when $N \ge 1$. Because the comparison is exact, the verdict depends on the resolution at which the parameter is stored, and on any rounding $\delta$ applied before the test, which changes the count to

$$N_{\delta} = \sum_{t=2}^{T} \mathbf{1}\,[\,\tilde v_t \ne \tilde v_{t-1}\,]\;b_t, \qquad \tilde v_t = \delta\,\operatorname{round}\!\left(\frac{v_t}{\delta}\right)$$

where $\delta$ is the rounding step in the unit of the parameter (years here) and $\operatorname{round}$ the nearest integer. The first cell adds a change of growing size to a parameter held at 4.0 years and read at three dates with a breach at the last one, from 10^-16 to 1 year, and then repeats a history of three dates under a breach at every date with a parameter that carries only numerical noise of growing standard deviation, 1,000 histories per value. The second cell rounds the fitted maturity of the C14 deposit model to steps from 0.01 to 5 years before the test and draws both sweeps.

````python
# Sweep the size of a change at a breach, and the noise of a re-estimated parameter
days3 = np.array(["2021-12-31", "2022-06-30", "2022-12-31"], dtype="datetime64[D]")
flag3 = np.array([False, False, True])
print("change added at the third date | light | flagged changes")
for eps in (0.0, 1e-16, 1e-15, 1e-12, 1e-9, 1e-6, 1e-3, 0.1, 1.0):
    run = rv_irrbb.behavioural_change_at_breach(np.array([4.0, 4.0, 4.0 + eps]), days3, flag3)
    print(f"{eps:8.0e} | {LIGHT[run.light.value]:5s} | {run.statistic:.0f}")
noise_rng = np.random.default_rng(11)
all_flags = np.array([True, True, True])
red_share = {}
print("noise of the re-estimation (sd) | share of 1000 histories that are red")
for sd in (0.0, 1e-16, 1e-14, 1e-9, 1e-4):
    reds = 0
    for _ in range(1000):
        values = 4.0 + sd * noise_rng.standard_normal(3)
        reds += rv_irrbb.behavioural_change_at_breach(values, days3, all_flags).light.value == "red"
    red_share[sd] = reds / 1000
    print(f"{sd:8.0e} | {red_share[sd]:.3f}")
````

The first table says that the test does not read the size of a change. A change of 10^-16 years is below the spacing of the floating-point numbers around 4.0 (about 8.9 x 10^-16), so the series does not change and the light is green with zero flagged changes; from 10^-15 to a full year the light is red with one flagged change, the same for a change of a femtoyear and of twelve months. The second table is the consequence for a fitted parameter: with no noise no history is red, with a noise of 10^-16 a share of 0.039 of the histories is red, and from 10^-14 to 10^-4 all 1,000 are red. A parameter that is re-estimated at each date, even to the last digit, is always flagged under a breach. The next cell rounds the fitted maturity of the committed deposit model before the test.

````python
# Round the fitted maturity to a grid before the test and draw both sweeps
steps = (0.0, 0.01, 0.1, 0.25, 0.5, 1.0, 2.0, 5.0)
flags_c14 = np.array([bool(d["outputs"]["scenarios"][0]["outlier"]["breached"]) for d in docs.values()])
dates_c14 = np.array([d["outputs"]["valuation_date"] for d in docs.values()], dtype="datetime64[D]")
by_date = {m["valuation_date"]: m["weighted_average_maturity_years"] for m in fit["nmd_models"]}
raw = np.array([by_date[str(day)] for day in dates_c14])
rounded = {}
print("rounding step (y) | rounded maturities | light | flagged changes")
for step in steps:
    values = raw if step == 0.0 else np.round(raw / step) * step
    run = rv_irrbb.behavioural_change_at_breach(values, dates_c14, flags_c14)
    rounded[step] = (values, run.statistic)
    print(f"{step:5.2f} | {np.round(values, 3)} | {LIGHT[run.light.value]:5s} | {run.statistic:.0f}")
fig, (left, right) = plt.subplots(1, 2, figsize=(11, 4.2))
left.semilogx([max(s, 1e-17) for s in red_share], list(red_share.values()), marker="o", color="tab:red")
left.set_xlabel("noise of the re-estimation (standard deviation)")
left.set_ylabel("share of histories that are red")
left.set_title("Noise alone turns the light")
for step in (0.0, 0.5, 2.0, 5.0):
    right.plot(dates_c14, rounded[step][0], marker="o", label=f"step {step:g}")
right.set_ylabel("rounded maturity (years)")
right.set_title("The verdict depends on the resolution")
right.legend(fontsize=8)
fig.tight_layout()
plt.show()
````

The rounding table shows the verdict as a function of the resolution. The fitted maturities 4.759, 4.205 and 5.0 give two flagged changes and a red light unfiltered and with steps of 0.01, 0.1 and 0.25 years; with steps of 0.5 and 1 year they become 5, 4 and 5 and the light stays red with two flagged changes; with a step of 2 years all three values become 4.0 and with a step of 5 years all become 5.0, and the light turns green with none. In the left panel the red share is zero at zero noise, near zero at 10^-16 and one from 10^-14, on a logarithmic axis; in the right panel the lines for steps of 0 and 0.5 years move between the dates and the lines for steps of 2 and 5 years are flat.

The method fails where the parameter is a fitted one. The strict comparison was built for a parameter that a bank states and holds, a core share or a core maturity, and there it separates a held value from a changed one, as the constructed histories showed. On a parameter that is re-estimated at every date it flags the pattern by construction, whatever the size of the move, and a rounding that would silence the noise also silences a real change of two years: the coarse step hides the very change the test exists to find. The red of the C14 series is therefore an invitation to ask whether anyone chose the number, which the data of this notebook cannot answer.

### Using it on your own data

The two proposed treatments of this notebook have inputs of their own. For the UF as a currency, a bank brings its two legs with the cash flows by band (the UF leg converted to pesos with the UF value of the date), the two curves, the sizes of the Annex 1 table for each currency (`shock_size(currency, "cmf-annex1")`, with `CLF` for the UF) and their floors; the engine returns the delta EVE of each leg and `annex1_delta_eve` pairs them. For the breach test, a bank brings the series of every behavioural parameter it fits, the dates of the valuation and the breach flag of each date, which in the case documents is `outputs.scenarios[0].outlier.breached`, and the committed model document lists the changes it recorded in `fit.nmd_parameter_changes` with the fields `parameter`, `from_date`, `to_date`, `from_value`, `to_value`, `limit_breach_in_window` and `flagged`.

The first cell below builds an own bank with a peso leg and a UF leg, shocks each leg with its own table and floor, and compares the Annex 1 value with a reading that puts all the flows in pesos and applies the peso sizes and the peso floor to them. It uses the committed curves of the case C15 as a stand-in for your curves. The second cell runs the breach test on every scalar of the committed fitted deposit model, not only on the one the document flags, and prints the parameters that the document records as flagged. The flows and the amounts of the bank are invented for the illustration; replace them with your own.

````python
# An own bank with a UF leg: the two-currency Annex 1 value against a single-currency reading
asset_o = {"CLP": ([0.5, 2.0, 6.0], [300.0, 400.0, 200.0]), "UR": ([3.0, 9.0], [0.010, 0.005])}
liab_o = {"CLP": ([0.1, 1.0], [500.0, 250.0]), "UR": ([0.2, 4.0], [0.0075, 0.006])}
uf_o = 40000.0
net_o = {}
for leg in asset_o:
    a_flow = slotting.slot_cash_flows(np.array(asset_o[leg][0]), np.array(asset_o[leg][1]))
    l_flow = slotting.slot_cash_flows(np.array(liab_o[leg][0]), np.array(liab_o[leg][1]))
    net_o[leg] = (a_flow - l_flow) * (uf_o if leg == "UR" else 1.0)
size_o = {"CLP": shocks.shock_size("CLP", "cmf-annex1"), "UR": shocks.shock_size("CLF", "cmf-annex1")}
d_o = {"CLP": [], "UR": []}
total_o = net_o["CLP"] + net_o["UR"]
single_o = []
for s in shocks.SHOCKS:
    for leg in d_o:
        z = size_o[leg]
        shape = shocks.shock_curve(s, z.parallel_bp, z.short_bp, z.long_bp, MID)
        shaped = shocks.apply_floor(rates[leg], shape, floors[leg])
        d_o[leg].append(eve_nii.delta_eve(net_o[leg], net_o[leg], rates[leg], shaped, MID))
    z = size_o["CLP"]
    peso = shocks.apply_floor(rates["CLP"], shocks.shock_curve(s, z.parallel_bp, z.short_bp, z.long_bp, MID), 0.0)
    single_o.append(eve_nii.delta_eve(total_o, total_o, rates["CLP"], peso, MID))
    print(f"{s:14s} CLP {d_o['CLP'][-1]:8.2f} | UR {d_o['UR'][-1]:8.2f} | single currency {single_o[-1]:8.2f}")
annex_o = eve_nii.annex1_delta_eve(np.array(d_o["CLP"]), np.array(d_o["UR"]), True)
print("Annex 1 delta EVE (two currencies):", round(annex_o.delta_eve, 2), "|", annex_o.worst_shock)
print("single-currency reading, largest loss:", round(max(0.0, max(single_o)), 2))
````

The delta EVE of the own bank is 35.97 and 27.81 million pesos on the two legs for the parallel-up shock, so the Annex 1 value is 63.78, the sum of the two, and its worst shock is the parallel up. The single-currency reading gives 67.89 for the same shock, 4.11 more, because it applies the peso sizes and the peso curve to the converted UF flows. In this bank the two-currency treatment gives the lower number, the opposite of what the committed variants of C15 showed in the comparison above, where the Annex 1 value was the higher one; the direction depends on the sizes, the curves, the legs and the indexation charge, and it is not a property of the method. The next cell turns to the breach test.

````python
# The breach test on every scalar of the fitted deposit model
models = fit["nmd_models"]
dates_m = np.array([m["valuation_date"] for m in models], dtype="datetime64[D]")
params = {"weighted average maturity (y)": [m["weighted_average_maturity_years"] for m in models]}
params["margin volatility"] = [m["margin_volatility"] for m in models]
params["deposit beta, up"] = [m["deposit_beta"]["up"] for m in models]
params["deposit beta, down"] = [m["deposit_beta"]["down"] for m in models]
print("fields of a recorded change:", list(fit["nmd_parameter_changes"][0]))
recorded = sorted({c["parameter"] for c in fit["nmd_parameter_changes"] if c["flagged"]})
print("parameters flagged in the document:", recorded)
for name, values in params.items():
    run = rv_irrbb.behavioural_change_at_breach(np.array(values), dates_m, flags_c14)
    print(f"{name:30s} {np.round(values, 5)} | {LIGHT[run.light.value]:5s} | {run.extras['flagged_dates']}")
unchanged = {k: bool(len(set(v)) == 1) for k, v in params.items()}
print("parameters that never change:", [k for k, v in unchanged.items() if v])
````

The document records one flagged parameter, the weighted average maturity of the replicating portfolio of the wholesale deposits. Run on every scalar of the same fitted model, the test flags three: the maturity (4.759, 4.205 and 5.0 years), the margin volatility (0.00803, 0.00768 and 0.00823) and the deposit beta for rising rates (0.02363, 0.07482 and 0.26112), each red with the changes flagged at 2022-06-30 and 2022-12-31. Only the beta for falling rates, which is 0.03147 at the three dates, is green, and it is the only one of the four that never changes. All four are fitted again at each date from the sample that ends there, so the three that move do so by construction, as the sweep above explained. The committed document tracks the maturity and the two stated requests and does not list the other two fitted parameters; a validator who wants the test to cover the whole fitted model can extend the list on the model document, and should read each red as the question of whether the change was chosen or estimated.

What to check on your own data: that the series are stored at the resolution you intend, since the test compares exactly; that the breach flag is defined for the window ending at each date and not for the date alone; that the parameters you pass include the stated ones and the fitted ones, with the distinction recorded in the document; that the dates are strictly increasing; and that the currency treatment you compare uses the table of the Annex for each leg and the floor of each leg, because a single-currency reading with the peso table is a different measure, not a check of the first.

## Results

**Answer:** yes to both, with limits. Treating the UF flows as pesos gives a delta NII of 1,356 for the short-UF variant against the Annex 1 value of 11,067, and a delta EVE of 64,213 against 79,643, so the measure of the regulator is not the one a single-currency calculation gives. On the changes that the committed document records, the breach test flags the fitted average maturity of the wholesale deposits at 2022-06-30 and 2022-12-31, while the requested core share (0.6) and maturity (5 years) are green and the same fitted series with no breach is green: the breach flag decides. The document records only those three parameters; run on every fitted scalar of the model, the test also flags the margin volatility and the deposit beta for rising rates.

- **The UF claim, the separation.** Neither Basel table has the peso or the UF (21 currencies each), and the engine refuses a size for them. The Chilean table gives the peso 300/400/200 bp and the UR 200/250/150 bp for the EVE and 300 bp for the NII, and each currency is on its own curve: 0.04499 in the first band for the peso and 0.01205 for the UR. The floor has to be the currency's own: a 200 bp fall takes the UR curve to -0.00795 in the first band, and the peso floor of 0 would bind in nine bands while the UR floor of -2 percent binds in none.
- **The UF claim, what a single currency loses.** The delta NII is 1,356, 1,648 and 1,254 against 11,067, 11,213 and 11,016 for the short-UF, long-UF and offset-on-off variants: 6.8 to 8.8 times smaller. The delta EVE is 64,213, 84,710 and 59,422 against 79,643, 93,816 and 81,173: 10 to 27 percent smaller. No ratio to Tier 1 crosses a line in either treatment in these four variants. These ranges are those of the committed book, one seed. Over 30 synthetic books per variant the UF paper finds ratios of the mean delta NII of 4.9, 5.5 and 5.7 (the mean of the per-book ratios is 5.8, 6.9 and 7.5, heavy-tailed) and a delta EVE 5.7, 15.4 and 24.2 percent smaller, with the charge supplying 52 to 53 percent of the delta NII. Three of its statements were refuted: the pair rule does not have the largest Shapley value for the delta EVE (the own curve does); the Annex 1 delta NII is not always above the positive-part aggregation (it falls below it in the CLP steepener variant, in 30 of 30 books, and that statement was written after a value that already contradicted it); and a binding UF floor raised the delta NII in 208 of 960 comparisons. The statement that Annex 1 flags most books where the single-currency treatment flags none held at 110 percent of Tier 1 only (17 of 20 books against 0), through the delta EVE test.
- **The UF claim, the offset and the charge.** The offset lowers the delta NII from 14,597 to 11,067 in the short-UF variant and the delta EVE from 94,721 to 81,173 in the offset-on-off variant. The charge is 0.5586, 0.5513 and 0.5612 of the delta NII in the three variants with a net UF position of 30 percent of Tier 1, and 0.1560 in the one with 5 percent.
- **The UF claim, the limit.** The floors never bind: the closest post-shock rate is 20.48 bp from its floor in the four variants, so their effect on a number is not demonstrated here. There is no second implementation of Annex 1 to reconcile with; the dossier's search found none, which stays unverified beyond it. The bank is synthetic, and the claim is regional.
- **The breach-test claim, the discrimination.** On the five constructed histories the test gives the verdict the rule says: green for a parameter held fixed through a breach, green for a change with no breach, green for a change before the breach, red for a change at the breach (flagging 2022-12-31), and not evaluated for a single date.
- **The breach-test claim, the committed bank.** The requested core share (0.6) and the requested core maturity (5.0 years) do not change and the test is green; the fitted average maturity moves, 4.7595, 4.2049 and 5.0 years, and the test is red at 2022-06-30 and 2022-12-31. With the breach flags set to false the same series is green. The model document records the changes of these three parameters only; the margin volatility and the deposit beta for rising rates are also re-estimated at each date, and the test run on them flags both, each by construction.
- **The breach-test claim, the limit.** The outlier test is breached at the three dates, so the flag never varies, and the parameter that moves is a fitted one that is re-estimated at each date, so it differs by construction. The red is therefore not a finding that anyone tuned the maturity to relieve a breach; it shows that the test operates on those parameters and would do so on a parameter that a manager changed. What a validator does with a red: ask who moved the parameter, when and with what approval; compare the new value with the data that support it (notebook 05 shows that the fitted weights are not stable across windows); and read the EVE with and without behaviour at the same date. The bank's behaviour is synthetic.
- **What the notebook does not show.** A reconciliation of the Chilean figures with a second implementation, the effect of a floor on a number, and a case where a manager changed a stated assumption at a breach: the committed data has none.

## Summary

- The UF claim: the Chilean Annex treats the UF as a currency of its own; the engine implements its sizes, floor, offset and charge, and the committed variants recompute from it. A single-currency treatment of the same flows would give a very different delta NII, because the charge and the pairing are what make the NII measure.
- The UF claim has limits: the ranges of the committed book are one seed (over 30 books the single-currency ratio of the delta NII is 4.9, 5.5 and 5.7), three statements of its paper were refuted, no second implementation to reconcile with, floors that never bind on these curves, a synthetic bank, and a regional claim.
- The breach-test claim: the breach test is red when a parameter changes at a date with a breach and green otherwise, and the verdict is decided by the breach flag.
- The breach-test claim has limits: in the committed bank every date is a breach and the parameter that moves is a fitted one, so the red shows that the test operates, not that the bank did what the review of the failed bank found. Its error rates are measured in the behavioural-assumption paper on simulated banks, not in this notebook: on a stated parameter it flags 100.0 % of histories with a change at a breach date and 17.0 to 66.3 % of histories with unrelated changes (the 5 % size prediction is refuted); on a re-estimated parameter it flags 82.2 % of banks with nothing planted and ranks a 2-year lengthening at an ROC area of 0.579, and only a tolerance calibrated at 5 % (3.9 % false flags, area 0.937, 18.0 % of 2-year and 99.5 % of 4-year lengthenings detected) repairs the size, which then drifts to 5.5 % at another breach prevalence.
- Both claims are stated in the SDD only as far as this evidence goes.
