## Método

Los tres primeros peldaños modelan el tiempo desde la fecha de reporte hasta el incumplimiento, con el prepago como censura, y dan la PD acumulada $F(t)$ que el notebook 01 convierte en una PCE. Difieren en cómo dejan que la tasa de riesgo dependa del préstamo y de la macroeconomía.

**L0, Kaplan-Meier.** Una curva agrupada por edad del préstamo. El estimador producto-límite de la supervivencia es $\hat S(t) = \prod_{t_i \le t}\bigl(1 - d_i/n_i\bigr)$, con $d_i$ los incumplimientos y $n_i$ los préstamos en riesgo a la edad $t_i$. No tiene covariables ni macroeconomía, así que su orden es solo el orden por edad.

**L1, tasa de riesgo en tiempo discreto con covariables macro.** Una regresión logística sobre registros préstamo-período (Shumway, 2001; Botha y Verster, 2025):

$$\mathrm{logit}\,h_i(t) = \alpha(a_{it}) + \boldsymbol\beta^\top \mathbf{x}_i + \boldsymbol\gamma^\top \mathbf{m}_{t-\ell}$$

con $\alpha$ una función escalonada de la edad del préstamo (bandas de edad), $\mathbf{x}_i$ las covariables del préstamo (tamaño, plazo, sector, rotativo) y $\mathbf{m}_{t-\ell}$ las variables macro rezagadas $\ell = 1$ trimestre, estandarizadas. Es el peldaño que lleva los escenarios.

**L2, riesgos proporcionales de Cox** (Cox, 1972): $h_i(t) = h_0(t)\exp(\boldsymbol\beta^\top \mathbf{x}_i)$, con la edad del préstamo como escala de tiempo y empates de Breslow. La implementación de `statsmodels` que usa el flujo de cálculo no tiene covariables que varíen en el tiempo, así que este peldaño **no tiene macroeconomía** y no puede llevar un escenario.

Para una tasa de riesgo por período pequeño $h$, $\mathrm{logit}\,h = \log\bigl(h/(1-h)\bigr) \approx \log h$, así que los coeficientes de L1 y L2 en las covariables comunes deberían coincidir. La macroeconomía entra a la estructura de plazos solo en L1.

**La validación de un peldaño** es la prueba retrospectiva de vida completa del notebook 05 y la comparación de peldaños: la concordancia de Harrell (la probabilidad de que, de dos préstamos, el de menor supervivencia tenga el mayor riesgo previsto, con empates como un medio), el puntaje de Brier integrado de la predicción de supervivencia y la diferencia de calibración a 12 meses entre la PD acumulada prevista y la observada. La mirada de un factor (BCBS 2005; Gordy 2003) vincula la macroeconomía con la PD mediante un factor sistemático $z$ que un satélite regresa sobre variables macro; los pesos de los pronósticos de corto plazo decaen hacia valores de largo plazo (carta del BCE del 1 de abril de 2020).

Dos identidades sostienen cada paso siguiente, y una tercera define la rama neuronal del paso 8. Una tasa de riesgo $h(t)$ por período da la supervivencia, la PD marginal y la PD acumulada que `from_hazard` y `from_cumulative` del motor se convierten entre sí,

$$S(t) = \prod_{u \le t}\bigl(1 - h(u)\bigr), \qquad q(t) = S(t-1)\,h(t), \qquad F(t) = \sum_{u \le t} q(u) = 1 - S(t)$$

y cuando el prepago compite con el incumplimiento, con tasas $h_D$ y $h_P$, la incidencia de incumplimiento no es $1 - S$ sino

$$F_D(t) = \sum_{u \le t} S_{\mathrm{all}}(u-1)\,h_D(u), \qquad S_{\mathrm{all}}(t) = \prod_{u \le t}\bigl(1 - h_D(u) - h_P(u)\bigr)$$

que es lo que `competing_risks` devuelve como `cumulative_default`, siempre bajo la probabilidad de incumplir o prepagar $1 - S_{\mathrm{all}}$. La rama PC-Hazard toma una tasa constante dentro de cada intervalo de edad $k$ de ancho $w$, de modo que la log-verosimilitud de los incumplimientos $d_{sk}$ y de la exposición $e_{sk}$ del estrato $s$ es de Poisson,

$$\ell = \sum_{s,k} \Bigl[ d_{sk} \log \lambda_{sk} - \lambda_{sk}\, e_{sk} \Bigr], \qquad \lambda_{sk} = \mathrm{red}_k(\mathbf{x}_s)$$

con $\lambda_{sk}$ la tasa que la red asigna a las covariables del estrato $s$ en el intervalo $k$.

## Comparación

La investigación (dossiers 02 y 04) encontró los métodos siguientes para la estructura de plazos del incumplimiento bajo IFRS 9. La última columna dice qué hace el producto con cada uno.

| Kaplan-Meier y curvas de cosecha | el incumplimiento acumulado agrupado y por cohorte, con censura | Botha y Verster 2025 | peldaño L0, leído aquí |

| Riesgos proporcionales de Cox, con o sin covariables que varían en el tiempo | efectos de covariables sin supuesto de línea base; macro que varía en el tiempo | Cox 1972 | peldaño L2 (sin covariables que varíen en el tiempo), leído aquí |

### Paso 1: los coeficientes del logit y del modelo de Cox en las covariables comunes

Ambos peldaños tienen las covariables del préstamo: tres tramos de tamaño, tres de plazo, once grupos de sector y la marca de rotativo. La celda alinea los 18 coeficientes que comparten el logit de tiempo discreto y el modelo de Cox y mide cuán lejos están, en términos absolutos, como correlación y en errores estándar del logit. La figura grafica el coeficiente de Cox de cada covariable contra su coeficiente del logit; si los dos métodos coincidieran exactamente, todos los puntos quedarían sobre la diagonal.

````python
# Alinear los coeficientes que comparten el logit de tiempo discreto y el modelo de Cox y compararlos
dth = {c["name"]: c for c in ladder["L1-dth"]["coefficients"]}
cox = {c["name"]: c for c in ladder["L2-cox"]["coefficients"]}
common = [name for name in cox if name in dth]
logit_values = np.array([dth[n]["estimate"] for n in common])
cox_values = np.array([cox[n]["estimate"] for n in common])
std_errors = np.array([dth[n]["std_error"] for n in common])
print("covariables comunes:", len(common), "| solo en el peldaño de tiempo discreto:", len(dth) - len(common))
print("mayor diferencia absoluta:", round(float(np.abs(logit_values - cox_values).max()), 4))
print("correlación:", round(float(np.corrcoef(logit_values, cox_values)[0, 1]), 5))
print("mayor diferencia en errores estándar del logit:",
      round(float(np.max(np.abs(logit_values - cox_values) / std_errors)), 2))

fig, ax = plt.subplots(figsize=(5.5, 4.5))
ax.scatter(logit_values, cox_values, color="tab:blue")
span = [logit_values.min(), logit_values.max()]
ax.plot(span, span, color="grey", linestyle=":", label="coeficientes iguales")
ax.set_xlabel("coeficiente del logit")
ax.set_ylabel("coeficiente de Cox")
ax.set_title("Covariables comunes: logit frente a Cox")
ax.legend(fontsize=8)
fig.tight_layout()
plt.show()
````

Los 18 coeficientes compartidos quedan casi sobre la diagonal: su correlación es 0,9998 y la mayor diferencia absoluta es 0,0389, que son 3,2 errores estándar del logit. La coincidencia es esperable. Para una tasa de riesgo $h$ por período pequeña el logit $\log\bigl(h/(1-h)\bigr)$ es cercano a $\log h$, que es la escala del modelo de Cox, de modo que los dos métodos estiman casi los mismos efectos de las covariables. Los puntos que se alejan de la diagonal pertenecen a las covariables con mayor tasa de riesgo por período, donde la aproximación es peor. La elección entre los dos no trata entonces de las covariables del préstamo sino de lo que cada uno puede llevar: solo el logit puede tomar la macroeconomía.

### Paso 2: los coeficientes macro y sus signos

Solo el peldaño logit tiene covariables macro, estandarizadas con una media y una desviación estándar. Un validador verifica cada signo frente a la expectativa económica antes de leer nada más: una mayor tasa de desempleo y una mayor tasa preferencial deberían subir la tasa de riesgo de incumplimiento, y un mayor crecimiento del PIB real debería bajarla. La expectativa es un insumo declarado de la lección. La celda imprime cada estimación con su error estándar, su estadístico z y su razón de riesgo por desviación estándar, y si el signo es el esperado.

````python
# Leer los coeficientes macro del peldaño logit y verificar sus signos frente a la expectativa declarada
EXPECTED_SIGN = {"unemployment_rate": +1, "real_gdp_growth": -1, "prime_rate": +1}  # lesson input
for name, sign in EXPECTED_SIGN.items():
    c = dth[name]
    z = c["estimate"] / c["std_error"]
    matches = "sí" if np.sign(c["estimate"]) == sign else "no"
    print(f"{MACRO[name]:26s} | estimación {c['estimate']:+.4f} | error estándar {c['std_error']:.4f}")
    print(f"{'':26s} | z {z:+.2f} | razón de riesgo {np.exp(c['estimate']):.3f} | signo esperado: {matches}")
````

El crecimiento del PIB real baja la tasa de riesgo como se espera, con una estimación de -0,1920, un z de -49,11 y una razón de riesgo de 0,825 por desviación estándar, y la tasa preferencial la sube, +0,1138 con una razón de 1,120. La tasa de desempleo tiene el signo equivocado: su estimación es -0,0534, con un z de -5,99, de modo que un mayor desempleo baja la tasa de riesgo y el coeficiente es estadísticamente distinto de cero. Es la sensibilidad que el hallazgo versionado `F-SCENARIO-ORDER` culpa de que el escenario severamente adverso dé una PCE menor que el base en el notebook 03. Un validador rechaza un coeficiente macro de signo equivocado aunque sea significativo, porque un efecto significativo que se mueve contra la economía puede señalar colinealidad entre las variables macro en una historia corta, lo que este notebook no prueba.

### Paso 3: los tres peldaños sobre los mismos datos, puntuados por el motor

Las filas de la prueba retrospectiva guardan, para cada peldaño y cohorte de originación, la brecha con signo entre la PD acumulada observada y la predicha en cada horizonte, definida como observada menos predicha. La celda recupera la curva predicha de cada peldaño desde la observada y la brecha, y entrega las tres curvas predichas a `lifetime_pd_backtest` del motor, que devuelve la mayor brecha, la mayor brecha relativa y una luz. Lo hace para la cohorte agrupada `all` y para la cosecha 2007, los préstamos originados justo antes de la crisis.

````python
# Puntuar cada peldaño en dos cohortes con la prueba de vida completa del motor
rows = case["outputs"]["lifetime_backtest"]["rows"]
lifetime_rows = [t for t in case["tests"] if t["test_id"] == "ifrs9.lifetime_pd_backtest"]
tests = {(t["model_id"], t["segment"]): t for t in lifetime_rows}
curves = {}
for cohort in ("all", "2007"):
    cohort_rows = [r for r in rows if r["origination_cohort"] == cohort]
    months = np.array([r["months"] for r in cohort_rows])
    obs = np.array([r["observed_cumulative_pd"] for r in cohort_rows])
    curves[cohort] = (months, obs, {})
    for rung in RUNGS:
        row = tests[(rung, cohort)]
        predicted = obs - np.array(row["extras"]["signed_gap"])
        curves[cohort][2][rung] = predicted
        at_risk = np.full(months.size, row["n"], dtype=float)
        events = np.full(months.size, row["n_events"], dtype=float)
        score = ifrs9.lifetime_pd_backtest(months, predicted, obs, at_risk, events, model_id=rung, segment=cohort)
        relative = max(score.extras["relative_gap"])
        print("cohorte", "agrupada" if cohort == "all" else cohort, "|", RUNG_NAME[rung])
        print(f"   mayor brecha {score.metric:.4f} | mayor brecha relativa {relative:.4f}")
        print("   luz sin la correlación:", LIGHT[score.light.value])
````

La prueba de vida completa del motor, llamada sin la correlación de activos, lee la brecha relativa y no la banda de Vasicek del notebook 05, de modo que estas luces son más estrictas que las versionadas. En la cohorte agrupada el logit de tiempo discreto tiene la menor de las mayores brechas, 0,0167 con una brecha relativa de 0,3228 y una luz ámbar, mientras que los peldaños de Kaplan-Meier y de Cox tienen 0,0427 y brechas relativas de 0,9147 y 1,0791, rojas. En la cosecha 2007 las tres brechas son 0,1240, 0,0859 y 0,1163 y las tres son rojas, con brechas relativas sobre 1. El orden de los peldaños no cambia entre las cohortes, con el logit primero, pero incluso el mejor peldaño está lejos del resultado de la cosecha que enfrentó la crisis.

La curva observada de la cosecha 2007 sube a cerca de 0,20 a los 36 meses, y las tres predicciones quedan bajo ella en todos los horizontes: el peldaño de Kaplan-Meier llega a 0,0766 a los 36 meses, el de Cox a 0,0863 y el logit a 0,1267. El logit es el más cercano porque sus covariables macro suben la tasa de riesgo en los trimestres de crisis, y aun así cubre poco más de la mitad del incumplimiento realizado. Las curvas de Kaplan-Meier y de Cox son ciegas a la macroeconomía y a la cosecha, de modo que describen un préstamo promedio de la cartera y no ven los préstamos otorgados en la cima del ciclo. Las covariables macro acortan la brecha y no la cierran.

### Paso 4: lo que cuesta en PCE la elección del peldaño

Una PD de vida completa importa a través de la PCE. La celda valora una cuenta declarada, una exposición de 100.000 USD al 8% anual con la pérdida dado el incumplimiento del segmento de préstamos de más de 20 años, a un horizonte de 36 meses con cuatro curvas: la observada y la curva predicha de cada peldaño para la cohorte agrupada. Cada curva pasa por `from_cumulative` para obtener su PD marginal, y `ecl_scenario` la convierte en dinero. La exposición y la tasa son insumos de la lección, y la figura se deja a los números.

````python
# Valorar la cuenta declarada a 36 meses con la curva predicha de cada peldaño y con la observada
months_all, obs_all, predicted_all = curves["all"]
LGD = {s["segment_id"]: s["lgd"] for s in fit["lgd"]["segments"]}["term-gt-240"]
EAD, EIR = 100_000.0, np.array([0.08])  # stated inputs of the lesson
priced = {}
for label, curve in (("observada", obs_all), *[(RUNG_NAME[r], predicted_all[r]) for r in RUNGS]):
    structure = tsm.from_cumulative(curve, months_all)
    marginal = structure.marginal[None, :]
    priced[label] = float(ecl_engine.ecl_scenario(marginal, LGD, EAD, EIR, months_all, 36)[0])
reference = priced["observada"]
for label, value in priced.items():
    print(f"{label:24s} | PCE a 36 meses, USD {value:9.2f} | sobre la observada {value / reference:6.3f}")
````

La elección del peldaño es una elección sobre dinero. Valorada con el incumplimiento acumulado observado de la cohorte agrupada, que es lo que ocurrió y no un pronóstico, la cuenta declarada habría costado 5.175,62 USD de PCE a 36 meses. La curva del logit da 4.548,98 USD, 0,879 de eso, y las curvas de Kaplan-Meier y de Cox dan 3.114,33 y 3.095,79 USD, cerca de 0,60 de eso. Un banco que hubiera provisionado con cualquiera de los peldaños ciegos a la macroeconomía habría mantenido 40% menos que el resultado, e incluso el logit queda 12% corto. La cohorte agrupada incluye la crisis, que es la razón de que todo peldaño subpredice, y el orden es el mismo que en la prueba retrospectiva.

### Paso 5: el prepago como censura frente a una salida competidora

C06 trata el prepago como censura (hallazgo `F-PREPAYMENT`): la tasa de riesgo de incumplimiento por causa es toda la PD de vida completa. La investigación recomienda una vista de riesgos competidores, en que un préstamo que se paga ya no puede incumplir (Fine y Gray, 1999; Botha y Verster, 2025). La celda toma la curva de Kaplan-Meier, recupera su tasa de riesgo con `from_cumulative`, agrega una tasa de riesgo de prepago declarada de 0, 0,5%, 1% y 2% por trimestre con `competing_risks`, y valora la PCE de vida completa de la cuenta declarada del paso 4 bajo cada incidencia de incumplimiento. La figura dibuja las curvas de incidencia.

````python
# Agregar una tasa de riesgo de prepago declarada a la curva de Kaplan-Meier y comparar incidencia y PCE
km = ladder["L0-km"]["baseline"]
km_months = np.array(km["months"])
single = tsm.from_cumulative(np.array(km["cumulative_pd"]), km_months)
at_36 = int(np.flatnonzero(km_months == 36)[0])
incidence, ecl_life = {}, {}
for prepay in (0.0, 0.005, 0.01, 0.02):
    competing = tsm.competing_risks(single.hazard, np.full(single.hazard.shape, prepay), km_months)
    incidence[prepay] = competing.cumulative_default
    marginal = competing.marginal_default[None, :]
    ecl_life[prepay] = float(ecl_engine.ecl_scenario(marginal, LGD, EAD, EIR, km_months, int(km_months[-1]))[0])
    print("tasa de riesgo de prepago por trimestre", prepay)
    print("   incidencia de incumplimiento a 36 meses:", round(float(incidence[prepay][at_36]), 5))
    print("   PCE de vida completa, USD:", round(ecl_life[prepay], 2))
    print("   censura sobre competidora:", round(ecl_life[0.0] / ecl_life[prepay], 4))

fig, ax = plt.subplots(figsize=(7, 4))
for prepay, curve in incidence.items():
    ax.plot(km_months, curve, label=f"{prepay}")
ax.set_xlabel("meses desde la fecha de reporte")
ax.set_ylabel("incidencia acumulada de incumplimiento")
ax.set_title("Incidencia de incumplimiento bajo censura y bajo una salida competidora")
ax.legend(title="tasa de riesgo de prepago por trimestre", fontsize=8)
fig.tight_layout()
plt.show()
````

Agregar una salida por prepago baja la incidencia de incumplimiento, y la vista de censura la sobrestima. Con una tasa de riesgo de prepago declarada de 0,5%, 1% y 2% por trimestre el incumplimiento acumulado a 36 meses baja de 0,04394 a 0,04222, 0,04056 y 0,03744, y la PCE de vida completa de la cuenta declarada baja de 5.915,44 USD a 5.453,55, 5.034,42 y 4.306,99. La razón entre la PCE con censura y la competidora es 1,0847, 1,175 y 1,3734. Las curvas se separan más cuanto más largo es el horizonte, de modo que la elección importa más para los préstamos largos de la cartera. Las tasas de riesgo de prepago son insumos de la lección y no estimaciones, así que el resultado es una sensibilidad: el lado prudente de la vista de censura tiene un precio que crece con el prepago.

### Paso 5b: dónde cambia o falla el método, el riesgo de prepago y el horizonte

El paso 5 valoró la vista de censura con cuatro riesgos de prepago declarados y un horizonte. El método tiene dos parámetros que deciden cuán errada es la vista de censura, el riesgo de prepago y el horizonte en que se mide la PCE, y un validador necesita toda la superficie y no cuatro puntos. La celda repite la comparación del paso 5 en una malla de seis riesgos de 0 a 5% por trimestre y tres horizontes, 12, 24 y 36 meses: para cada par valora la cuenta declarada con `ecl_scenario` del motor desde la incidencia de incumplimiento de Kaplan-Meier (censura) y desde la incidencia que devuelve `competing_risks` cuando el prepago es una salida competidora, y divide la primera por la segunda. Hace lo mismo con el incumplimiento acumulado. El panel izquierdo dibuja la razón de la PCE contra el riesgo, una línea por horizonte, y el derecho la razón de incidencias. La celda termina ofreciendo un riesgo de prepago de 1 por período, que junto al riesgo de incumplimiento no deja sobrevivientes, para ver el rechazo del motor.

````python
# Barrer el riesgo de prepago y el horizonte: la sobrestimación de la incidencia y de la PCE bajo la vista de censura
prepay_grid = np.array([0.0, 0.005, 0.01, 0.02, 0.03, 0.05])
cuts = [m for m in (12, 24, 36) if m in km_months]
ecl_ratio, inc_ratio = np.zeros((len(cuts), prepay_grid.size)), np.zeros((len(cuts), prepay_grid.size))
for j, prepay in enumerate(prepay_grid):
    rival = tsm.competing_risks(single.hazard, np.full(single.hazard.shape, prepay), km_months)
    for i, cut in enumerate(cuts):
        k = int(np.flatnonzero(km_months == cut)[0])
        with_exit = ecl_engine.ecl_scenario(rival.marginal_default[None, :], LGD, EAD, EIR, km_months, cut)
        censored = ecl_engine.ecl_scenario(single.marginal[None, :], LGD, EAD, EIR, km_months, cut)
        ecl_ratio[i, j] = float(censored[0]) / float(with_exit[0])
        inc_ratio[i, j] = single.cumulative[k] / rival.cumulative_default[k]
print("horizontes en meses:", cuts, "tasa de riesgo de prepago por trimestre:", prepay_grid.tolist())
print("PCE con censura sobre PCE competidora, por horizonte (filas) y tasa:", np.round(ecl_ratio, 4).tolist())
print("incidencia con censura sobre incidencia competidora, misma disposición:", np.round(inc_ratio, 4).tolist())
try:
    tsm.competing_risks(single.hazard, np.full(single.hazard.shape, 1.0), km_months)
except ValueError as error:
    print("una tasa de incumplimiento más prepago de 1 o más se rechaza con", type(error).__name__)

fig, (left, right) = plt.subplots(1, 2, figsize=(10, 3.8))
for i, cut in enumerate(cuts):
    left.plot(100 * prepay_grid, ecl_ratio[i], marker="o", label=f"{cut}")
    right.plot(100 * prepay_grid, inc_ratio[i], marker="o", label=f"{cut}")
left.set_ylabel("PCE con censura sobre PCE competidora")
right.set_ylabel("incidencia con censura sobre competidora")
for ax in (left, right):
    ax.set_xlabel("riesgo de prepago por trimestre (%)")
    ax.legend(title="months", fontsize=8)
fig.suptitle("Lo que cuesta tratar el prepago como censura")
fig.tight_layout()
plt.show()
````

La sobrestimación crece con ambos parámetros, y casi en proporción a su producto. Con un riesgo de prepago de 0,5% por trimestre la vista de censura sobrestima la PCE en 1,4% a 12 meses, 2,8% a 24 y 4,0% a 36; con 2% por trimestre en 5,8%, 11,9% y 17,1%; y con 5% por trimestre en 15,5%, 32,7% y 48,7%. La razón de las incidencias es casi el mismo número (1,4946 frente a 1,4867 con 5% y 36 meses), de modo que el sesgo de la PCE es el sesgo de la incidencia de incumplimiento y no un efecto del descuento. Una lectura por regímenes: si los préstamos son cortos y se prepagan despacio, el tratamiento por censura de C06 es un error pequeño, bajo 5% hasta cerca de 1,7% por trimestre a 12 meses, 0,9% a 24 y 0,6% a 36 (leído por interpolación lineal entre los puntos de la malla), y puede mantenerse con el hallazgo `F-PREPAYMENT` declarado; para los préstamos largos de una cartera de tipo hipotecario con prepago de 3% o más por trimestre sobrestima la PCE de vida completa en un cuarto o más a 36 meses, y un validador debe exigir la vista de riesgos competidores. Nada de este barrido se estima desde los préstamos, porque los riesgos son insumos de la lección, así que la figura dice cómo escala el error y no cuán grande es en la cartera. El motor rechaza un par de riesgos que suma 1 o más, ya que ningún préstamo sobreviviría el período: ese es el borde superior del dominio del método.

### Paso 6: el satélite y la historia del factor

La conversión de un factor da el factor sistemático $z$ de cada trimestre pasado desde la tasa de incumplimiento realizada de la cartera (`fit.pit.z_hat`), y un satélite regresa ese factor sobre las variables macro de modo que un escenario se pueda convertir en una trayectoria de $z$. La celda imprime los coeficientes del satélite con el signo que debería tener el factor, y convierte la historia de 63 trimestres del factor de vuelta en la tasa de incumplimiento que implica con `ttc_to_pit`, usando la PD a través del ciclo de la cartera y su correlación. La figura dibuja el factor y la tasa de incumplimiento que implica.

````python
# Leer el satélite y la historia del factor, y convertir la historia en las tasas de incumplimiento que implica
pit = fit["pit"]
satellite = pit["satellite"]
params = case["outputs"]["non_linearity"]["parameters"]
EXPECTED_FACTOR_SIGN = {"unemployment_rate": -1, "real_gdp_growth": +1, "prime_rate": -1}  # stated expectation
print("intercepto del satélite y término autorregresivo:", round(satellite["intercept"], 4), satellite["ar1"])
for c in satellite["coefficients"]:
    sign_ok = "sí" if np.sign(c["value"]) == EXPECTED_FACTOR_SIGN[c["name"]] else "no"
    print(f"{MACRO[c['name']]:26s} | coeficiente {c['value']:+.4f} | signo esperado para el factor: {sign_ok}")
history = np.array([p["z"] for p in pit["z_hat"]])
years = [int(p["period"][:4]) + (int(p["period"][5:7]) - 1) / 12 for p in pit["z_hat"]]
implied = tsm.ttc_to_pit(np.full(history.size, params["pd_ttc"]), pit["rho"], history)
print("trimestres de historia del factor:", history.size)
print("media y desviación estándar del factor:")
print(round(float(history.mean()), 3), round(float(history.std()), 3))
print("último valor del factor:", round(float(history[-1]), 3))
print("tasa de incumplimiento implicada, menor y mayor:")
print(round(float(implied.min()), 4), round(float(implied.max()), 4))

fig, (left, right) = plt.subplots(1, 2, figsize=(10, 3.6))
left.plot(years, history, color="tab:blue")
left.axhline(0.0, color="grey", linewidth=0.8)
left.set_xlabel("año")
left.set_ylabel("factor sistemático z")
left.set_title("Historia del factor sistemático")
right.plot(years, implied, color="tab:red")
right.set_xlabel("año")
right.set_ylabel("tasa de incumplimiento implicada")
right.set_title("Tasa de incumplimiento implicada por el factor")
fig.tight_layout()
plt.show()
````

Convertido de vuelta, el factor implica una tasa de incumplimiento entre 0,30% y 6,71%, un rango de más de veinte veces, que es el ciclo que debe llevar la estructura de plazos. Hay un desacuerdo entre las dos capas: en el modelo de tasa de riesgo la tasa de desempleo baja la PD, mientras que en el satélite baja el factor, lo que sube la PD. El archivo no dice por qué, y un validador lo lee como un hallazgo del modelo.

### Paso 8: un peldaño neuronal, PC-Hazard

Los tres peldaños anteriores comparten una restricción que la investigación nombra como su punto débil: el peldaño de Cox supone
riesgos proporcionales, así que el riesgo de cada préstamo es el mismo perfil por edad escalado hacia arriba o hacia abajo, y el
logit de tiempo discreto también suma las covariables macro a un solo perfil. Kvamme y Borgan (2019) la relajan con PC-Hazard: el
riesgo es constante por tramos sobre una grilla de intervalos de edad, y una red de alimentación hacia adelante lleva las
covariables de un préstamo a la tasa de cada intervalo, de modo que cada estrato tiene su propio perfil por edad. Como la
verosimilitud es la de Poisson de los incumplimientos y la exposición de cada celda (estrato, intervalo), el entrenamiento lee solo
esas dos tablas. El producto la agrega como peldaño L3 (`pipeline.cases.c06.neural`, dos capas ocultas de 32 unidades, detención
temprana sobre un quinto reservado).

Los datos de C06 son 1,7 millones de préstamos SBA leídos de la raíz de datos, así que este notebook sortea sus propios préstamos
donde la respuesta se conoce: cuatro estratos con perfiles que se cruzan (plano, un pico temprano, un riesgo creciente y una joroba
a los dos años), prepago de 0,8 % mensual como salida competidora y un seguimiento de dos a cinco años. La celda siguiente ajusta la
red del producto sobre cuatro quintos de ellos y se detiene sobre el quinto reservado.

````python
# Sortear 24.000 préstamos en cuatro estratos de riesgos que se cruzan con la edad, con prepago y 2 a 5 años
from pipeline.cases.c06 import neural as pch

N_INT, WIDTH = 20, 3.0
age = (np.arange(N_INT) + 0.5) * WIDTH
truth = np.vstack([np.full(N_INT, 0.0020), 0.0050 * np.exp(-age / 15.0), 0.0004 + 0.00008 * age,
                   0.0005 + 0.0035 * np.exp(-(((age - 24.0) / 10.0) ** 2))])
rng = np.random.default_rng(20261030)
stratum = np.repeat(np.arange(4), 6000)
knots = np.concatenate([np.zeros((4, 1)), np.cumsum(truth * WIDTH, axis=1)], axis=1)
e = rng.exponential(size=stratum.size)
default_t = np.full(stratum.size, np.inf)
for s in range(4):
    m = stratum == s
    j = np.searchsorted(knots[s], e[m], side="right") - 1
    inside = j < N_INT
    t = np.full(m.sum(), np.inf)
    t[inside] = WIDTH * j[inside] + (e[m][inside] - knots[s, j[inside]]) / truth[s, j[inside]]
    default_t[m] = t
prepay_t = rng.exponential(1 / 0.008, stratum.size)
follow = rng.uniform(24.0, 60.0, stratum.size)
time_obs = np.minimum.reduce([default_t, prepay_t, follow])
event = default_t <= np.minimum(prepay_t, follow)
cause = np.where(event, 1, np.where(prepay_t <= follow, 2, 0))
defaults_label = "incumplimientos"
print(f"{defaults_label}: {int(event.sum())} / {stratum.size}")
````

De los 24.000 préstamos 1.535 incumplen dentro de su seguimiento; el resto prepaga o sigue vivo cuando la observación
termina. La red nunca ve un préstamo: la celda siguiente convierte cada parte de la muestra en las dos tablas de la verosimilitud,
los incumplimientos y los meses de exposición de cada estrato en cada intervalo de tres meses de edad, y entrena sobre ellas.

````python
# Armar las dos tablas de cada parte (incumplimientos y exposición por estrato e intervalo) y entrenar la red
grid = {"n_strata": 4, "n_intervals": N_INT, "width": WIDTH}
val = rng.random(stratum.size) < 0.2
ev_tr, ex_tr = pch.interval_statistics(stratum[~val], time_obs[~val], event[~val], **grid)
ev_va, ex_va = pch.interval_statistics(stratum[val], time_obs[val], event[val], **grid)
net = pch.fit_network(np.eye(4), ev_tr, ex_tr, val_events=ev_va, val_exposure=ex_va, seed=pch.SEED)
epoch_label = "mejor época / épocas corridas"
params_label = "pesos"
print(f"{epoch_label}: {net.best_epoch} / {net.epochs} | {params_label}: {net.n_parameters}")
````

La red conserva la época con la menor pérdida en la parte reservada, la 11.ª de 71 corridas, con 1.876 pesos para 80 tasas:
la detención temprana es lo que impide que un modelo de este tamaño ajuste el ruido de 1.500 incumplimientos. El modelo de
comparación es el de riesgos proporcionales ajustado sobre las mismas tablas: una base por intervalo por un factor por estrato,
mediante las actualizaciones alternadas de la verosimilitud de Poisson. La figura dibuja el riesgo verdadero, el de la red y el
proporcional de cada estrato.

````python
# Ajustar riesgos proporcionales sobre las mismas tablas (una base, un factor por estrato) y dibujar los tres riesgos
ev_all, ex_all = ev_tr + ev_va, ex_tr + ex_va
factor = np.ones(4)
for _ in range(200):
    base = ev_all.sum(0) / np.maximum((ex_all * factor[:, None]).sum(0), 1e-12)
    factor = ev_all.sum(1) / np.maximum((ex_all * base[None, :]).sum(1), 1e-12)
    factor = factor / factor[0]
ph = factor[:, None] * base[None, :]
fig, axes = plt.subplots(1, 4, figsize=(12, 3.2), sharey=True)
for s, ax in enumerate(axes):
    ax.plot(age, 1000 * truth[s], color="k", label="verdad")
    ax.plot(age, 1000 * net.rates[s], "--", color="tab:blue", label="PC-Hazard")
    ax.plot(age, 1000 * ph[s], ":", color="tab:red", label="riesgos proporcionales")
    ax.set_title(f"{"estrato"} {s}")
    ax.set_xlabel("antigüedad del préstamo (meses)")
axes[0].set_ylabel("riesgo de incumplimiento por mes (por 1.000)")
axes[0].legend(fontsize=8)
fig.tight_layout()
plt.show()
````

El ajuste proporcional solo puede subir y bajar una base, así que dibuja cuatro curvas paralelas: aplana el pico temprano
del estrato 1 y la joroba del estrato 3 y le da al estrato 2, que crece, un comienzo decreciente. PC-Hazard sigue cada perfil, con
el ruido de unos cientos de incumplimientos por estrato alrededor de la línea verdadera, y sus últimos intervalos se abren donde
quedan pocos préstamos observados. Lo que las formas significan para una provisión es la PD acumulada de los primeros años de un
préstamo, que la celda siguiente imprime por estrato junto a la incidencia de Aalen-Johansen que el producto usa como lado observado
de su calibración.

````python
# PD acumulada a 12, 24 y 36 meses de edad por estrato: la verdad, ambos ajustes y Aalen-Johansen con prepago
horizons = np.array([12.0, 24.0, 36.0])
pds = {}
for name, rates in (("verdad", truth), ("PC-Hazard", net.rates), ("riesgos proporcionales", ph)):
    pds[name] = 100 * pch.conditional_cumulative_pd(rates, np.arange(4), np.zeros(4), horizons, width=WIDTH)
aj = [100 * pch.aalen_johansen(time_obs[stratum == s], cause[stratum == s], horizons) for s in range(4)]
for name, values in pds.items():
    print(f"{name:24s}", " | ".join(" ".join(f"{v:5.2f}" for v in row) for row in values))
print(f"{'Aalen-Johansen':24s}", " | ".join(" ".join(f"{v:5.2f}" for v in row) for row in aj))
for name in ("PC-Hazard", "riesgos proporcionales"):
    error = np.abs(pds[name] - pds["verdad"]).mean(axis=0)
    print(f"{name:24s} {"error absoluto medio a 12, 24, 36 meses (pp)"}: " + ", ".join(f"{v:.3f}" for v in error))
````

A los 12 meses la PD de la red queda a menos de un cuarto de punto de la verdad en todos los estratos, un error medio de
0,131 puntos, contra 1,015 de riesgos proporcionales, que le da al estrato de pico temprano 2,11 % donde la verdad es 4,04 %. El
error se estrecha hacia los 36 meses en ambos, porque la PD acumulada promedia el perfil a lo largo de los años. La fila de
Aalen-Johansen es menor que la verdad porque cuenta el prepago como una salida que termina la exposición del préstamo, que es la
incidencia que necesita una provisión. La evaluación versionada del producto compara las mismas dos ideas sobre los préstamos SBA
fuera de tiempo.

````python
# La comparación versionada fuera de tiempo de PC-Hazard y Cox en las seis fechas de reporte del caso C06
dates, c_index, brier = [], {"pch": [], "cox": []}, {"pch": [], "cox": []}
for path in sorted(DERIVED.glob("models-rd-*.json")):
    doc = json.loads(path.read_text(encoding="utf-8"))
    ev = next(m["details"]["evaluation"] for m in doc["model"] if "evaluation" in m.get("details", {}))
    dates.append(path.stem.removeprefix("models-rd-"))
    for k in ("pch", "cox"):
        c_index[k].append(ev["c_index"][k])
        brier[k].append(ev["integrated_brier"][k])
fig, (left, right) = plt.subplots(1, 2, figsize=(11, 3.6))
for k, color in (("pch", "tab:blue"), ("cox", "tab:red")):
    left.plot(dates, c_index[k], marker="o", color=color, label=k)
    right.plot(dates, brier[k], marker="o", color=color, label=k)
left.set_ylabel("índice C (mayor es mejor)")
right.set_ylabel("puntaje de Brier integrado (menor es mejor)")
for ax in (left, right):
    ax.set_xlabel("fecha de reporte")
    ax.legend(fontsize=8)
fig.tight_layout()
plt.show()
better = sum(p < c for p, c in zip(brier["pch"], brier["cox"]))
print(f"{"fechas donde PC-Hazard tiene el menor puntaje de Brier integrado"}: {better} / {len(dates)}")
````

Sobre los préstamos reales la red ordena a los deudores mucho mejor que Cox en todas las fechas de reporte, con un índice C
de 0,73 a 0,81 contra 0,56 a 0,64, y su puntaje de Brier integrado es menor en cuatro de las seis fechas (2010-12, 2019-12, 2020-06

### Úsalo con tus propios datos

La maquinaria de vida completa del producto parte de un objeto, una curva de PD acumulada por meses desde la fecha de reporte, y cada paso posterior es una llamada al motor. Su forma de contrato es la que los archivos de modelos versionados guardan en `fit.ladder[].baseline` (`months` y `cumulative_pd`); un desarrollador que tiene una curva propia, de una estimación de Kaplan-Meier, una tasa de riesgo logística o una base de Cox, la escribe en esa forma, comprueba que es estrictamente creciente y está dentro de (0, 1), lo que `from_cumulative` exige, y obtiene la tasa de riesgo, la supervivencia, la PD marginal y la PD a futuro entre dos meses cualesquiera de la curva. Desde ahí las llamadas son las de los pasos 4 y 5: `competing_risks` para una salida por prepago y `ecl_scenario` para la PCE de una cuenta. La celda usa, como sustituto de una curva propia, la base de la rama Kaplan-Meier versionada para la fecha de reporte 2020-06-30 (`data/derived/C06/models-rd-2020-06.json`), y valora la cuenta declarada del paso 4 con y sin un riesgo de prepago de 1% por trimestre. Termina entregando al motor la misma curva invertida, que ninguna PD acumulada puede ser.

````python
# Una curva de PD acumulada propia con la forma del contrato: estructura de plazos, salida competidora, PCE y rechazos
own_doc = json.loads((DERIVED / "models-rd-2020-06.json").read_text(encoding="utf-8"))
own_km = next(r for r in own_doc["fit"]["ladder"] if r["rung_id"] == "L0-km")["baseline"]
own_months, own_cum = np.array(own_km["months"]), np.array(own_km["cumulative_pd"])
own_ts = tsm.from_cumulative(own_cum, own_months)
k36 = int(np.flatnonzero(own_months == 36)[0])
print("primer mes, último mes, puntos:", int(own_months[0]), int(own_months[-1]), own_months.size)
print("F(36), S(36):", round(float(own_ts.cumulative[k36]), 5), round(float(own_ts.survival[k36]), 5))
print("PD a futuro 12 a 36:", round(float(tsm.forward_pd(own_ts, 12, 36)), 5))
for label, prepay in zip(("PCE a 12 y 36 meses, con censura:", "igual, prepago 1% por trimestre:"), (0.0, 0.01)):
    rival = tsm.competing_risks(own_ts.hazard, np.full(own_ts.hazard.shape, prepay), own_months)
    pair = [float(ecl_engine.ecl_scenario(rival.marginal_default[None, :], LGD, EAD, EIR, own_months, cut)[0])
            for cut in (12, 36)]
    print(label, [round(v, 2) for v in pair])
try:
    tsm.from_cumulative(own_cum[::-1], own_months)
except ValueError as error:
    print("una PD acumulada que baja se rechaza con", type(error).__name__)
````

La curva tiene 40 puntos, uno cada tres meses del mes 3 al mes 120. Su PD acumulada a 36 meses es 0,04631 y su supervivencia 0,95369, y la PD a futuro entre los meses 12 y 36, condicionada a haber sobrevivido 12 meses, es 0,04527. La cuenta declarada cuesta 54,33 USD de PCE a 12 meses y 2.086,79 a 36 con censura; con un riesgo de prepago de 1% por trimestre las cifras son 52,81 y 1.923,54, es decir 2,8% y 7,8% menos, en línea con el barrido anterior (1,0287 y 1,082 para el riesgo versionado de 1%). Una curva que baja se rechaza con un `ValueError`. Qué revisar en una curva propia: que sea acumulada desde la fecha de reporte y no por edad de originación salvo que la cuenta sea nueva; que los meses sean los meses finales de los períodos y estrictamente crecientes; que su extremo superior no se extrapole más allá de los datos (las tasas de riesgo de los últimos puntos descansan en pocos préstamos, como mostró el paso 8); que el prepago se trate de forma explícita, con la sensibilidad del barrido, en vez de por omisión como censura; y que la curva se someta a la prueba retrospectiva por cohorte con la prueba de vida completa del notebook 05 antes de citar cualquier PCE desde ella.
