## Método

### Paso 1: la distribución de la cartera grande frente a la integral de su densidad

La función de distribución y la densidad del cuaderno 01 son dos funciones del motor. Si están bien, la integral de la densidad de 0 a $x$ es la función de distribución en $x$. La integral la calcula la cuadratura adaptativa de SciPy, que no usa ninguna de las dos formas cerradas.

````python
# Integrar la densidad del motor con la cuadratura de SciPy y compararla con la función de distribución del motor
from riskvalidation.engines.vasicek import lhp_cdf, lhp_pdf

print(f"{'x':>7s} {'integral de la densidad':>26s} {'función de distribución':>26s} {'diferencia':>12s}")
for x in (0.005, 0.01, 0.02, 0.05):
    area, _ = integrate.quad(lambda t: float(lhp_pdf(t, pd_hat, rho_hat)), 0.0, x, limit=200)
    cdf = float(lhp_cdf(x, pd_hat, rho_hat))
    print(f"{x:7.3f} {area:26.12f} {cdf:26.12f} {area - cdf:12.1e}")
````

### Paso 2: el capital IRB frente a la fórmula escrita con SciPy

El requerimiento de capital IRB de Basilea para una exposición de PD y LGD dadas es el capital al 99,9% de la cartera grande con la correlación regulatoria $R$ (BCBS, 2005, https://www.bis.org/publications/explanatory-note-basel-ii-irb-risk-weight-functions.pdf; Basel Framework CRE31, https://www.bis.org/basel_framework/chapter/CRE/31.htm). Para una exposición minorista renovable, que tiene una correlación fija $R=0{,}04$ y ningún ajuste por vencimiento, es

$$K=\mathrm{LGD}\left[\Phi\!\left(\frac{\Phi^{-1}(\mathrm{PD})+\sqrt{R}\,\Phi^{-1}(0.999)}{\sqrt{1-R}}\right)-\mathrm{PD}\right]$$

La celda siguiente la evalúa con la distribución normal de SciPy en la PD ajustada de la cartera de tarjetas de crédito, y la compara con tres respuestas del motor: `capital_requirement` del módulo IRB, el capital de la cartera grande `lhp_measures` con la misma correlación, y el `k_irb` versionado.

````python
# Evaluar la fórmula IRB con SciPy y compararla con el módulo IRB, el capital de la cartera grande y el valor versionado
from riskvalidation.portfolio.portfolio import lhp_measures
from riskvalidation.regulatory.irb import capital_requirement

c_out = cards["outputs"]
pd_cards = c_out["fit"]["maximum_likelihood"]["pd"]["estimate"]
reference = c_out["irb_reference"]
lgd, corr = 0.45, reference["irb_correlation"]

k_scipy = lgd * (stats.norm.cdf((stats.norm.ppf(pd_cards) + math.sqrt(corr) * stats.norm.ppf(0.999)) / math.sqrt(1 - corr)) - pd_cards)
k_module = float(capital_requirement(reference["asset_class"], pd_cards, lgd, regime="basel3").k)
k_lhp = lhp_measures(1.0, lgd, pd_cards, corr, (0.999,))[0].ec
print("clase de activos:", reference["asset_class"], "| PD:", round(pd_cards, 6), "| R:", corr, "| LGD:", lgd)
print(f"K fórmula con SciPy      {k_scipy:.12f}")
print(f"K módulo IRB             {k_module:.12f}   diferencia {k_module - k_scipy:.1e}")
print(f"K cartera grande        {k_lhp:.12f}   diferencia {k_lhp - k_scipy:.1e}")
print(f"K versionado              {reference['k_irb']:.12f}   diferencia {reference['k_irb'] - k_scipy:.1e}")
````

### Paso 3: CreditRisk+ frente a sus formas cerradas

La recursión de CreditRisk+ del cuaderno 04 calcula toda la distribución de la pérdida en unidades. Dos propiedades del modelo se conocen en forma cerrada desde su función generadora, así que anclan la recursión sin una segunda implementación de ella. Con un sector, sin peso idiosincrático y varianza Gamma $\sigma^2$, la probabilidad de ninguna pérdida es $P(X=0)=(1+\sigma^2\mu)^{-1/\sigma^2}$ con $\mu=\sum_ip_i$ (Haaf, Reiss y Schoenmakers, 2004, https://doi.org/10.21314/jor.2004.097), y, condicionando en el factor Gamma $S$ (media 1, varianza $\sigma^2$), la media y la varianza de la pérdida en unidades son

$$E[X]=\sum_i\nu_ip_i,\qquad \mathrm{Var}(X)=\sum_i\nu_i^2p_i+\sigma^2\Big(\sum_i\nu_ip_i\Big)^2$$

(el primer término es la varianza de Poisson dado $S$, el segundo la varianza de la media condicional; es una derivación del modelo, no una fórmula citada). La celda siguiente corre la recursión estable en la cartera comercial y compara la media y la varianza de su función de probabilidad con estas.

````python
# Correr la recursión estable y comparar P(X = 0), la media y la varianza con las formas cerradas
from riskvalidation.portfolio import creditrisk_plus as cr

crp = rungs["c4-creditrisk-plus"]
sigma2 = next(p["value"] for p in crp["parameters"] if p["name"] == "sigma2")
loss_unit = next(p["value"] for p in crp["parameters"] if p["name"] == "loss_unit")
n_max = int(next(d["value"] for d in crp["diagnostics"] if d["name"] == "n_max"))
ead = np.concatenate([np.full(g["n_names"], g["ead_each"]) for g in book["groups"]])
pd_names = np.concatenate([np.full(g["n_names"], g["pd"]) for g in book["groups"]])
units = np.rint(ead * book["lgd_mean"] / loss_unit).astype(int)
inp = cr.CreditRiskPlusInput(nu=units, pd=pd_names, w0=np.zeros(ead.size), weights=np.ones((ead.size, 1)), sigma2=np.array([sigma2]), loss_unit=loss_unit, currency=book["currency"])
pmf = cr.pmf_stable(inp, n_max)

n = np.arange(pmf.size)
mean_pmf = float((n * pmf).sum())
var_pmf = float(((n - mean_pmf) ** 2 * pmf).sum())
mu = float(pd_names.sum())
mean_closed = float((units * pd_names).sum())
var_closed = float((units ** 2 * pd_names).sum() + sigma2 * mean_closed ** 2)
print("masa fuera de la grilla:", f"{1 - pmf.sum():.1e}")
print(f"P(X = 0)  recursión {pmf[0]:.12e} | forma cerrada {(1 + sigma2 * mu) ** (-1 / sigma2):.12e}")
print(f"media      recursión {mean_pmf:.6f} | forma cerrada {mean_closed:.6f}")
print(f"varianza   recursión {var_pmf:.4f} | forma cerrada {var_closed:.4f} | brecha relativa {abs(var_pmf - var_closed) / var_closed:.1e}")
````

### Paso 4: el vector de prueba de granularidad del artículo

Gordy y Lutkebohmert (2013, https://www.ijcb.org/sites/default/files/journal/v9n3/ijcb-v9n3-granularity-adjustment-regulatory-capital-assessment.pdf) imprimen un número que cualquiera que implemente su ajuste puede reproducir: $\delta=4{,}83$ con $\xi=0{,}25$ y $q=0{,}999$. La investigación lo leyó directamente del artículo. El motor calcula $\delta=(\alpha_q-1)(\xi+(1-\xi)/\alpha_q)$ a partir del cuantil de una distribución Gamma. La segunda identidad es la que implica la fórmula del artículo: para $n$ nombres idénticos el ajuste es proporcional a $1/n$, así que $\mathrm{GA}\times n$ no se mueve. La celda siguiente reproduce el vector y comprueba la identidad.

````python
# Reproducir el vector de prueba publicado y comprobar que GA por n es constante para nombres idénticos
from riskvalidation.portfolio import granularity as gr

gran = out["granularity"]
delta = gr.delta(gran["xi"], 0.999)
print("delta (xi = 0.25, q = 0.999):", round(delta, 4), "| publicado:", gran["delta_reference"], "| el motor coincide hasta los dígitos impresos:", si_no(round(delta, 2) == gran["delta_reference"]))

k_names = float(capital_requirement("corporate", pd_hat, 0.45, regime="basel3", maturity=2.5).k)
products = []
for n_names in (10, 100, 1000, 10_000):
    result = gr.granularity_adjustment(np.full(n_names, 1 / n_names), np.full(n_names, k_names), np.full(n_names, pd_hat), np.full(n_names, 0.45), gran["gamma"], gran["xi"], 0.999)
    products.append(result.ga * n_names)
    print(f"n = {n_names:6d} | GA = {result.ga:.8f} | GA x n = {result.ga * n_names:.8f}")
print("mayor dispersión relativa de GA x n:", f"{(max(products) - min(products)) / max(products):.1e}")
````

El Paso 5 regenera el peldaño de cópula gaussiana versionado con el motor publicado. La escalera se calculó con el motor 0.7.0 y el motor instalado es el 0.8.1, y el peldaño se sorteó con un generador con semilla y un tamaño de bloque fijo, así que un motor fiel debe devolver la misma muestra. La corrida usa los 1.000.000 de escenarios del peldaño versionado, porque la igualdad bit a bit es el punto (toma alrededor de medio minuto y es la única celda del tema con el tamaño versionado). La celda siguiente la repite y compara VaR, ES y sus errores estándar, dígito a dígito, con los valores versionados.

````python
# Regenerar el peldaño de cópula gaussiana de un factor versionado con el motor publicado y compararlo dígito a dígito
from riskvalidation.portfolio import copula
from riskvalidation.portfolio import portfolio as pf

committed = rungs["c2-gaussian-1f"]
sim = committed["simulation"]
one = pf.portfolio_from_groups(book["groups"], book["currency"], book["lgd_gamma"], one_factor=True)
sample = copula.simulate_gaussian(one, sim["n_scenarios"], sim["seed"], sim["chunk_size"])
rerun = pf.measures_from_sample(sample.losses, (0.99, 0.999), pf.expected_loss(one))

print("escenarios:", sim["n_scenarios"], "| semilla:", sim["seed"], "| bloque:", sim["chunk_size"], "| peldaño versionado calculado con riskvalidation", ci["provenance"]["riskvalidation_version"])
identical = []
for m, c in zip(rerun, committed["measures"]):
    for name, mine, theirs in (("VaR", m.var, c["var"]), ("ES", m.es, c["es"]), ("error del VaR", m.var_se, c["var_se"]), ("error del ES", m.es_se, c["es_se"])):
        identical.append(math.isclose(mine, theirs, rel_tol=1e-8))
        print(f"q = {m.level}: {name:16s} {mine:16,.4f} | versionado {theirs:16,.4f} | diferencia relativa {abs(mine - theirs) / theirs:.1e}")
print("los ocho valores coinciden a 8 dígitos:", si_no(all(identical)))
````

### Paso 6: el estimador de correlación frente a una verdad conocida

La grilla del simulador es el propio estimador del motor contrastado con una verdad que los datos no pueden ocultar. Cada celda sortea 400 series de fracciones de incumplimiento anuales con una correlación conocida, ajusta los dos estimadores del cuaderno 02, y registra la estimación media, el **sesgo** (la estimación media menos el valor conocido), el error cuadrático medio y la **cobertura** del intervalo nominal de 95%, la parte de las réplicas cuyo intervalo contiene la correlación verdadera. Un intervalo bien calibrado tiene cobertura cercana a 0,95. La celda siguiente imprime esto para la grilla de 22 años en cuatro correlaciones conocidas, y para 40 años en las mismas, desde las `cells` versionadas.

````python
# Imprimir la estimación media, el sesgo, el error y la cobertura de ambos estimadores en cuatro correlaciones conocidas para 22 y 40 años
print(f"{'años':>6s} {'rho conocida':>12s} {'estimador':>20s} {'media':>8s} {'sesgo':>8s} {'RMSE':>7s} {'cobertura':>9s}")
for years, grid in ((22, grid22), (40, grid40)):
    for cell in grid["outputs"]["cells"]:
        if round(cell["known_rho"], 2) not in (0.02, 0.06, 0.12, 0.24):
            continue
        for e in cell["estimators"]:
            print(f"{years:6d} {cell['known_rho']:12.2f} {e['estimator']:>20s} {e['mean_estimate']:8.4f} {e['bias']:8.4f} {e['rmse']:7.4f} {e['coverage']:9.3f}")
failed = sum(c["n_failed"] for g in (grid22, grid40) for c in g["outputs"]["cells"])
print("réplicas que no lograron ajustarse, todas las celdas de ambas grillas:", failed)
````

### Paso 7: una verificación en vivo de una celda, y el tamaño y la potencia de la prueba

La grilla versionada le tomó horas al pipeline, así que no la repetimos. Verificamos una celda, la de tamaño, con un número menor de réplicas sorteadas aquí desde la definición del modelo: 22 años de $x_t=p(z_t)$ con $z_t\sim N(0,1)$, la PD verdadera de la grilla (2%) y la correlación verdadera 0,12, luego el ajuste de máxima verosimilitud del motor y el intervalo de perfil del cuaderno 02. La prueba de H0: $\rho=0{,}12$ rechaza cuando 0,12 queda fuera del intervalo, así que la tasa de rechazo en el valor verdadero es el tamaño, y la parte de intervalos que contienen 0,12 es la cobertura. Con 100 réplicas el error de Monte Carlo es grande, así que el resultado se lee con su intervalo de Wilson (Wilson, 1927, https://doi.org/10.1080/01621459.1927.10502953) y se pone junto a las tasas versionadas de la grilla de 400 réplicas, cuyas filas vienen de `correlation_size_power` del motor.

````python
# Verificar en vivo la celda de tamaño con 100 réplicas y compararla con las filas de tamaño versionadas
from riskvalidation.engines.vasicek import conditional_pd
from riskvalidation.portfolio.asrf_fit import fit_ml, profile_interval
from riskvalidation.validation.portfolio import correlation_size_power

generator = grid22["outputs"]["generator"]
true_rho, true_pd, years_n, null_rho = 0.12, generator["known_pd"], generator["series_length_years"], generator["null_rho"]
rng = np.random.default_rng(20261021)
rejected, estimates = [], []
for _ in range(100):
    rates = np.asarray(conditional_pd(true_pd, true_rho, rng.standard_normal(years_n)))
    fitted = fit_ml(rates)
    low, high = profile_interval(rates, 0.95)["rho"]
    rejected.append(not (low <= null_rho <= high))
    estimates.append(fitted.rho)

live = correlation_size_power(np.array(rejected), true_rho, null_rho, alpha=generator["alpha"], estimator="maximum-likelihood")
print("en vivo, 100 réplicas: tasa de rechazo (tamaño)", round(live.metric, 3), "| error", round(live.extras["se"], 3), "| Wilson", round(live.extras["ci_low"], 3), "-", round(live.extras["ci_high"], 3), "| estimación media", round(float(np.mean(estimates)), 4))
for years, grid in ((22, grid22), (40, grid40)):
    for row in grid["tests"]:
        if row["extras"]["role"] == "size":
            print(f"grilla versionada {years} años, {row['extras']['estimator']:20s}: tamaño {row['metric']:.4f} | error {row['extras']['se']:.4f} | Wilson {row['extras']['ci_low']:.3f} - {row['extras']['ci_high']:.3f} | luz {si_no(row['light'])}")
committed_size = next(r for r in grid22["tests"] if r["extras"]["role"] == "size" and r["extras"]["estimator"] == "maximum-likelihood")["metric"]
print("el tamaño versionado de 22 años cae dentro del intervalo de Wilson en vivo:", si_no(live.extras["ci_low"] <= committed_size <= live.extras["ci_high"]))
````

La última figura pone el tamaño y la potencia versionados en una imagen para el estimador de máxima verosimilitud: la tasa de rechazo frente a la correlación conocida con 22 y con 40 años, con el intervalo de Wilson de cada tasa, el nivel nominal y la puerta de potencia de 80%. Es la respuesta, en este generador, a la pregunta que un validador hace de cualquier prueba: con qué frecuencia rechaza cuando no debe, y con qué frecuencia no rechaza cuando debe.

````python
# Dibujar el tamaño y la potencia versionados de la prueba de máxima verosimilitud con 22 y 40 años con sus intervalos de Wilson
fig, ax = plt.subplots(figsize=(8, 3.8))
for (years, grid), color in zip(((22, grid22), (40, grid40)), ("tab:blue", "tab:red")):
    rows = [r for r in grid["tests"] if r["extras"]["estimator"] == "maximum-likelihood"]
    rho_known = np.array([r["extras"]["known_rho"] for r in rows])
    rate = np.array([r["extras"]["rate"] for r in rows])
    ax.plot(rho_known, rate, marker="o", color=color, label=f"{years} años")
    ax.fill_between(rho_known, [r["extras"]["ci_low"] for r in rows], [r["extras"]["ci_high"] for r in rows], color=color, alpha=0.15)
ax.axhline(0.05, color="gray", linestyle=":", label="tamaño nominal 0.05")
ax.axhline(0.80, color="green", linestyle=":", label="puerta de potencia 0.80")
ax.axvline(0.12, color="black", linestyle="--", linewidth=0.8)
ax.set_xlabel("correlación conocida")
ax.set_ylabel("tasa de rechazo de H0: rho = 0.12")
ax.set_title("Tamaño y potencia de la prueba de correlación de máxima verosimilitud")
ax.legend(fontsize=8)
fig.tight_layout()
plt.show()
````
