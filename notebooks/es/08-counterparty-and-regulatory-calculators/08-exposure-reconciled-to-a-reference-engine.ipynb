## Método

### Paso 1: el lado del motor, recalculado

Para que la afirmación signifique algo el lado del motor no puede ser un número copiado de un archivo. Reconstruimos la exposición esperada del swap único y del conjunto de neteo desde la curva, la calibración, la semilla y el número de trayectorias (la construcción del notebook 01, con la misma grilla de tiempo), en la medida forward a la que se lleva el perfil descontado de ORE. La celda siguiente lo hace para ambas variantes, de modo que lo que comparamos con ORE se calcula aquí y no se copia de un archivo.

````python
# Reconstruir la simulación del notebook 01 y la EE del swap único y del conjunto de neteo
base = docs["netting-set"]["outputs"]
tenors = np.array([p["tenor_years"] for p in base["curve"]["points"]])
zeros = np.array([p["zero_rate"] for p in base["curve"]["points"]])
a, sigma = base["calibration_used"]["a"], base["calibration_used"]["sigma"]
p0, forward = hw1f.curve_functions(zeros, tenors)
named = base["named_parameters"]
report = np.array(named["time_grid_years"]["value"])
main = np.concatenate([[0.0], report])
swaps = [
    sw.Swap(t["trade_id"], t["notional"]["amount"], t["fixed_rate"], t["side"] == "payer",
            t["start_years"], t["maturity_years"], t["fixed_frequency_per_year"])
    for t in base["book"]["trades"]
]
merged = np.unique(np.round(np.concatenate([main] + [main + d for d in (14 / 360, 28 / 360)]), 12))
grid = sw.simulation_grid(merged, swaps)
r = hw1f.simulate_short_rate(grid, a, sigma, forward, named["path_count"]["value"], named["seed"]["value"])
numeraire = hw1f.bank_account(grid, r, a, sigma, p0, forward)
values = sw.swap_values_on_grid(swaps, grid, r, a, sigma, p0, forward)
ix = np.searchsorted(grid, report - sw.TIME_TOL)
p0r, nm = p0(report), numeraire[:, ix]

live = {}
for key, members in (("single-swap", [0]), ("netting-set", [0, 1, 2, 3])):
    e_d, e_se = ex.expected_exposure(ex.net_values(values[members][:, :, ix]), nm)
    live[key] = (e_d / p0r, e_se / p0r)
    print(f"{key:12s} máximo de la EE reconstruida:", round(float(live[key][0].max()), 0),
          "en", float(report[live[key][0].argmax()]), "a")
````

### Paso 2: la conciliación con ORE

Ahora el perfil reconstruido del motor se enfrenta al versionado de ORE, punto a punto, con `reconcile_ee_profile` del motor, cada lado con su propio error de Monte Carlo. Esta es la afirmación en su forma más nítida: dos implementaciones independientes, una en Python y una en C++, del mismo modelo de Hull-White sobre la misma curva y operaciones, con la diferencia medida en unidades de su propio ruido. Para la EEPE y el CVA el error combinado es el registrado en el documento. La celda siguiente imprime, para las dos variantes de la afirmación, el estadístico, el número de fechas dentro del umbral, y las diferencias de EEPE y CVA.

````python
# Conciliar la EE reconstruida de las dos variantes con el perfil de ORE, y leer las diferencias de EEPE y CVA
print(f"{'variante':13s}{'estadístico':>12s}{'fechas dentro de 2 se':>26s}{'luz':>8s}"
      f"{'EEPE (se)':>12s}{'CVA (se)':>11s}")
for key in ("single-swap", "netting-set"):
    o = docs[key]["outputs"]["ore_reconciliation"]
    ref, ref_se = np.array(o["ore_ee"]["values"]), np.array(o["ore_ee"]["mc_se"])
    t = vc.reconcile_ee_profile(live[key][0], live[key][1], ref, ref_se, k_se=2.0)
    inside = int(np.sum(np.abs(live[key][0] - ref) <= 2.0 * np.sqrt(live[key][1] ** 2 + ref_se**2)))
    e, c = o["eepe_agreement"], o["cva_agreement"]
    print(f"{key:13s}{t.statistic:12.3f}{inside:22d} /{report.size:3d}{LIGHT[t.light.value]:>8s}", end="")
    print(f"{e['difference'] / e['combined_se']:12.2f}{c['difference'] / c['combined_se']:11.2f}")
print("umbral (k_se):", t.extras["thresholds"]["k_se"])
````

La celda siguiente dibuja la EE reconstruida del swap único y del conjunto de neteo sobre el perfil de ORE, con una banda de dos errores estándar combinados alrededor del de ORE.

````python
# Dibujar la EE reconstruida sobre el perfil de ORE para ambas variantes, con una banda de 2 se combinados
fig, axes = plt.subplots(1, 2, figsize=(10, 3.8))
for ax, key in zip(axes, ("single-swap", "netting-set")):
    o = docs[key]["outputs"]["ore_reconciliation"]
    ref, ref_se = np.array(o["ore_ee"]["values"]), np.array(o["ore_ee"]["mc_se"])
    band = 2 * np.sqrt(live[key][1] ** 2 + ref_se**2)
    ax.fill_between(report, (ref - band) / 1e3, (ref + band) / 1e3, color="grey", alpha=0.3, label="ORE ± 2 se")
    ax.plot(report, ref / 1e3, "--", color="tab:blue", label="ORE 1.8.17.0")
    ax.plot(report, live[key][0] / 1e3, color="black", lw=1.5, label="motor, reconstruido")
    ax.set_xlabel("años")
    ax.set_ylabel("miles de USD")
    ax.set_title(f"EE: {key}")
    ax.legend()
fig.tight_layout()
plt.show()
````

En ambos paneles la línea negra corre dentro de la banda gris, el error combinado de los dos motores, de la primera fecha a la última, y sigue la línea segmentada de ORE a través del máximo. Esta es la afirmación en su forma más nítida: una simulación en Python y una en C++ del mismo modelo de Hull-White, escritas de forma independiente, producen perfiles que no se distinguen dado su propio ruido. La banda tiene dos errores estándar combinados de ancho, un umbral fijado antes de mirar los datos. Los paneles cubren dos carteras sin garantía; la sección siguiente pasa a las carteras donde la afirmación no se cumple.

## Comparación

Qué ofrece la práctica, tal como la investigación persistida del caso la registra desde las fuentes primarias. Las herramientas de exposición de contraparte dan los estimadores y la simulación y dejan la conciliación al usuario: ORE es el motor abierto más completo y la referencia natural, y QuantLib aporta las formas cerradas del modelo de Hull-White; ninguno trae una declaración de cuánto coincide con él una implementación independiente. La regla de la propia investigación para Contraste es que los números de ORE y de QuantLib pueden presentarse como una validación del producto **solo hasta donde llegan las conciliaciones realmente corridas**, nunca más allá. La práctica de ingeniería detrás de una conciliación es la que se usa en este tema: ambos lados llevan un error de Monte Carlo, la diferencia se enuncia en errores estándar combinados, y se fija un umbral antes de la comparación, aquí dos errores estándar combinados, un umbral interno de Contraste y no uno regulatorio.

## Evidencia

### Paso 3: las brechas, informadas

La afirmación cubre las variantes donde la conciliación se cumple, y el documento informa las otras. La celda siguiente lee las seis variantes y las ordena con la regla del propio documento: conciliada, brecha informada o no aplicable. Para cada una muestra el número de fechas de EE dentro del umbral, las diferencias de EEPE y CVA en errores estándar combinados, y para la variante con margen inicial la razón de los dos DIM medios. Nada se oculta promediando: se imprime la mayor brecha de cada perfil.

````python
# Ordenar las seis variantes con la regla del propio documento e imprimir el tamaño de cada brecha informada
claimed, gaps, na, worst = [], [], [], {}
print(f"{'variante':15s}{'veredicto':18s}{'fechas de EE en 2 se':>24s}{'mayor brecha de EE (se)':>28s}"
      f"{'EEPE (se)':>11s}{'CVA (se)':>10s}")
for n in names:
    o = docs[n]["outputs"]
    rec = o["ore_reconciliation"]
    if rec["status"] == "not-applicable":
        na.append(n)
        print(f"{n:15s}{'no aplicable':18s}")
        continue
    ag = rec["agreement"]
    inside = sum(p["within_two_se"] for p in ag)
    worst[n] = max(abs(p["difference"]) / p["combined_se"] for p in ag)
    verdict = "afirmada" if rec["all_within_two_se"] else "brecha informada"
    (claimed if rec["all_within_two_se"] else gaps).append(n)
    e, c = rec["eepe_agreement"], rec["cva_agreement"]
    print(f"{n:15s}{verdict:18s}{inside:20d} /{len(ag):3d}{worst[n]:28.1f}", end="")
    print(f"{e['difference'] / e['combined_se']:11.2f}{c['difference'] / c['combined_se']:10.2f}")
print("afirmadas:", claimed, "| brecha informada:", gaps, "| no aplicables:", na)
````

La variante con margen inicial tiene un número más que leer: el margen inicial dinámico medio de ambos motores, y los hallazgos que el propio caso registró contra ORE para las variantes con brecha. La celda siguiente imprime ambos.

### Paso 4: las otras referencias

Otras dos referencias respaldan la afirmación. Primero, las formas cerradas del modelo de Hull-White contra QuantLib 1.44, como cuatro filas con ambos valores y un umbral de $10^{-6}$ en términos relativos. Segundo, la variante de correlación adversa no tiene referencia en ORE, porque el ejemplo de ORE es otro modelo; su referencia es el Apéndice de Hull y White (2012), que imprime una calibración de tres trayectorias. Los dígitos del artículo son los valores de referencia: el $a(t^*)$ impreso y las tasas de riesgo $h_{ij}$. `wwr_calibrate_a`, del motor, debe darlos, y en $b=0$ `wwr_cva` debe devolver el CVA independiente de forma exacta. Las dos celdas siguientes verifican las cuatro cosas.

````python
# Las cuatro filas de QuantLib, y los valores del Apéndice frente a la calibración de correlación adversa
ql = [row for row in docs["single-swap"]["tests"] if row["test_id"].startswith("ccr.quantlib")]
rel = [abs(row["extras"]["engine"] - row["extras"]["quantlib"]) / abs(row["extras"]["quantlib"]) for row in ql]
print("filas de QuantLib:", len(ql), "| mayor diferencia relativa:", f"{max(rel):.2e}",
      "| el umbral 1e-6 se cumple:", "sí" if max(rel) < 1e-6 else "no")

w_app = np.array([[100.0, 100.0], [200.0, 300.0], [300.0, 400.0]])
target = cva.survival_from_spread(np.array([0.01, 0.01]), np.array([0.5, 1.0]), 1.0)
a_app, h_app = cva.wwr_calibrate_a(w_app, target, np.array([0.5, 0.5]), 0.01)
printed_a = np.array([-6.9128, -7.8509])
printed_h = np.array([[0.00270, 0.00735, 0.01998], [0.00106, 0.00782, 0.02126]])
unit_a, unit_h = 1e-4, 1e-5
print("a(t*) brecha con los dígitos impresos, en unidades del último dígito impreso:",
      np.round(np.abs(a_app - printed_a) / unit_a, 2))
print("h mayor brecha, en unidades del último dígito impreso:",
      round(float(np.max(np.abs(h_app.T - printed_h)) / unit_h), 2))
mean_survival = np.exp(-np.cumsum(h_app * 0.5, axis=1)).mean(axis=0)
print("mayor brecha entre la supervivencia promedio tras cada paso y la meta:",
      float(np.max(np.abs(mean_survival - target))))
````

La segunda celda recalcula la variante de correlación adversa: en $b=0$ el CVA con correlación adversa debe ser exactamente el CVA independiente, y en $b=1$ y $b=2$ debe subir en los montos que da el barrido.

````python
# El CVA con correlación adversa en b = 0 contra el CVA independiente, y su alza en b = 1 y b = 2
wwr = docs["wrong-way-risk"]["outputs"]
credit = wwr["cva_dva"]
survival = np.array([b["survival"] for b in credit["buckets"]])
dt = np.diff(np.concatenate([[0.0], report]))
v_net = ex.net_values(values[:, :, ix])
w, x_disc = v_net / 1e6, np.maximum(v_net, 0.0) / nm
epe_d = np.array(wwr["exposure"]["ee_discounted"]["values"])
pd_cp = cva.default_probabilities(np.concatenate([[1.0], survival]))
independent, _ = cva.cva(epe_d, pd_cp, credit["lgd"])
value0, _ = cva.wwr_cva(w, x_disc, survival, dt, credit["lgd"], 0.0)
print("el CVA con correlación adversa en b = 0 es igual al CVA independiente:",
      "sí" if value0 == independent else "no", "|", round(independent, 2))
for b in (1.0, 2.0):
    value, _ = cva.wwr_cva(w, x_disc, survival, dt, credit["lgd"], b)
    print(f"b = {b:.0f} | CVA {value:10.2f} | cambio frente al independiente {value / independent - 1:+.1%}")
````

### Paso 5: la hipótesis detrás del umbral y su tamaño

El Paso 2 informó un número por variante, y un número de una corrida del motor es una extracción. La afirmación de este notebook descansa en una hipótesis: si el motor y ORE implementan el mismo modelo sobre las mismas operaciones y curva, sus perfiles difieren solo por ruido de Monte Carlo. Bajo esa hipótesis la brecha estandarizada en cada fecha tiene media cero y varianza uno, y la afirmación es una regla sobre su máximo:

$$z_k = \frac{EE_k^{engine} - EE_k^{ORE}}{\sqrt{se_k^2 + \left(se_k^{ORE}\right)^2}}, \qquad Z = \max_{k=1,\dots,33}|z_k|, \qquad \text{conciliada si y solo si } Z \le 2$$

El umbral tiene un tamaño que se puede derivar y luego medir. Si las 33 fechas fueran independientes, un motor correcto pasaría con probabilidad

$$\Pr(Z \le 2) = \left(2\Phi(2) - 1\right)^{33} = 0{,}9545^{33} \approx 0{,}215$$

lo que haría al umbral demasiado estricto. Las fechas no son independientes, porque la exposición esperada de fechas vecinas se calcula con los mismos caminos, así que el tamaño hay que medirlo. La celda siguiente vuelve a correr el motor 30 veces con semillas distintas, con 20.000 caminos cada vez como en el caso, y calcula $Z$ frente al perfil versionado de ORE para el swap único y para el conjunto de neteo. El perfil de ORE es una corrida versionada, así que su ruido es una constante que comparten las 30 corridas.

````python
# El estadístico frente a ORE en 30 corridas independientes del motor (20.000 caminos), carteras sin garantía
n_runs = 30
books = {"single-swap": [0], "netting-set": [0, 1, 2, 3]}
refs = {}
for n in books:
    ore_block = docs[n]["outputs"]["ore_reconciliation"]["ore_ee"]
    refs[n] = (np.array(ore_block["values"]), np.array(ore_block["mc_se"]))
live_stat = {}
for n in books:
    live_stat[n] = vc.reconcile_ee_profile(live[n][0], live[n][1], *refs[n], k_se=2.0).statistic
big_z = {name: [] for name in books}
for run_seed in range(101, 101 + n_runs):
    r_s = hw1f.simulate_short_rate(grid, a, sigma, forward, named["path_count"]["value"], run_seed)
    v_s = sw.swap_values_on_grid(swaps, grid, r_s, a, sigma, p0, forward)[:, :, ix]
    nm_s = hw1f.bank_account(grid, r_s, a, sigma, p0, forward)[:, ix]
    for name, members in books.items():
        e_d, e_se = ex.expected_exposure(ex.net_values(v_s[members]), nm_s)
        ref, ref_se = refs[name]
        big_z[name].append(vc.reconcile_ee_profile(e_d / p0r, e_se / p0r, ref, ref_se, k_se=2.0).statistic)
````

La celda siguiente reduce los 30 estadísticos de cada cartera a su media, sus percentiles 5% y 95% y la parte que pasa el umbral, con el intervalo de Wilson de esa parte, y pone el resultado junto a la probabilidad que darían fechas independientes.

````python
# La tasa de aprobación con su intervalo de Wilson, frente a lo que darían fechas independientes
import math


def wilson(k, n, z=1.96):
    # Intervalo de puntaje de Wilson de una proporción k / n
    centre = (k / n + z**2 / (2 * n)) / (1 + z**2 / n)
    half = z * np.sqrt(k / n * (1 - k / n) / n + z**2 / (4 * n**2)) / (1 + z**2 / n)
    return centre - half, centre + half


for name, values_z in big_z.items():
    arr = np.array(values_z)
    k_pass = int(np.sum(arr <= 2.0))
    low, high = wilson(k_pass, arr.size)
    print(f"{name:12s} | media {arr.mean():.3f} | 5%-95% {np.percentile(arr, 5):.3f}-{np.percentile(arr, 95):.3f}"
          f" | corrida versionada {live_stat[name]:.3f} | pasan {k_pass}/{arr.size} = {k_pass / arr.size:.2f}"
          f" (intervalo de 95% {low:.2f}-{high:.2f})")
independent = math.erf(2 / math.sqrt(2)) ** 33
print("probabilidad de pasar con 33 fechas independientes, (2 Phi(2) - 1)^33:", round(independent, 3))
````

Los 30 estadísticos del swap único tienen una media de 0,998, con 90% de las corridas entre 0,721 y 1,292; los del conjunto de neteo tienen una media de 1,049 y van de 0,717 a 1,369. La corrida versionada, 0,775 y 0,715, es una extracción típica y no afortunada, bien dentro de ambos rangos. Las 30 corridas de cada cartera pasan el umbral de 2, una proporción de 1,00 con un intervalo de Wilson de 0,89 a 1,00, frente a los 0,215 que darían fechas independientes: la correlación entre fechas hace al umbral mucho menos estricto de lo que sugiere el argumento de independencia. El intervalo dice que la probabilidad de aprobar de un motor correcto es mayor que 89% con 95% de confianza, y ahí se detiene, porque 30 corridas no pueden mostrar que sea exactamente uno. La prueba es generosa con un motor correcto. Lo que limita la afirmación es el otro lado, cuán grande es un defecto que la prueba aún ve.

### Sensibilidad y límites

Una conciliación puede fallar de dos maneras. Puede rechazar un motor correcto, lo que las corridas de arriba muestran que casi no hace, y puede aceptar uno errado, que es el caso que limita la afirmación. Su potencia se mide plantando un defecto: la celda siguiente corre el motor con una volatilidad $\sigma$ errada en 0, 2, 5 y 10 por ciento, ocho corridas cada una, y concilia cada corrida con el perfil de ORE del modelo verdadero.

````python
# Potencia: el motor con sigma errado en 0, 2, 5 y 10 por ciento frente al perfil de ORE, 8 corridas cada una
rel_errors = (0.0, 0.02, 0.05, 0.10)
ref, ref_se = refs["single-swap"]
power = {}
for rel_err in rel_errors:
    stats = []
    for run_seed in range(201, 209):
        s_wrong = sigma * (1.0 + rel_err)
        r_w = hw1f.simulate_short_rate(grid, a, s_wrong, forward, named["path_count"]["value"], run_seed)
        v_w = sw.swap_values_on_grid(swaps[:1], grid, r_w, a, s_wrong, p0, forward)[0][:, ix]
        nm_w = hw1f.bank_account(grid, r_w, a, s_wrong, p0, forward)[:, ix]
        e_d, e_se = ex.expected_exposure(v_w, nm_w)
        stats.append(vc.reconcile_ee_profile(e_d / p0r, e_se / p0r, ref, ref_se, k_se=2.0).statistic)
    power[rel_err] = np.array(stats)

print(f"{'sigma errado en':>16s}{'estadístico medio':>19s}{'menor':>8s}{'mayor':>9s}{'rechazado':>11s}")
for rel_err, arr in power.items():
    print(f"{rel_err:16.0%}{arr.mean():19.2f}{arr.min():8.2f}{arr.max():9.2f}{float(np.mean(arr > 2.0)):11.2f}")
````

Con el $\sigma$ correcto el estadístico promedia 1,02, entre 0,85 y 1,25, y ninguna corrida se rechaza. Con un $\sigma$ 2% más alto promedia 1,98, las corridas van de 1,64 a 2,38 y la mitad se rechaza: un error de 2% en la volatilidad está en el borde de lo que la conciliación ve con 20.000 caminos. Con 5% la media es 3,73 y las ocho corridas se rechazan, y con 10% la media es 6,76. El método resuelve entonces un error de volatilidad de cerca de 2% y no dice nada bajo él. Un sesgo menor que el ruido combinado de los dos motores es invisible, y la afirmación "conciliada" significa "ninguna diferencia mayor que cerca de 2% de volatilidad", no "idéntica". Las tres variantes con garantía están en 7,1, 9,9 y 11,7 errores estándar combinados, sobre los 6,76 que produce un error de 10% en la volatilidad, así que con esta vara sus brechas son grandes.

````python
# Dibujar el estadístico en 30 corridas (izquierda), el rechazo con sigma errado (centro), las cinco variantes
fig, (left, mid, right) = plt.subplots(1, 3, figsize=(15, 4.0))
for name, color in (("single-swap", "tab:green"), ("netting-set", "tab:olive")):
    left.hist(big_z[name], bins=np.linspace(0.4, 3.0, 14), alpha=0.6, color=color, label=name)
    left.axvline(live_stat[name], color=color, ls="--", lw=1.5)
left.axvline(2.0, color="black", ls=":", lw=1.2, label="umbral: 2")
left.set_xlabel("mayor |motor - ORE| / se combinado, una corrida")
left.set_ylabel("corridas de 30")
left.set_title("El estadístico de un motor correcto")
left.legend(fontsize=8)
mid.plot([100 * e_ for e_ in rel_errors], [float(np.mean(power[e_] > 2.0)) for e_ in rel_errors], "o-", color="tab:red",
         label="parte de corridas rechazadas")
mid.plot([100 * e_ for e_ in rel_errors], [power[e_].mean() / 10 for e_ in rel_errors], "s--", color="tab:blue",
         label="estadístico medio / 10")
mid.set_xlabel("error de sigma, por ciento")
mid.set_ylabel("proporción, o estadístico / 10")
mid.set_title("Potencia frente a una volatilidad errada")
mid.legend(fontsize=8)
worst_all = dict(worst)
bar_colors = ["tab:green" if v <= 2.0 else "tab:red" for v in worst_all.values()]
right.bar(range(len(worst_all)), list(worst_all.values()), color=bar_colors)
band = max(np.percentile(big_z[n], 95) for n in books)
right.axhspan(0, band, color="grey", alpha=0.25, label="95% de corridas correctas")
right.set_xticks(range(len(worst_all)), list(worst_all), rotation=25, ha="right", fontsize=8)
right.set_yscale("log")
right.set_ylabel("estadístico versionado")
right.set_title("Las cinco variantes frente a la banda de ruido")
right.legend(fontsize=8)
fig.tight_layout()
plt.show()
````

A la izquierda, los histogramas de los 30 estadísticos del swap único y del conjunto de neteo se juntan entre 0,7 y 1,4 y nunca alcanzan el umbral punteado de 2, y las líneas segmentadas de las corridas versionadas quedan dentro de la masa. En el centro, la parte de corridas rechazadas (rojo) sube de 0 sin error a 0,5 con 2% y a 1 con 5%, y el estadístico medio dividido por 10 (azul) sube de 0,10 a 0,68 con 10%: la curva es empinada entre 0 y 5%, que es donde está la resolución de la prueba. A la derecha, los cinco estadísticos versionados se dibujan en escala logarítmica frente a una banda gris que contiene 95% de las corridas de un motor correcto: las dos barras verdes quedan dentro y las tres barras rojas, las variantes con garantía, se alzan bien sobre ella. Los tres paneles juntos sostienen la afirmación tal como se enuncia: dos variantes son indistinguibles del ruido, tres no, y un error de 2% en la volatilidad ya se vería.

### Úsalo con tus propios datos

Una afirmación de conciliación sobre un motor propio tiene la forma de las celdas de arriba y necesita cuatro entradas: el perfil del motor con su error de Monte Carlo en las fechas de la referencia, la referencia con su error (cero para una forma cerrada), un umbral fijado de antemano y varias corridas con semillas distintas, de modo que la afirmación sea una tasa con un intervalo y no un solo número. La celda siguiente escribe un swap propio, un pagador a 4 años por 8 millones de USD, corre el motor diez veces, calcula cada vez el estadístico frente a la forma cerrada en las fechas de reinicio e informa la parte de corridas bajo el umbral con su intervalo de Wilson. Termina con un valor de referencia para el riesgo de camino equivocado que sigue la misma lógica: con $b=0$ todo camino tiene la misma intensidad y la calibración tiene una forma cerrada, $a = \ln(s/\mathrm{LGD})$ para un spread plano $s$, que una grilla de $b$ propia debe reproducir antes de confiar en ella con $b \ne 0$. El contrato `engine-ccr-api` (`docs/contracts/engine/ccr-api.md`) fija la llamada, `wwr_calibrate_a(w, survival_target, dt, b)`, con `w` en millones de USD y $b$ por millón de USD.

````python
# Tu propia afirmación: un pagador a 4 años con diez semillas frente a su forma cerrada, y un valor de referencia
own_swap = sw.Swap("OWN-1", 8e6, 0.0370, True, 0.0, 4.0, 2)
own_grid = sw.simulation_grid(main, [own_swap])
own_ix = np.searchsorted(own_grid, report - sw.TIME_TOL)
own_resets = sw.reset_times(own_swap)
own_pick = [k for k, t in enumerate(report) if np.min(np.abs(own_resets - t)) < 1e-9]
closed_own = ex.analytic_ee_single_swap(own_swap, report[own_pick], a, sigma, p0, forward)
own_stats = []
no_err = np.zeros(len(own_pick))
for run_seed in range(301, 311):
    r_o = hw1f.simulate_short_rate(own_grid, a, sigma, forward, named["path_count"]["value"], run_seed)
    v_o = sw.swap_values_on_grid([own_swap], own_grid, r_o, a, sigma, p0, forward)[0][:, own_ix]
    e_d, e_se = ex.expected_exposure(v_o, hw1f.bank_account(own_grid, r_o, a, sigma, p0, forward)[:, own_ix])
    stat = vc.reconcile_ee_profile((e_d / p0r)[own_pick], (e_se / p0r)[own_pick], closed_own, no_err, 2.0)
    own_stats.append(stat.statistic)
passes = int(np.sum(np.array(own_stats) <= 2.0))
low, high = wilson(passes, len(own_stats))
print("estadístico en diez semillas: media", round(float(np.mean(own_stats)), 3), "| mayor", round(max(own_stats), 3))
print(f"corridas bajo el umbral: {passes} de {len(own_stats)} | intervalo de 95% de la proporción {low:.2f}-{high:.2f}")

spread, own_lgd, step = 0.02, 0.6, 0.5
target = cva.survival_from_spread(np.array([spread]), np.array([step]), own_lgd)
a_engine, _ = cva.wwr_calibrate_a(np.zeros((3, 1)), target, np.array([step]), 0.0)
a_closed = math.log(spread / own_lgd)
print("valor de referencia con b = 0, a = ln(s / LGD):")
print("  motor", round(float(a_engine[0]), 9), "| forma cerrada", round(a_closed, 9))
````

Qué mirar en la salida. **La tasa y su intervalo.** Las diez corridas pasan, el estadístico medio es 0,996 y el mayor 1,334, y el intervalo de Wilson de la parte que pasa es de 0,72 a 1,00: ancho, porque hay solo diez corridas. Diez corridas sostienen "sobre 72%" y no más, donde 30 dieron 0,89, así que una afirmación debe decir cuántas corridas la respaldan. **El error de la referencia.** Para una forma cerrada es cero, lo que hace a la prueba algo más exigente que una comparación con una simulación del mismo tamaño. **El umbral fijado primero.** Es de 2 errores estándar combinados aquí y en el Paso 2; el barrido del notebook anterior mostró qué hace ampliarlo después. **La resolución.** Con 20.000 caminos la prueba ve un error de volatilidad de cerca de 2% y no menos; dilo junto a la afirmación. **El valor de referencia.** El motor da -3,401197382 y la forma cerrada $\ln(0{,}02/0{,}6)$ da -3,401197382: nueve decimales, que es la precisión del buscador de raíces, y un valor de referencia con forma cerrada es la prueba de regresión más barata de una calibración. **El contrato.** La llamada y las unidades son las de `docs/contracts/engine/ccr-api.md`: un `w` dejado en USD, en vez de millones de USD, cambia toda intensidad.

## Resultados

**Respuesta:** Dos de las seis variantes concilian con ORE 1.8.17.0 dentro del error de Monte Carlo: el swap único, con una mayor brecha de 0,775 errores estándar combinados, y el conjunto de neteo de cuatro swaps, con 0,715, ambos con las 33 fechas dentro del umbral de 2. Tres no concilian y lo dicen: el CSA de 10 días (9,9 errores estándar), el CSA de 20 días (7,1) y la variante con margen inicial (11,7), donde el margen inicial dinámico medio de ORE es 0,884 del del motor; una no tiene referencia. Las swaptions del motor coinciden con QuantLib hasta una diferencia relativa de 9,5e-8, y su calibración de correlación adversa reproduce los valores impresos del artículo con a lo más 1,32 unidades del último dígito impreso.

- **El lado del motor se calcula aquí.** La exposición esperada del swap único y del conjunto de neteo se reconstruye desde la curva, la calibración, la semilla y el número de trayectorias, así que lo que se compara con ORE no es un número copiado de un archivo. La EEPE difiere en 0,30 y 0,31 errores estándar y el CVA en 0,37 y 0,40. El motor corrió 20.000 trayectorias; ORE corrió 10.000 muestras antitéticas de forma nativa en cerca de medio minuto.
- **Dónde no, y por cuánto.** El CSA de 10 días tiene 17 de 33 fechas dentro del umbral, una EEPE dentro del umbral (-1,55) y un CVA a 15,9 errores estándar (ORE es menor). El CSA de 20 días tiene 21 de 33 fechas dentro, y tanto la EEPE (-2,20) como el CVA (-9,88) fuera del umbral. Con margen inicial 32 de 33 fechas de EE están dentro del umbral (mayor brecha 11,7, en una fecha), la EEPE y el CVA coinciden (-0,71 y -1,40), y ninguno de los 33 puntos del margen inicial dinámico medio lo está. El caso registra cada una como un hallazgo.
- **Formas cerradas y valores de referencia.** Las cuatro filas de QuantLib coinciden hasta una diferencia relativa de 9,5e-8 como máximo, dentro del umbral de 1e-6. La calibración de correlación adversa reproduce las tasas de riesgo del Apéndice del artículo hasta 0,45 de una unidad del último dígito impreso, el $a(t_1^*)$ impreso hasta 0,33 y el $a(t_2^*)$ impreso hasta 1,32 unidades (el artículo imprime -7,8509 y la raíz de su propia ecuación, que el motor encuentra, se redondea a -7,8508); la supervivencia promedio es igual a la meta hasta 1e-12. En $b=0$ el CVA con correlación adversa es exactamente el CVA independiente, y en $b=1$ y $b=2$ está 31,2% y 64,5% sobre él.
- **Qué es la afirmación.** Dos de las seis variantes concilian con el motor de referencia dentro del error de Monte Carlo; tres no y lo dicen; una no tiene referencia. Un lector que quiera cuestionar la afirmación tiene los perfiles de ambos motores en los documentos versionados y el umbral en la prueba del motor.

- **Sensibilidad y límites.** El umbral de 2 tiene un tamaño medido: 30 de 30 corridas del motor pasan en cada cartera (intervalo de Wilson de 0,89 a 1,00, frente a 0,215 si las 33 fechas fueran independientes), y su potencia se mide con defectos plantados: una volatilidad errada en 2% se rechaza en la mitad de las corridas (estadístico medio 1,98), con 5% en todas (3,73) y con 10% en todas (6,76). La afirmación "conciliada" significa entonces ninguna diferencia mayor que cerca de 2% de volatilidad con 20.000 caminos, y las variantes con garantía, de 7,1 a 11,7, están muy más allá.
- **Con tus propios datos.** Un pagador a 4 años por 8 millones de USD pasa el umbral en 10 de 10 corridas frente a su forma cerrada (estadístico medio 0,996, intervalo de Wilson de 0,72 a 1,00), y la calibración de riesgo de camino equivocado con $b=0$ iguala su forma cerrada, -3,401197382, a nueve decimales.
