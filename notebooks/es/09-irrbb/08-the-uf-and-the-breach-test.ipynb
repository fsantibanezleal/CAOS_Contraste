La investigación propone la prueba ("una prueba de validación de Contraste debería entonces marcar todo cambio en los parámetros de comportamiento de los depósitos que coincida con un incumplimiento de límite") y no cita ninguna herramienta que la implemente.

## Evidencia

### La UF como moneda propia

**Lo que agrega la afirmación.** El Anexo 1 del capítulo 21-13 trata el riesgo de reajustabilidad como una moneda separada con sus propios choques, su propio piso, el cargo por reajustabilidad y la compensación de 50 por ciento; el motor lo implementa, y las variantes versionadas de C15 se recalculan desde él. La primera celda muestra la separación en los documentos versionados: los tamaños y el piso de cada moneda y la curva en que se descuenta cada una. La segunda muestra por qué el piso debe ser el propio de la moneda: la curva UR es una curva de tasas reales, de 1,2 a 2,8 por ciento, y un choque a la baja de 200 pb empuja su extremo corto bajo cero, donde el piso del peso de 0 por ciento actuaría en nueve bandas y el piso UR de -2 por ciento en ninguna.

````python
# Los tamaños, el piso y la curva de cada moneda en los documentos versionados, y el tamaño del motor para cada una
out = variants["short-uf"]["outputs"]
rates = {"CLP": np.array(out["curves"]["clp"]["zero_rates_at_midpoints"])}
rates["UR"] = np.array(out["curves"]["ur"]["zero_rates_at_midpoints"])
floors = {"CLP": out["floors"]["other_floor"], "UR": out["floors"]["ur_floor"]}
print("tamaños EVE (par/corto/largo) | tamaño NII | piso | tasa banda 1 | tasa banda 19")
for committed in out["eve_shock_sizes"]:
    leg = "CLP" if committed["currency"] == "CLP" else "UR"
    nii = next(x["parallel_bp"] for x in out["nii_shock_sizes"] if x["currency"] == committed["currency"])
    z = shocks.shock_size(committed["currency"], "cmf-annex1")
    print(f"{leg:4s} {z.parallel_bp:.0f}/{z.short_bp:.0f}/{z.long_bp:.0f} | {nii:.0f} | {floors[leg]:.2f}")
    print(f"    {rates[leg][0]:.5f} | {rates[leg][18]:.5f}")
fig, ax = plt.subplots(figsize=(8, 4.2))
for leg, colour in (("CLP", "tab:blue"), ("UR", "tab:green")):
    ax.plot(MID, rates[leg] * 100, color=colour, label=leg)
    ax.axhline(floors[leg] * 100, color=colour, linestyle=":", label="piso " + leg)
ax.set_xlabel("punto medio de la banda (años)")
ax.set_ylabel("tasa cero (%)")
ax.set_title("Las dos monedas: curva y piso")
ax.legend(fontsize=8)
plt.show()
````

Las dos líneas continuas son las curvas en que se descuenta cada moneda, la curva en pesos entre cerca de 4,5 y 5,9 por ciento y la curva real en UF entre 1,2 y 2,8 por ciento, y las líneas punteadas son los pisos de cada moneda: 0 por ciento para el peso y -2 por ciento para la UR. Los tamaños impresos arriba también difieren, 300/400/200 pb para el peso y 200/250/150 pb para la UR en el EVE, y los mismos 300 pb en el NII. Todo lo que el Anexo 1 separa se ve aquí: una curva, un conjunto de tamaños y un piso por moneda, que un tratamiento de una sola moneda reemplazaría por los del peso. La figura muestra los insumos de la afirmación; el paso siguiente muestra qué hacen con un resultado.

Ahora el piso. La celda siguiente aplica una caída de 200 pb a la curva UR de la primera variante con el piso del peso y luego con el piso UR, y lista las bandas donde cada piso actúa.

````python
# Aplicar una baja de 200 pb a la curva UR con el piso en pesos y con el piso UR y ver dónde actúa cada uno
shape = np.full(19, -200.0)
raw = shocks.apply_floor(rates["UR"], shape, None)
fig, ax = plt.subplots(figsize=(8, 4.2))
ax.plot(MID, rates["UR"] * 100, label="curva UR")
ax.plot(MID, raw * 100, linestyle="--", label="curva UR tras -200 pb, sin piso")
for label, floor in (("piso en pesos (0)", 0.0), ("piso UR (-2 por ciento)", -0.02)):
    floored = shocks.apply_floor(rates["UR"], shape, floor)
    bound = [int(k) + 1 for k in np.flatnonzero(np.abs(floored - raw) > 1e-12)]
    print(f"{label:28s} bandas donde actúa: {bound}")
    ax.plot(MID, floored * 100, label=label)
print("tasa más baja de la curva UR tras -200 pb, antes del piso:", round(float(raw.min()), 5))
ax.set_xlabel("punto medio de la banda (años)")
ax.set_ylabel("tasa cero (%)")
ax.set_title("Una baja de 200 pb sobre la curva UR")
ax.legend(fontsize=8)
plt.show()
````

La línea de trazos es la curva real tras una baja de 200 pb, con su extremo corto en -0,795 por ciento. El piso en pesos devuelve las nueve primeras bandas a cero, la línea que queda sobre el eje, y así cambiaría una curva de tasas reales que no tiene razón para ser no negativa; el piso UR de -2 por ciento queda bajo toda la curva con choque y no actúa en ninguna banda. Por eso el piso tiene que ser el propio de la moneda: el piso en pesos aplicado a la UF fabricaría una pérdida en una moneda cuyas tasas eran bajas pero legítimamente positivas al comienzo y negativas tras el choque. La figura muestra un choque sobre una curva en una fecha y no muestra que alguno de los pisos actúe en las variantes versionadas, lo que el límite de la afirmación sobre la UF más abajo dice con claridad.

**La compensación y el cargo.** La regla de compensación da la mitad del crédito al tramo con ganancia cuando los dos deltas tienen signos opuestos. La variante `offset-on-off` guarda el par con y sin la compensación, y las mismas funciones del motor dan las medidas del Anexo de ambas formas para las cuatro variantes. El cargo se suma una vez al delta NII sea cual sea la compensación. La celda siguiente imprime ambas medidas con y sin la compensación, la proporción del delta NII que es el cargo, y la mayor cercanía de alguna tasa posterior al choque a su piso, y dibuja el efecto de la compensación.

````python
# Las medidas del Anexo con y sin compensación, el cargo como proporción del delta NII y el piso más cercano
annex = {}
for v, d in variants.items():
    o = d["outputs"]
    d_clp, d_ur = (np.array([r["pair"][k] for r in o["eve"]["rows"]]) for k in ("delta_clp", "delta_ur"))
    n_clp, n_ur = (np.array([r["pair"][k] for r in o["nii"]["rows"]]) for k in ("delta_clp", "delta_ur"))
    eve_on, eve_off = (eve_nii.annex1_delta_eve(d_clp, d_ur, flag) for flag in (True, False))
    pn = o["indexation_charge"]["pn_ur"]
    nii_on, nii_off = (eve_nii.annex1_delta_nii(n_clp, n_ur, pn, flag) for flag in (True, False))
    closest = min(x["distance_bp"] for x in o["floors"]["distances"])
    annex[v] = (eve_on.delta_eve, nii_on.delta_nii, eve_off.delta_eve, nii_off.delta_nii)
    print(f"{v:14s} ΔEVE {eve_on.delta_eve:10,.0f} {eve_off.delta_eve:10,.0f}")
    print(f"    ΔNII {nii_on.delta_nii:10,.0f} {nii_off.delta_nii:10,.0f}")
    print(f"    cargo / ΔNII {nii_on.indexation_charge / nii_on.delta_nii:.4f} | piso más cercano (pb) {closest:.2f}")
x = np.arange(len(annex))
fig, (left, right) = plt.subplots(1, 2, figsize=(11, 4))
panels = ((left, (0, 2), "delta EVE (millones de CLP)"), (right, (1, 3), "delta NII (millones de CLP)"))
for ax, (k_on, k_off), ylabel in panels:
    ax.bar(x - 0.2, [annex[v][k_on] for v in annex], 0.4, label="con compensación")
    ax.bar(x + 0.2, [annex[v][k_off] for v in annex], 0.4, label="sin compensación")
    ax.set_xticks(x, list(annex), rotation=20, ha="right", fontsize=8)
    ax.set_ylabel(ylabel)
left.set_title("Efecto de la compensación sobre las medidas del Anexo 1")
left.legend(fontsize=8)
fig.tight_layout()
plt.show()
````

La compensación actúa sobre todo en el NII. En la variante short-uf baja el delta NII de 14.597 a 11.067, 24 por ciento, y en la variante offset-on-off baja el delta EVE de 94.721 a 81.173, 14 por ciento; en la variante short-uf no cambia el delta EVE porque el peor choque tiene a ambos tramos perdiendo. El cargo es 55,9, 55,1 y 56,1 por ciento del delta NII en las tres variantes con una posición neta en UF de 30 por ciento del Tier 1, y 15,6 por ciento en la variante con 5 por ciento. La mayor cercanía de alguna tasa con choque a su piso es 20,48 pb en las cuatro. La figura muestra el efecto de una regla sobre cuatro variantes sintéticas, no una afirmación sobre ningún banco real.

**El contrafactual: lo que pierde un tratamiento de una sola moneda.** Sin la UR, un banco pondría sus flujos en UF, convertidos a pesos, con los flujos en pesos, los descontaría todos con la curva en pesos, los sometería a los choques con los tamaños del peso y el piso en cero, y no sumaría ningún cargo: una sola moneda. La celda hace exactamente eso con el motor para las cuatro variantes, usando los flujos netos versionados de cada tramo, y compara las dos medidas y las razones al Tier 1. Es un contrafactual, no un método rival: muestra lo que se pierde cuando la UF no es una moneda propia. La diferencia tiene cuatro partes que actúan juntas, la curva (nominal en vez de real), los tamaños (los del peso para los flujos en UF), la compensación total de los tramos dentro de una sola suma, y el cargo ausente.

````python
# Medidas del Anexo 1 frente a una sola moneda con los mismos flujos: curva, tamaños y piso en pesos, sin cargo
peso = shocks.shock_size("CLP", "cmf-annex1")
single = {}
for v, d in variants.items():
    o = d["outputs"]
    uf = o["ur_currency"]["uf_value"]
    rows = {leg["leg"]: [r for item in leg["items"] for r in item["cash_flows"]] for leg in o["legs"]}
    vec = {k: np.bincount([r["bucket"] - 1 for r in rs], [r["amount"] for r in rs], 19) for k, rs in rows.items()}
    total = vec["CLP"] + uf * vec["UR"]
    r_peso = np.array(o["curves"]["clp"]["zero_rates_at_midpoints"])
    loss_eve = []
    for row in o["eve"]["rows"]:
        net = np.array(row["clp"]["net_cash_flows"]) + np.array(row["ur"]["net_cash_flows"])
        shape = shocks.shock_curve(row["shock"], peso.parallel_bp, peso.short_bp, peso.long_bp, MID)
        shocked = shocks.apply_floor(r_peso, shape, 0.0)
        loss_eve.append(eve_nii.delta_eve(total, net, r_peso, shocked, MID))
    flat = [shocks.apply_floor(r_peso, np.full(19, bp), 0.0) for bp in (300.0, -300.0)]
    loss_nii = [eve_nii.delta_nii(total, r_peso, shocked, MID) for shocked in flat]
    single[v] = (max(0.0, max(loss_eve)), max(0.0, max(loss_nii)))
    print(f"{v:14s} ΔEVE Anexo 1 {annex[v][0]:10,.0f} | una moneda {single[v][0]:10,.0f}")
    print(f"    ΔNII Anexo 1 {annex[v][1]:10,.0f} | una moneda {single[v][1]:10,.0f}")
````

La celda siguiente dibuja las cuatro medidas de cada variante como proporción del Tier 1, con la línea de 15 por ciento del EVE y la de 5 por ciento del NII.

````python
# Dibujar las medidas del Anexo 1 y el contrafactual de una sola moneda como proporción del Tier 1
tier = {v: d["outputs"]["thresholds"]["tier1"] for v, d in variants.items()}
series = []
specs = [("ΔEVE Anexo 1", annex, 0), ("ΔEVE una moneda", single, 0)]
specs += [("ΔNII Anexo 1", annex, 1), ("ΔNII una moneda", single, 1)]
for label, source, k in specs:
    series.append((label, [source[v][k] / tier[v] for v in variants]))
x = np.arange(len(variants))
fig, ax = plt.subplots(figsize=(9, 4.2))
for k, (label, values) in enumerate(series):
    ax.bar(x + (k - 1.5) * 0.2, [100 * value for value in values], 0.2, label=label)
ax.axhline(15, color="tab:red", linestyle="--", linewidth=0.8)
ax.axhline(5, color="tab:orange", linestyle="--", linewidth=0.8)
ax.set_xticks(x, list(variants), rotation=20, ha="right", fontsize=8)
ax.set_ylabel("porcentaje del Tier 1")
ax.set_title("La UF como moneda frente a una sola moneda")
ax.legend(fontsize=7)
plt.show()
````

Tratar los flujos en UF como pesos da un delta NII de 1.356, 1.648 y 1.254 para las variantes short-uf, long-uf y offset-on-off frente a los valores del Anexo 1 de 11.067, 11.213 y 11.016: entre 6,8 y 8,8 veces menor, porque la suma de una sola moneda no tiene cargo y deja que los tramos se neteen por completo. El delta EVE es 64.213, 84.710 y 59.422 frente a 79.643, 93.816 y 81.173: entre 10 y 27 por ciento menor. En la variante floors-active, con una posición neta en UF de 5 por ciento del Tier 1, la brecha es menor, 4.653 frente a 6.606 para el NII. En estas cuatro variantes ninguna razón cruza su línea con ninguno de los dos tratamientos, así que la afirmación no es que un banco aprobaría o fallaría distinto sino que la medida que define el regulador no es la que da un cálculo de una sola moneda.

**El límite de la afirmación sobre la UF.** La evidencia muestra que el motor implementa el Anexo 1 como una moneda separada y que las variantes versionadas se recalculan desde él. No muestra que el Anexo se aplique como lo aplica la CMF: no hay una segunda implementación con la cual reconciliar, y la búsqueda del dossier no encontró ninguna (NO VERIFICADO más allá de esa búsqueda). Los pisos están implementados y probados, pero nunca actúan sobre las curvas versionadas, siendo el más cercano uno a 20,48 pb, así que su efecto sobre un número no queda demostrado con estos datos. El banco es sintético. Y la afirmación es regional: el estándar no tiene una unidad como la UF, y solo el Anexo chileno la necesita aquí.

### Una prueba que marca un supuesto de comportamiento cambiado ante un incumplimiento de límite

**Lo que agrega la afirmación.** `irrbb.behavioural_change_at_breach` recibe la serie de un parámetro de comportamiento, las fechas y una marca de incumplimiento por fecha, y cuenta las fechas en que el parámetro cambió y la marca es verdadera. Es roja con uno o más, verde con ninguno, y `not_evaluated` con menos de dos fechas. La primera celda muestra que la prueba discrimina, sobre cinco historiales construidos cuya respuesta es conocida: un parámetro mantenido fijo durante un incumplimiento, un parámetro que cambia sin incumplimiento, un parámetro cambiado ante un incumplimiento, un parámetro cambiado antes del incumplimiento pero no en él, y una sola fecha. Son números construidos que ejercitan la regla, no datos de ningún banco.

### Sensibilidad y límites: dónde cambia o falla el método

La prueba que convierte el hallazgo de la Reserva Federal en un conteo compara cada valor de un parámetro de comportamiento con el anterior por desigualdad exacta (Junta de Gobernadores del Sistema de la Reserva Federal, 2023, https://www.federalreserve.gov/publications/files/svb-review-20230428.pdf). Con $v_t$ el valor del parámetro en la fecha $t$ y $b_t$ igual a 1 cuando se incumplió un límite en la ventana que termina en $t$, cuenta

$$N = \sum_{t=2}^{T} \mathbf{1}\,[\,v_t \ne v_{t-1}\,]\;b_t$$

y la luz es roja cuando $N \ge 1$. Como la comparación es exacta, el veredicto depende de la resolución con que se guarda el parámetro, y de cualquier redondeo $\delta$ aplicado antes de la prueba, que cambia el conteo a

$$N_{\delta} = \sum_{t=2}^{T} \mathbf{1}\,[\,\tilde v_t \ne \tilde v_{t-1}\,]\;b_t, \qquad \tilde v_t = \delta\,\operatorname{round}\!\left(\frac{v_t}{\delta}\right)$$

donde $\delta$ es el paso de redondeo en la unidad del parámetro (años aquí) y $\operatorname{round}$ el entero más cercano. La primera celda suma un cambio de tamaño creciente a un parámetro mantenido en 4,0 años y leído en tres fechas con un incumplimiento en la última, de 10^-16 a 1 año, y luego repite una historia de tres fechas con incumplimiento en todas con un parámetro que solo lleva ruido numérico de desviación estándar creciente, 1.000 historias por valor. La segunda celda redondea el plazo ajustado del modelo de depósitos de C14 a pasos de 0,01 a 5 años antes de la prueba y dibuja ambos barridos.

````python
# Barrer el tamaño de un cambio en un incumplimiento y el ruido de un parámetro reestimado
days3 = np.array(["2021-12-31", "2022-06-30", "2022-12-31"], dtype="datetime64[D]")
flag3 = np.array([False, False, True])
print("cambio sumado en la tercera fecha | luz | cambios marcados")
for eps in (0.0, 1e-16, 1e-15, 1e-12, 1e-9, 1e-6, 1e-3, 0.1, 1.0):
    run = rv_irrbb.behavioural_change_at_breach(np.array([4.0, 4.0, 4.0 + eps]), days3, flag3)
    print(f"{eps:8.0e} | {LIGHT[run.light.value]:5s} | {run.statistic:.0f}")
noise_rng = np.random.default_rng(11)
all_flags = np.array([True, True, True])
red_share = {}
print("ruido de la reestimación (de) | proporción de 1.000 historias en rojo")
for sd in (0.0, 1e-16, 1e-14, 1e-9, 1e-4):
    reds = 0
    for _ in range(1000):
        values = 4.0 + sd * noise_rng.standard_normal(3)
        reds += rv_irrbb.behavioural_change_at_breach(values, days3, all_flags).light.value == "red"
    red_share[sd] = reds / 1000
    print(f"{sd:8.0e} | {red_share[sd]:.3f}")
````

La primera tabla dice que la prueba no lee el tamaño de un cambio. Un cambio de 10^-16 años es menor que el espaciado de los números de punto flotante alrededor de 4,0 (unos 8,9 x 10^-16), de modo que la serie no cambia y la luz es verde con cero cambios marcados; de 10^-15 a un año completo la luz es roja con un cambio marcado, igual para un cambio de un femtoaño y de doce meses. La segunda tabla es la consecuencia para un parámetro ajustado: sin ruido ninguna historia es roja, con un ruido de 10^-16 una proporción de 0,039 de las historias es roja, y de 10^-14 a 10^-4 las 1.000 son rojas. Un parámetro que se reestima en cada fecha, aunque sea hasta el último dígito, siempre se marca bajo un incumplimiento. La celda siguiente redondea el plazo ajustado del modelo de depósitos comprometido antes de la prueba.

````python
# Redondear el plazo ajustado a una malla antes de la prueba y dibujar ambos barridos
steps = (0.0, 0.01, 0.1, 0.25, 0.5, 1.0, 2.0, 5.0)
flags_c14 = np.array([bool(d["outputs"]["scenarios"][0]["outlier"]["breached"]) for d in docs.values()])
dates_c14 = np.array([d["outputs"]["valuation_date"] for d in docs.values()], dtype="datetime64[D]")
by_date = {m["valuation_date"]: m["weighted_average_maturity_years"] for m in fit["nmd_models"]}
raw = np.array([by_date[str(day)] for day in dates_c14])
rounded = {}
print("paso de redondeo (a) | plazos redondeados | luz | cambios marcados")
for step in steps:
    values = raw if step == 0.0 else np.round(raw / step) * step
    run = rv_irrbb.behavioural_change_at_breach(values, dates_c14, flags_c14)
    rounded[step] = (values, run.statistic)
    print(f"{step:5.2f} | {np.round(values, 3)} | {LIGHT[run.light.value]:5s} | {run.statistic:.0f}")
fig, (left, right) = plt.subplots(1, 2, figsize=(11, 4.2))
left.semilogx([max(s, 1e-17) for s in red_share], list(red_share.values()), marker="o", color="tab:red")
left.set_xlabel("ruido de la reestimación (desviación estándar)")
left.set_ylabel("proporción de historias en rojo")
left.set_title("El ruido solo enciende la luz")
for step in (0.0, 0.5, 2.0, 5.0):
    right.plot(dates_c14, rounded[step][0], marker="o", label=f"paso {step:g}")
right.set_ylabel("plazo redondeado (años)")
right.set_title("El veredicto depende de la resolución")
right.legend(fontsize=8)
fig.tight_layout()
plt.show()
````

La tabla de redondeo muestra el veredicto como función de la resolución. Los plazos ajustados 4,759, 4,205 y 5,0 dan dos cambios marcados y luz roja sin filtrar y con pasos de 0,01, 0,1 y 0,25 años; con pasos de 0,5 y 1 año pasan a 5, 4 y 5 y la luz sigue roja con dos cambios marcados; con un paso de 2 años los tres valores pasan a 4,0 y con un paso de 5 años todos pasan a 5,0, y la luz se pone verde con ninguno. En el panel izquierdo la proporción en rojo es cero con ruido cero, cercana a cero en 10^-16 y uno desde 10^-14, en un eje logarítmico; en el panel derecho las líneas de pasos de 0 y 0,5 años se mueven entre las fechas y las de pasos de 2 y 5 años son planas.

El método falla donde el parámetro es uno ajustado. La comparación estricta se construyó para un parámetro que un banco declara y mantiene, una proporción estable o un plazo estable, y ahí separa un valor mantenido de uno cambiado, como mostraron las historias construidas. Sobre un parámetro que se reestima en cada fecha marca el patrón por construcción, sea cual sea el tamaño del movimiento, y un redondeo que silenciaría el ruido silencia también un cambio real de dos años: el paso grueso oculta justamente el cambio que la prueba existe para encontrar. El rojo de la serie de C14 es entonces una invitación a preguntar si alguien eligió el número, lo que los datos de este notebook no pueden responder.

La primera celda siguiente arma un banco propio con una pierna en pesos y una pierna en UF, aplica choques a cada pierna con su propia tabla y piso, y compara el valor del Anexo 1 con una lectura que pone todos los flujos en pesos y les aplica los tamaños y el piso del peso.
````python
# Un banco propio con pierna en UF: el valor del Anexo 1 en dos monedas frente a una lectura en una moneda
asset_o = {"CLP": ([0.5, 2.0, 6.0], [300.0, 400.0, 200.0]), "UR": ([3.0, 9.0], [0.010, 0.005])}
liab_o = {"CLP": ([0.1, 1.0], [500.0, 250.0]), "UR": ([0.2, 4.0], [0.0075, 0.006])}
uf_o = 40000.0
net_o = {}
for leg in asset_o:
    a_flow = slotting.slot_cash_flows(np.array(asset_o[leg][0]), np.array(asset_o[leg][1]))
    l_flow = slotting.slot_cash_flows(np.array(liab_o[leg][0]), np.array(liab_o[leg][1]))
    net_o[leg] = (a_flow - l_flow) * (uf_o if leg == "UR" else 1.0)
size_o = {"CLP": shocks.shock_size("CLP", "cmf-annex1"), "UR": shocks.shock_size("CLF", "cmf-annex1")}
d_o = {"CLP": [], "UR": []}
total_o = net_o["CLP"] + net_o["UR"]
single_o = []
for s in shocks.SHOCKS:
    for leg in d_o:
        z = size_o[leg]
        shape = shocks.shock_curve(s, z.parallel_bp, z.short_bp, z.long_bp, MID)
        shaped = shocks.apply_floor(rates[leg], shape, floors[leg])
        d_o[leg].append(eve_nii.delta_eve(net_o[leg], net_o[leg], rates[leg], shaped, MID))
    z = size_o["CLP"]
    peso = shocks.apply_floor(rates["CLP"], shocks.shock_curve(s, z.parallel_bp, z.short_bp, z.long_bp, MID), 0.0)
    single_o.append(eve_nii.delta_eve(total_o, total_o, rates["CLP"], peso, MID))
    print(f"{s:14s} CLP {d_o['CLP'][-1]:8.2f} | UR {d_o['UR'][-1]:8.2f} | una moneda {single_o[-1]:8.2f}")
annex_o = eve_nii.annex1_delta_eve(np.array(d_o["CLP"]), np.array(d_o["UR"]), True)
print("delta EVE del Anexo 1 (dos monedas):", round(annex_o.delta_eve, 2), "|", annex_o.worst_shock)
print("lectura en una moneda, mayor pérdida:", round(max(0.0, max(single_o)), 2))
````

- **La afirmación sobre la prueba de exceso, la discriminación.** En las cinco historias construidas la prueba da el veredicto que dice la regla: verde para un parámetro fijo durante un exceso, verde para un cambio sin exceso, verde para un cambio antes del exceso, roja para un cambio en el exceso (marcando 2022-12-31), y no evaluada para una sola fecha.
- **La afirmación sobre la prueba de exceso, el banco versionado.** La proporción estable pedida (0,6) y el plazo estable pedido (5,0 años) no cambian y la prueba es verde; el plazo promedio ajustado se mueve, 4,7595, 4,2049 y 5,0 años, y la prueba es roja en 2022-06-30 y 2022-12-31. Con las marcas de exceso en falso la misma serie es verde.
- **La afirmación sobre la prueba de exceso, el límite.** La prueba de valores atípicos se incumple en las tres fechas, así que la marca nunca varía, y el parámetro que se mueve es uno ajustado que se reestima en cada fecha, así que difiere por construcción. El rojo no es entonces un hallazgo de que alguien ajustó el plazo para aliviar un exceso; muestra que la prueba opera sobre esos parámetros y lo haría sobre un parámetro que un gerente cambió. Qué hace un validador con un rojo: pregunta quién movió el parámetro, cuándo y con qué aprobación; compara el valor nuevo con los datos que lo respaldan (el notebook 05 muestra que los pesos ajustados no son estables entre ventanas); y lee el EVE con y sin conducta en la misma fecha. El comportamiento del banco es sintético.
- **Lo que el notebook no muestra.** Una reconciliación de las cifras chilenas con una segunda implementación, el efecto de un piso sobre un número, y un caso en que un gerente cambió un supuesto declarado ante un exceso: los datos versionados no traen ninguno.

## Resumen

- La afirmación sobre la UF: el Anexo chileno trata la UF como una moneda propia; el motor implementa sus tamaños, piso, compensación y cargo, y las variantes versionadas se recalculan desde él. Un tratamiento de una sola moneda con los mismos flujos daría un delta NII muy distinto, porque el cargo y el emparejamiento son lo que hace la medida de NII.
- La afirmación sobre la UF tiene límites: ninguna segunda implementación con la cual reconciliar, pisos que nunca actúan en estas curvas, un banco sintético y una afirmación regional.
- La afirmación sobre la prueba de exceso: la prueba de exceso es roja cuando un parámetro cambia en una fecha con exceso y verde en otro caso, y el veredicto lo decide la marca de exceso.
- La afirmación sobre la prueba de exceso tiene límites: en el banco versionado toda fecha es un exceso y el parámetro que se mueve es uno ajustado, así que el rojo muestra que la prueba opera, no que el banco hiciera lo que encontró la revisión del banco quebrado. Sus tasas de error las mide el artículo de la prueba de supuestos de comportamiento sobre bancos simulados, no este notebook: con un parámetro declarado marca 100,0 % de las historias con un cambio en una fecha de exceso y 17,0 a 66,3 % de las historias con cambios no relacionados (la predicción de tamaño de 5 % se refuta); con un parámetro reestimado marca 82,2 % de los bancos sin nada plantado y ordena un alargamiento de 2 años con un área ROC de 0,579, y solo una tolerancia calibrada al 5 % (3,9 % de falsas marcas, área 0,937, 18,0 % de los alargamientos de 2 años y 99,5 % de los de 4 años detectados) repara el tamaño, que luego sube a 5,5 % con otra prevalencia de excesos.
- Ambas afirmaciones se declaran en el SDD solo hasta donde llega esta evidencia.
