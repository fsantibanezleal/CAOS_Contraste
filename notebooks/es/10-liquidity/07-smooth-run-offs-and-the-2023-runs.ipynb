## Datos

````python
# Cargar los cuatro documentos versionados (el notebook está tres carpetas bajo la raíz del repositorio)
ROOT = Path("../../..").resolve()
documents = {}
for name in ("list-adverse", "list-extreme", "fast-run", "lcr-standard"):
    path = ROOT / "data" / "derived" / "C16" / f"{name}.json"
    documents[name] = json.loads(path.read_text(encoding="utf-8"))
    baked = documents[name]["provenance"]["riskvalidation_version"]
    print(name, "| calculado con riskvalidation", baked, "| instalado", rv.__version__)

variants = {name: documents[name] for name in ("list-adverse", "list-extreme", "fast-run")}
````

## Método

### Paso 1: el período de supervivencia de las tres variantes

La comparación necesita una medida que todas las variantes compartan: el período de supervivencia, el primer día en que la posición neta de liquidez se vuelve negativa, con "sobrevive el horizonte" cuando nunca lo hace (notebook 01). Las tres variantes difieren en la velocidad de la fuga: LiST Adverso y Extremo usan las tasas diarias suaves de las tablas del BCE, y la corrida rápida pone en el segmento corporativo no asegurado la tasa calibrada con las corridas anteriores a 2023, en un horizonte agudo de 5 días. La celda siguiente calcula el período de supervivencia y la NLP mínima de cada variante a partir de su serie NLP versionada con el motor.

````python
# El período de supervivencia y la NLP mínima de cada variante, desde su serie NLP versionada, con el motor
from riskvalidation.liquidity.ladder import minimum_nlp, survival_period

print(f"{'variante':14s}{'horizonte':>10s}{'supervivencia':>24s}{'NLP mínima (mil mill. USD)':>29s}{'día':>6s}")
for name, document in variants.items():
    nlp = np.array(document["outputs"]["path"]["ladder"]["nlp"])
    result = survival_period(nlp)
    low, low_day = minimum_nlp(nlp)
    survival_text = "sobrevive el horizonte" if result.survives else f"día {result.day}"
    print(f"{name:14s}{result.horizon:10d}{survival_text:>24s}{low / 1e6:29.2f}{low_day:6d}")
````

### Paso 2: la proporción de depósitos perdida al horizonte de cada corrida

El período de supervivencia depende del CBC y de las salidas a la vez. Para comparar la velocidad de la fuga con lo ocurrido usamos solo la salida: para una corrida que perdió una proporción $F_i$ de sus depósitos en $n_i$ días (fila $i$ de la Tabla A.1) la contraparte del modelo es la salida acumulada de los segmentos en esos mismos $n_i$ días, como proporción de la base de depósitos $B$:

$$\hat F_i = \frac{1}{B} \sum_{s} \sum_{d \le n_i} O_s(d), \qquad \text{error}_i = F_i - \hat F_i$$

Una fila cuyo horizonte excede la escalera de una variante no puede evaluarse en ella (la corrida rápida tiene 5 días). La celda siguiente calcula cada proporción modelada a partir del saldo inicial y las tasas diarias de los segmentos versionados con el motor, e imprime las filas del horizonte más corto al más largo. Un error positivo significa que el modelo pierde menos de lo que perdió la corrida.

````python
# La proporción modelada al horizonte de cada corrida de la Tabla A.1, para cada variante, desde las salidas del motor
from riskvalidation.liquidity.ladder import behavioural_outflows
from riskvalidation.liquidity.runoff import TABLE_A1

modelled = {}
for name, document in variants.items():
    segments = document["outputs"]["path"]["ladder"]["segments"]
    opening = np.array([s["opening_balance"] for s in segments])
    horizon = len(segments[0]["outflows"])
    rates = np.array([s["daily_rates"] or np.ones(horizon) for s in segments])
    base = document["outputs"]["scoring"]["model_deposit_base"]
    outflow = np.cumsum(behavioural_outflows(opening, rates).sum(axis=0))
    shares = {}
    for run in TABLE_A1:
        shares[run.row_id] = None if run.days > len(outflow) else outflow[run.days - 1] / base
    modelled[name] = shares

print(f"{'banco':16s}{'año':>6s}{'días':>6s}{'observada':>11s}{'Adverso':>10s}{'Extremo':>10s}{'rápida':>10s}")
for run in sorted(TABLE_A1, key=lambda r: (r.days, r.year)):
    cells = ["-" if modelled[n][run.row_id] is None else f"{modelled[n][run.row_id]:.4f}" for n in modelled]
    left = f"{run.bank:16s}{run.year:6d}{run.days:6d}{run.outflow_share:11.2f}"
    print(left + f"{cells[0]:>10s}{cells[1]:>10s}{cells[2]:>10s}")
````

La celda siguiente dibuja la proporción modelada frente a la proporción observada de cada corrida que una variante puede evaluar, en ejes logarítmicos: un punto sobre la diagonal es un modelo que pierde lo que perdió la corrida, un punto bajo ella un modelo que pierde menos.

````python
# Dibujar la proporción modelada frente a la observada de cada corrida, para las tres variantes
fig, ax = plt.subplots(figsize=(7, 4.5))
for (name, shares), color, marker in zip(modelled.items(), ("tab:blue", "tab:red", "tab:green"), ("o", "s", "^")):
    points = [(run.outflow_share, shares[run.row_id]) for run in TABLE_A1 if shares[run.row_id] is not None]
    ax.scatter([p[0] for p in points], [p[1] for p in points], color=color, marker=marker, label=name)
ax.plot([0.05, 1.0], [0.05, 1.0], color="black", linestyle="--", linewidth=0.8, label="modelada = observada")
ax.set_xscale("log")
ax.set_yscale("log")
ax.set_xlabel("proporción observada de la corrida (escala logarítmica)")
ax.set_ylabel("proporción modelada al mismo horizonte (escala logarítmica)")
ax.set_title("La velocidad de las corridas")
ax.legend(fontsize=8)
fig.tight_layout()
plt.show()
````

Los puntos de las corridas rápidas, las de uno y dos días a la derecha del gráfico, quedan bajo la diagonal, lo que significa que el modelo pierde menos de lo que perdió la corrida. Las tres variantes pierden 0,0854, 0,0887 y 0,1263 de la base de depósitos en las corridas de un día (SVB UK, 0,30, y Signature, 0,20) y 0,0915, 0,0980 y 0,1708 en los dos días de SVB (0,85). La corrida rápida, calibrada antes de 2023, es la que más se acerca y aun así llega a cerca de una quinta parte de la pérdida de SVB. Las corridas lentas de 2008 y Credit Suisse quedan sobre la diagonal para las variantes suaves, es decir, esas variantes las cubren, mientras Northern Rock (0,20 en 4 días, 0,2528 para la corrida rápida) solo la cubre la corrida rápida y First Republic (0,57 en 90 días) solo la variante Extrema. La figura compara proporciones al horizonte de cada corrida y no dice nada del CBC que se les opone.

## Comparación

Lo que encontró la investigación, con la fuente de cada punto. Solo se citan las fuentes que lista el SDD.

- **La prueba de tensión de liquidez de 2019 del BCE (LiST).** Define la posición neta de liquidez, el período de supervivencia y el efecto acantilado, y hace salir los depósitos de vencimiento abierto con tasas diarias constantes capitalizadas en días relevantes; en sus resultados 103 bancos informaron una mediana del período de supervivencia de 176 días calendario bajo el choque Adverso y de 122 días calendario bajo el Extremo, y los choques son solo idiosincráticos (BCE, 2019, nota metodológica y anexo de resultados).
- **Los principios de Basilea.** El Principio 10 pide a los bancos someter a tensión su liquidez en una variedad de escenarios de corto y largo plazo, específicos de la institución y de todo el mercado, individualmente y en combinación (BCBS, 2008).
- **La turbulencia de 2023.** El informe del BCBS documenta la velocidad de las corridas de 2023 frente a las anteriores en la Tabla A.1 (SVB 85% en 2 días; First Republic 57% en 90 días) y plantea como preguntas abiertas métricas como una capacidad de contrapeso a cinco días, el período de supervivencia y una métrica de liquidez sin supuestos de fuga (BCBS, 2023). La revisión de la Reserva Federal encontró salidas de más de 40 mil millones de USD el 9 de marzo, con la administración esperando 100 mil millones más el día siguiente, y que la prueba de tensión interna de liquidez del banco se apoyaba en referencias de pares que no eran comparables (Federal Reserve, 2023).
- **Teoría y corrida digital.** Diamond y Dybvig (1983) muestran que un contrato de depósito a la vista tiene un equilibrio bueno y un equilibrio de corrida, y que el seguro de depósitos elimina el segundo; la proporción no asegurada es entonces un motor del riesgo de corrida, no un modelo calibrado. Cookson et al. (2026) relacionan la exposición a las redes sociales con las pérdidas bursátiles de los bancos en la corrida sobre Silicon Valley Bank; el producto no lo modela.
- **Pérdidas no realizadas.** Valorar a mercado los activos de los bancos de EE. UU. tras el alza de tasas de 2022 da grandes pérdidas no realizadas, y estas reducen el CBC efectivo y pueden gatillar corridas (Jiang et al., 2024; notebook 03).
- **La escalera de modelos.** El dosier de investigación ordena los modelos de una tensión de liquidez desde una escalera contractual estática a una escalera conductual, un motor de escenarios, la calibración de velocidades de corrida con sensibilidades, y la tensión de liquidez inversa; no encontró un motor de código abierto maduro para toda la escalera (UNVERIFIED más allá de la búsqueda del dosier). El producto implementa los cinco peldaños; el precio de transferencia de fondos no está implementado.

Los tres pasos siguientes corren el motor sobre los datos del caso frente a estos puntos: la tercera severidad de LiST, que el producto no usa; la velocidad de SVB frente a las tablas y al LCR; y los períodos de supervivencia del caso frente a las medianas que el BCE informó para 103 bancos.

### Paso 3: la severidad Base de LiST

La prueba de 2019 tiene tres severidades, y el producto corre dos de ellas como variantes. El motor guarda la tercera, Base, con las mismas clases (BCE, 2019, Tabla 3.3). La celda siguiente la compone sobre los segmentos y la escalera del documento LiST Adverso y calcula su período de supervivencia y su NLP mínima.

````python
# La severidad Base de LiST sobre los segmentos y la escalera del documento LiST Adverso
from riskvalidation.liquidity.ladder import net_liquidity_position
from riskvalidation.liquidity.scenarios import compose_scenario, list_scenario

document = variants["list-adverse"]["outputs"]
lad = document["path"]["ladder"]
row = next(r for r in document["conventions"]["variant_scenarios"]["rows"] if r["variant"] == "list-adverse")
segments = {s["segment"]: (s["list_class"], s["opening_balance"]) for s in lad["segments"]}
args = (row["scenario_family"], row["phase"], list_scenario("baseline"))
scenario = compose_scenario(*args, segments, row["horizon_days"])
contractual = np.sum([c["amounts"] for c in lad["other_outflows"]], axis=0)
inflows = np.sum([c["amounts"] for c in lad["inflows"]], axis=0)
total_outflows = scenario.outflows().sum(axis=0) + contractual
baseline_nlp = net_liquidity_position(np.array(lad["cbc_total"]), total_outflows, inflows)
baseline = survival_period(baseline_nlp)
low, low_day = minimum_nlp(baseline_nlp)

print("sobrevive el horizonte:", "sí" if baseline.survives else "no", "| horizonte (días):", baseline.horizon)
print("NLP mínima (miles de millones de USD) y su día:", round(low / 1e6, 2), low_day)
````

Bajo Base el banco sobrevive todo el horizonte, y su NLP nunca cae bajo 22,06 mil millones de USD (el día 1, cuando el segmento financiero sale completo): las clases de vencimiento abierto de los segmentos tienen tasa cero, así que las únicas salidas son ese segmento y las contractuales. La severidad no es una tensión para este balance, y por eso el producto conserva Adverso y Extremo; es un piso de la familia de tablas, sin corrida alguna.

### Paso 4: cuánto más rápido fue 2023 que las tablas

Poner las corridas y las tablas en la misma escala toma la tasa diaria implícita del notebook 02 y la razón de una corrida a una tabla:

$$q = 1-(1-F)^{1/n}, \qquad r = \frac{q_{\text{run}}}{q_{\text{table}}}$$

donde $F$ es la proporción perdida en $n$ días y $r$ dice cuántas veces más rápido se fugó la corrida que la tabla. La tasa de Silicon Valley Bank (85% en 2 días, contando la salida que la administración esperaba para el 10 de marzo; Federal Reserve, 2023) frente a la tasa que LiST Adverso y Extremo dan a la clase corporativa no asegurada, y frente a la mayor salida de depósitos a 30 días del LCR (40% en 30 días, el tope del rango de la Tabla A.1), muestra la distancia. La celda siguiente calcula las tasas y las razones.

````python
# La tasa diaria implícita de SVB frente a las tasas de LiST de la clase corporativa y al tope del rango LCR
from riskvalidation.liquidity.runoff import LIST_PARAMETERS, TABLE_A1_LCR_RANGE, daily_rate_from_cumulative

svb = next(run for run in TABLE_A1 if run.row_id == "svb-2023")
corporate = "non-operational-non-financial-corporates"
rates = {severity: LIST_PARAMETERS[severity]["open_maturity"][corporate] for severity in ("adverse", "extreme")}
low_share, high_share, lcr_days = TABLE_A1_LCR_RANGE
rates["lcr top"] = daily_rate_from_cumulative(high_share, lcr_days)
rates["svb"] = daily_rate_from_cumulative(svb.outflow_share, svb.days)

print("tasa diaria implícita de SVB (85 por ciento en 2 días):", round(rates["svb"], 6))
for name, label in (("adverse", "Adverso"), ("extreme", "Extremo"), ("lcr top", "tope LCR")):
    print(f"{label:9s} | tasa diaria: {rates[name]:.8f} | SVB sobre ella: {rates['svb'] / rates[name]:.1f}")
lcr = documents["lcr-standard"]["outputs"]["lcr"]
print("LCR del documento:", round(lcr["ratio"], 4))
print("salida neta del LCR sobre su stock de HQLA:", round(lcr["net_outflows"] / lcr["hqla"]["hqla"], 4))
````

La celda siguiente dibuja las cuatro tasas diarias en un eje logarítmico: las dos tasas de LiST para la clase corporativa no asegurada, el tope del rango del LCR y la tasa de SVB.

````python
# Dibujar las tasas diarias de las tablas LiST, del rango LCR y de SVB en un eje logarítmico
labels = ["LiST Adverso", "LiST Extremo", "LCR, tope del rango", "SVB, 2023"]
values = [rates["adverse"], rates["extreme"], rates["lcr top"], rates["svb"]]
fig, ax = plt.subplots(figsize=(7, 4))
ax.bar(labels, values, color=["tab:blue", "tab:blue", "tab:green", "tab:red"])
ax.set_yscale("log")
ax.set_ylabel("tasa diaria implícita (escala logarítmica)")
ax.set_title("Qué tan rápido se fuga cada supuesto")
ax.tick_params(axis="x", labelrotation=15)
fig.tight_layout()
plt.show()
````

Las dos tasas de LiST y la tasa del LCR quedan a un factor de 2,5 entre sí, de 0,00688541 a 0,016883 al día, y la tasa de SVB es 36 veces la más alta de ellas. Las tasas de LiST para la clase corporativa no asegurada, 0,00688541 y 0,01065229 al día, están abajo; el tope del rango del LCR, 0,016883, es el siguiente escalón; y la tasa de SVB, 0,612702, es 89,0 veces la tasa Adversa, 57,5 veces la Extrema y 36,3 veces la del LCR. La salida neta del LCR es 2,1158 veces el stock de HQLA, así que incluso el más rápido de los supuestos regulatorios deja al banco corto, pero no a la velocidad de 2023. Las barras comparan tasas implícitas de una corrida cada una: SVB es una sola observación.

### Paso 5: los períodos de supervivencia frente a las medianas de la prueba de 2019

La celda siguiente pone los períodos de supervivencia del producto junto a las medianas de la prueba de 2019. Las medianas son cifras propias del BCE, ingresadas aquí como constantes con su fuente, porque no pertenecen a ningún documento versionado. Describen 103 bancos reales, y este caso es un banco de comportamiento sintético, así que la comparación muestra el orden de magnitud del método, no un ranking.

Un punto de unidad decide la comparación. El BCE informa la mediana del período de supervivencia en **días calendario**: el eje de su figura dice "Calendar days", y cada banco informa la fecha exacta de su período de supervivencia entre los días relevantes de LiST 2019 (BCE, 2019, anexo de resultados), lo que da 176 y 122. El período de supervivencia del producto cuenta **días relevantes** del calendario del documento, los días hábiles de EE.UU. desde 2023-01-03 con los feriados que lista el documento (una convención del producto, mientras que los días relevantes del BCE siguen el calendario TARGET2), y 125 días relevantes son 181 días calendario, los seis meses de la prueba. Los dos números no son comparables tal como están. La celda convierte el día de supervivencia de cada variante a los días calendario tras la fecha de valoración con `numpy.busday_offset` y el calendario del documento, convierte las medianas del BCE a días relevantes con `numpy.busday_count` como segunda lectura, comprueba que la conversión va y vuelve sobre 62, 41 y 125 días relevantes, y compara días calendario con días calendario.

````python
# Los períodos de supervivencia del producto junto a las medianas de la prueba del BCE de 2019, ambos en días calendario
# mediana del período de supervivencia de los 103 bancos en días calendario, anexo de resultados del BCE
LIST_2019_MEDIAN = {"adverse": 176, "extreme": 122}

calendar = variants["list-adverse"]["outputs"]["conventions"]["day_calendar"]
valuation = np.datetime64(variants["list-adverse"]["outputs"]["valuation_date"])
first_day, holidays = np.datetime64(calendar["first_day"]), calendar["holidays"]


def calendar_days_of(relevant_day):
    """Días calendario tras la fecha de valoración de un día relevante del calendario del documento."""
    date = np.busday_offset(first_day, relevant_day - 1, holidays=holidays)
    return int((date - valuation) / np.timedelta64(1, "D"))


def relevant_days_within(calendar_days):
    """Días relevantes del calendario del documento dentro de los días calendario dados."""
    end = valuation + np.timedelta64(calendar_days + 1, "D")
    return int(np.busday_count(first_day, end, holidays=holidays))


ours, ours_calendar = {}, {}
headers = ("severidad", "día relevante", "días calendario", "mediana, cal.", "razón", "mediana en días relevantes")
print(f"{headers[0]:10s}" + "".join(f"{h:>{w}s}" for h, w in zip(headers[1:], (16, 18, 16, 8, 28))))
for severity, name in (("adverse", "list-adverse"), ("extreme", "list-extreme")):
    nlp = np.array(variants[name]["outputs"]["path"]["ladder"]["nlp"])
    ours[severity] = survival_period(nlp).day
    ours_calendar[severity] = calendar_days_of(ours[severity])
    median = LIST_2019_MEDIAN[severity]
    ratio = ours_calendar[severity] / median
    row = f"{severity:10s}{ours[severity]:16d}{ours_calendar[severity]:18d}{median:16d}{ratio:8.2f}"
    print(row + f"{relevant_days_within(median):28d}")
trip = [relevant_days_within(calendar_days_of(day)) for day in (62, 41, 125)]
print("ida y vuelta, día relevante a días calendario y de vuelta (62, 41 y 125 días relevantes):", trip)
````

La celda siguiente dibuja los períodos de supervivencia de este caso frente a las medianas de la prueba de 2019.

````python
# Dibujar los períodos de supervivencia de este caso frente a las medianas de la prueba de 2019, en días calendario
positions = np.arange(2)
fig, ax = plt.subplots(figsize=(6.5, 4))
ax.bar(positions - 0.2, [ours_calendar["adverse"], ours_calendar["extreme"]], width=0.4, label="este caso")
medians = [LIST_2019_MEDIAN["adverse"], LIST_2019_MEDIAN["extreme"]]
ax.bar(positions + 0.2, medians, width=0.4, label="mediana de los 103 bancos, 2019")
ax.set_xticks(positions)
ax.set_xticklabels(["Adverso", "Extremo"])
ax.set_ylabel("período de supervivencia (días calendario)")
ax.set_title("Supervivencia, este caso frente a LiST 2019")
ax.legend(fontsize=8)
fig.tight_layout()
plt.show()
````

Comparado en una sola unidad, el caso dura cerca de la mitad de lo que informó el banco mediano de la muestra de 2019: 90 días calendario frente a 176 bajo Adverso y 61 frente a 122 bajo Extremo, razones de 0,51 y 0,50. Leídas en días relevantes del calendario del documento las medianas son 120 y 84 frente a 62 y 41. Poner 62 días relevantes frente a 176 días calendario daría 0,35, una razón que parece un tercio y es solo un desajuste de unidades. Las dos razones son casi iguales, así que la forma de la respuesta a la severidad es la misma aunque el nivel difiera. La causa es el balance, no el método: el caso tiene una gran base corporativa no asegurada y un CBC que cubre solo una parte, y los 103 bancos son balances reales con sus propios depósitos y activos. La conversión es aproximada a unos pocos días, porque el calendario del documento es una convención de días hábiles de EE.UU. y el del BCE es TARGET2. La figura dice que el mismo método da períodos de supervivencia del orden de meses; no dice que el banco quede bajo la muestra.

### Sensibilidad y límites del método

La comparación con la muestra del BCE tiene un insumo que un lector puede mover y que el caso no comparte con la muestra: el tamaño de la capacidad de contrapeso. Con la CBC escalada por $\kappa$ y los flujos sin cambio, la posición y el período de supervivencia son

$$\mathrm{NLP}_\kappa(t) = \kappa\,\mathrm{CBC}(t) - \sum_{s \le t}\bigl(O(s) - I(s)\bigr), \qquad \mathrm{SP}_\kappa = \min\{\, t : \mathrm{NLP}_\kappa(t) < 0 \,\}$$

y una posición que nunca se vuelve negativa dentro de la escalera está censurada por la derecha en el horizonte, 125 días relevantes o 181 días calendario. La celda siguiente barre $\kappa$ de 0,5 a 3 en pasos de 0,05 sobre las escaleras Adversa y Extrema con `net_liquidity_position` y `survival_period`, convierte cada día de supervivencia a días calendario con la función del paso 5, y encuentra el menor múltiplo con el que el caso alcanza la mediana de los 103 bancos.

````python
# Barrido del tamaño de la CBC: supervivencia en días calendario frente a las medianas de la prueba de 2019
cbc_multiples = np.round(np.arange(0.5, 3.01, 0.05), 2)
horizon_calendar = calendar_days_of(125)
sweep_days = {}
for severity, name in (("adverse", "list-adverse"), ("extreme", "list-extreme")):
    ladder = variants[name]["outputs"]["path"]["ladder"]
    cbc_base, flows_out = np.array(ladder["cbc_total"]), np.array(ladder["total_outflows"])
    flows_in = np.array(ladder["total_inflows"])
    days = []
    for multiple in cbc_multiples:
        outcome = survival_period(net_liquidity_position(multiple * cbc_base, flows_out, flows_in))
        days.append(None if outcome.survives else calendar_days_of(outcome.day))
    sweep_days[severity] = days

print("horizonte de la escalera en días calendario:", horizon_calendar)
print(f"{'múltiplo de la CBC':>20s}{'LiST Adverse':>26s}{'LiST Extreme':>26s}")
for multiple in (0.5, 1.0, 1.25, 1.5, 1.6, 1.7, 2.0, 3.0):
    cells = []
    for severity in ("adverse", "extreme"):
        value = sweep_days[severity][int(round((multiple - 0.5) / 0.05))]
        cells.append("sobrevive el horizonte" if value is None else str(value))
    print(f"{multiple:20.2f}{cells[0]:>26s}{cells[1]:>26s}")
for severity in ("adverse", "extreme"):
    median = LIST_2019_MEDIAN[severity]
    reached = [m for m, d in zip(cbc_multiples, sweep_days[severity]) if d is None or d >= median]
    print("menor múltiplo de la CBC de la grilla que alcanza la mediana:", severity, float(reached[0]))
````

La celda siguiente dibuja el período de supervivencia en días calendario frente al tamaño de la CBC para ambas severidades, con la mediana de la prueba de 2019 de cada una como línea segmentada del mismo color y el horizonte de la escalera punteado.

````python
# Dibujar el período de supervivencia en días calendario frente al tamaño de la CBC, con las medianas y el horizonte
fig, ax = plt.subplots(figsize=(8, 4.2))
for severity, color, label in (("adverse", "tab:blue", "LiST Adverso"), ("extreme", "tab:red", "LiST Extremo")):
    shown = [horizon_calendar if d is None else d for d in sweep_days[severity]]
    ax.plot(cbc_multiples, shown, color=color, label=label)
    ax.axhline(LIST_2019_MEDIAN[severity], color=color, linestyle="--", linewidth=0.9)
ax.axhline(horizon_calendar, color="gray", linestyle=":", label="horizonte de la escalera")
ax.set_xlabel("múltiplo de la CBC del documento")
ax.set_ylabel("período de supervivencia (días calendario)")
ax.set_title("Qué necesita el caso para igualar la mediana")
ax.legend(fontsize=8)
fig.tight_layout()
plt.show()
````

Ambas curvas suben casi linealmente y llegan a la línea de censura. Con el tamaño del documento el caso dura 90 días calendario bajo Adverso y 61 bajo Extremo; con la mitad de la CBC dura 32 y 23, y con 1,5 veces la CBC 167 y 115. El caso alcanza la mediana Adversa (176 días) con 1,55 veces su CBC y la mediana Extrema (122 días) con las mismas 1,55 veces: para igualar al banco mediano de la muestra de 2019 en ambas severidades este balance necesitaría cerca de la mitad más de activos líquidos. Esa es una lectura más útil de la brecha que la razón de 0,5, porque dice en qué se distingue el caso de la muestra.

La comparación falla en lo alto de la escala. Desde 1,6 veces la posición Adversa sobrevive toda la escalera, 181 días calendario, y la mediana Adversa de 176 días está a 5 días de ese horizonte (el anexo cuenta 51 de los 103 bancos sobre seis meses): la propia mediana del BCE está en el borde de lo que puede medir una prueba de seis meses, así que para la severidad Adversa ninguna CBC mayor puede distinguirse de la mediana. La curva Extrema sigue bajo el horizonte con 1,7 veces (143 días) y lo sobrevive con 2 veces, de modo que discrimina por más tiempo. El barrido escala todos los componentes de la CBC juntos y mantiene las salidas, así que no dice qué activos serían la capacidad faltante, y la conversión a días calendario usa el calendario del documento, de modo que el múltiplo de 1,55 vale a la resolución de la grilla y a unos pocos días.

### Úsalo con tus propios datos

Lo que necesita la comparación. Un documento de variante de tu propio banco con su escalera (`outputs.path.ladder`: la serie `nlp`, los `segments` con `opening_balance` y `daily_rates`, `cbc_total` y los flujos), su `outputs.scoring.model_deposit_base`, su `outputs.conventions.day_calendar` (`first_day` y los `holidays`) y su fecha de valoración; las corridas observadas como pares de proporción de salida y días (aquí las filas de 2023 de `TABLE_A1`); y las medianas publicadas, cada una con la unidad que usa quien la publica escrita al lado. El calendario es lo que sale mal: el motor cuenta días relevantes desde 1, y un día de supervivencia es una posición en tu calendario, no un número de días calendario.

La celda siguiente envuelve la comparación de este notebook en `variant_report(document, runs)`, que devuelve el período de supervivencia (`survival_period`) y la proporción que la variante modela al horizonte de cada corrida (`behavioural_outflows`, con `None` cuando la corrida es más larga que la escalera), y la corre sobre las tres variantes del documento con las cinco corridas de 2023. Imprime la fecha y los días calendario de cada supervivencia y luego prueba cuatro llamadas que deben fallar.

````python
# Informar la supervivencia, su fecha y las proporciones modeladas de las corridas para variantes de cualquier documento
def variant_report(document, runs):
    """Supervivencia, fecha y proporción modelada al horizonte de cada corrida para un documento de variante."""
    produced = document["outputs"]
    nlp = np.array(produced["path"]["ladder"]["nlp"])
    outcome = survival_period(nlp)
    segments = produced["path"]["ladder"]["segments"]
    opening = np.array([s["opening_balance"] for s in segments])
    horizon = len(segments[0]["outflows"])
    rates = np.array([s["daily_rates"] or np.ones(horizon) for s in segments])
    cumulative_outflow = np.cumsum(behavioural_outflows(opening, rates).sum(axis=0))
    base = produced["scoring"]["model_deposit_base"]
    shares = []
    for share, days in runs:
        daily_rate_from_cumulative(share, days)
        shares.append(None if days > horizon else float(cumulative_outflow[days - 1] / base))
    return outcome, shares


runs_2023 = [run for run in TABLE_A1 if run.year == 2023]
run_ids = [run.row_id for run in runs_2023]
my_runs = [(run.outflow_share, run.days) for run in runs_2023]
for name, document in variants.items():
    outcome, shares = variant_report(document, my_runs)
    if outcome.survives:
        text = "sobrevive el horizonte"
    else:
        date = np.busday_offset(first_day, outcome.day - 1, holidays=holidays)
        text = f"{outcome.day}, {date}, {calendar_days_of(outcome.day)}"
    print("variante", name, "|", "supervivencia (día relevante, fecha, días calendario):", text)
    shown = {run_id: None if s is None else round(s, 4) for run_id, s in zip(run_ids, shares)}
    print("   proporción modelada al horizonte de cada corrida:", shown)
````

La celda siguiente prueba las cuatro llamadas que deben fallar.

````python
# Cuatro llamadas que deben fallar: proporción sobre 1, corrida de 0 días, segmentos desiguales, valor faltante
adverse_document = variants["list-adverse"]
trials = {
    "una corrida con proporción sobre 1": lambda: variant_report(adverse_document, [(1.2, 2)]),
    "una corrida de 0 días": lambda: variant_report(adverse_document, [(0.5, 0)]),
    "segmentos y tasas que no coinciden": lambda: behavioural_outflows(np.array([1.0, 2.0]), np.ones((3, 5))),
    "una serie con un valor faltante": lambda: survival_period(np.array([1.0, np.nan])),
}
for label, trial in trials.items():
    try:
        trial()
    except ValueError as error:
        print("rechazado por el motor:", label, "->", error)
````

Qué revisar en la salida. La supervivencia de cada variante debe conciliar con el paso 1: día 62 para Adverso, que el calendario pone el 2023-03-31, 90 días calendario tras la fecha de valoración, y día 41 para Extremo, el 2023-03-02, 61 días; la corrida rápida sobrevive su horizonte de 5 días y no tiene fecha. Las proporciones son las del paso 2 en el orden de las filas de 2023 (SVB, First Republic, SVB UK, Credit Suisse, Signature): 0,0915 y 0,0854 en las corridas de dos y un día bajo Adverso, 0,4955 en las dos corridas de 90 días, y `None` en las corridas que la corrida rápida de 5 días no alcanza, de modo que una corrida más larga que la escalera nunca se extrapola. Luego los rechazos: una proporción sobre 1, una corrida de 0 días, segmentos y tasas que no coinciden en número y una serie con un valor faltante se rechazan con un mensaje que nombra el problema. Dos verificaciones son tuyas y el motor no las hace: que el calendario de tu documento sea aquel bajo el cual se contó la cifra publicada, y que una comparación con una mediana publicada declare su unidad, como lo hizo el paso 5; un día de supervivencia puesto frente a una mediana en otra unidad da una razón que parece un hallazgo y es un desajuste de unidades.
