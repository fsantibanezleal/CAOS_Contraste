### El cuantil conforme dividido contra MAPIE

Para $n$ puntajes de calibración $s_{(1)} \le \dots \le s_{(n)}$ el cuantil conforme exacto es el estadístico de orden $s_{(k)}$ con $k = \lceil (n+1)(1-\alpha) \rceil$ (notebook 04). El motor calcula $k$ sobre el decimal exacto de $\alpha$. La referencia es el paquete mismo, corrido: el `SplitConformalRegressor` de MAPIE 1.5.0 con un estimador ya ajustado que predice siempre 0, de modo que los residuos absolutos que conformaliza son los puntajes de calibración y el extremo superior de su intervalo alrededor de esa predicción es su cuantil. Junto a él se muestran otras dos reglas. La primera es la llamada de NumPy

$$\hat q_{\text{NumPy}} = \texttt{np.quantile}\bigl(s,\; (n+1)(1-\alpha)/n,\; \texttt{method="higher"}\bigr)$$

(el nivel acotado a 1), que una lectura anterior del código de MAPIE atribuyó al paquete y que la documentación del motor señala como no referencia: puede devolver $s_{(k+1)}$ donde la regla exacta devuelve $s_{(k)}$. La segunda es la regla ingenua en flotantes, $s_{(\lceil (n+1)(1-\alpha) \rceil)}$ con el producto tomado en flotantes.

La primera celda de abajo define las tres reglas de referencia como funciones; la segunda saca puntajes aleatorios del tamaño de calibración versionado (con semilla; el archivo versionado entrega solo el tamaño), aplica las cuatro reglas con $\alpha = 0{,}1$ e imprime el estadístico de orden que devuelve cada una. Luego recorre la grilla $n = 5, \dots, 399$ y $\alpha = 0{,}01, \dots, 0{,}99$ y cuenta los pares en que cada regla difiere del motor. MAPIE rechaza un par cuando falla su propia verificación, que su mensaje de error enuncia como que el número de puntajes sea menor que $1/\alpha$ o que $1/(1-\alpha)$; la celda cuenta esos rechazos aparte de las diferencias y comprueba que todo par que MAPIE rechaza y el motor responde cumple esa condición enunciada. La tolerancia es la igualdad exacta del puntaje devuelto: toda regla devuelve uno de los puntajes de calibración, así que una diferencia es un estadístico de orden distinto, nunca un redondeo.

````python
# Tres reglas de referencia como funciones de los puntajes y alfa: el paquete MAPIE, la llamada NumPy, la ingenua
def mapie_quantile(scores, alpha):
    """El extremo superior del intervalo que MAPIE 1.5.0 arma alrededor de una predicción de 0; None si se niega."""
    estimator = DummyRegressor(strategy="constant", constant=0.0).fit(np.zeros((2, 1)), np.zeros(2))
    splitter = SplitConformalRegressor(estimator=estimator, confidence_level=1 - alpha, prefit=True)
    splitter.conformalize(np.zeros((scores.size, 1)), scores)
    try:
        _, interval = splitter.predict_interval(np.zeros((1, 1)))
    except ValueError:
        return None
    return float(interval[0, 1, 0])


def numpy_rule(s, a):
    return np.quantile(s, min(1.0, (len(s) + 1) * (1 - a) / len(s)), method="higher")


def naive_rule(s, a):
    return s[min(math.ceil((len(s) + 1) * (1 - a)), len(s)) - 1]
````

Las tres funciones de arriba devuelven un estadístico de orden cada una. La celda siguiente saca los puntajes con semilla
del tamaño de calibración versionado, aplica el motor y las tres reglas con alfa = 0,1 e imprime el rango del puntaje que
devuelve cada una.

````python
# Las cuatro reglas con el tamaño de calibración versionado
rng = np.random.default_rng(3)
scores = np.sort(rng.random(n_cal))
alpha = 0.1
engine_q = conformal.split_conformal_quantile(scores, alpha)
mapie_q = mapie_quantile(scores, alpha)
numpy_q = numpy_rule(scores, alpha)
naive_q = naive_rule(scores, alpha)


def rank(q):
    return int(np.searchsorted(scores, q)) + 1


print(
    "n_cal", n_cal, "| rango del cuantil del motor", rank(engine_q), "| paquete MAPIE", rank(mapie_q), end=" "
)
print("| llamada higher de NumPy", rank(numpy_q), "| regla ingenua de flotantes", rank(naive_q))
````

Un par no es una comparación. La celda siguiente recorre 39.105 pares de tamaño de calibración (5 a 399) y alfa (0,01 a
0,99), cuenta cuántas veces cada regla devuelve otro estadístico de orden que el motor, y separa los pares en que MAPIE
rechaza la entrada de aquellos en que responde distinto.

````python
# Un barrido de cuántas veces cada regla devuelve otro estadístico de orden que el motor
count = collections.Counter()
for n in range(5, 400):
    s = np.sort(rng.random(n))
    for a in np.arange(1, 100) / 100:
        count["pairs"] += 1
        engine = conformal.split_conformal_quantile(s, float(a))
        reference = mapie_quantile(s, float(a))
        if math.isinf(engine):
            count["el motor no tiene cuantil finito"] += 1
            count["... y MAPIE también se niega"] += int(reference is None)
            continue
        count["la llamada higher de NumPy difiere"] += int(numpy_rule(s, a) != engine)
        count["la regla ingenua de flotantes difiere"] += int(naive_rule(s, a) != engine)
        if reference is None:
            count["MAPIE se niega, el motor responde"] += 1
            count["... y n es menor que 1/alfa o 1/(1-alfa)"] += int(n < max(1 / a, 1 / (1 - a)))
            continue
        count["comparados con el paquete MAPIE"] += 1
        if reference != engine:
            count["el paquete MAPIE difiere"] += 1
            count["... e iguala la regla ingenua de flotantes"] += int(reference == naive_rule(s, a))
            count["... y es el siguiente estadístico de orden hacia arriba"] += int(
                np.searchsorted(s, reference) == np.searchsorted(s, engine) + 1
            )
for key, value in count.items():
    print(f"{key:56s} {value:7d}")
print(
    "proporción de pares respondidos en que el paquete MAPIE difiere:",
    f"{count['el paquete MAPIE difiere'] / count['comparados con el paquete MAPIE']:.2%}",
)
````

### Las pruebas de dos tasas contra identidades independientes

Para dos grupos con $x_1$ de $n_1$ y $x_2$ de $n_2$, el estadístico z combinado al cuadrado es igual al estadístico chi-cuadrado de Pearson de la tabla de $2 \times 2$ **sin corrección de continuidad**, y su valor p es igual al de la prueba chi-cuadrado con un grado de libertad. La prueba exacta de Fisher es la prueba hipergeométrica del `fisher_exact` de SciPy. La celda siguiente comprueba ambas identidades en los cinco contrastes del atributo raza de HMDA versionado, y la diferencia de paridad contra la definición que usa la `demographic_parity_difference` de fairlearn para dos grupos, la mayor tasa de selección menos la menor.

````python
# Comprobar la prueba z contra el chi-cuadrado de Pearson, Fisher contra SciPy y la diferencia de paridad de dos grupos
race = hmda["outputs"]["attributes"][0]
groups = {g["group"]: (g["n_approved"], g["n"]) for g in race["groups"]}
x_ref, n_ref = groups[race["reference_group"]]
identity = {}
print(f"{'group':44s} {'z²':>10s} {'chi-cuadrado':>11s} {'p z':>10s} {'p chi-cuadrado':>13s}", end=" ")
print(f"{'p Fisher, motor':>17s} {'p Fisher, SciPy':>16s}")
for name, (x, n) in groups.items():
    if name == race["reference_group"]:
        continue
    z = inference.two_proportion_z_test(x, n, x_ref, n_ref)
    table = [[x, n - x], [x_ref, n_ref - x_ref]]
    chi2, p_chi2, _, _ = stats.chi2_contingency(table, correction=False)
    fisher = inference.fisher_exact_test(x, n, x_ref, n_ref)
    p_scipy = stats.fisher_exact(table).pvalue
    identity[name] = (z.p_value, p_chi2, fisher.p_value, p_scipy)
    print(f"{name:44s} {z.statistic**2:10.3f} {chi2:11.3f} {z.p_value:10.3e} {p_chi2:13.3e}", end=" ")
    print(f"{fisher.p_value:17.3e} {p_scipy:16.3e}")
edge_decision = np.r_[np.ones(60), np.zeros(40), np.ones(75), np.zeros(25)].astype(int)
edge_group = np.r_[np.repeat("A", 100), np.repeat("B", 100)]
rates = metrics.group_selection_rates(edge_decision, edge_group)
spread = max(r.rate for r in rates.values()) - min(r.rate for r in rates.values())
parity = abs(metrics.demographic_parity_difference(rates, "B")["A"])
print("diferencia de paridad para dos grupos, motor:", parity, "| mayor menos menor tasa:", round(spread, 12))
````

Cinco contrastes de dos pruebas son diez pares de valores p, y su concordancia es lo que se afirma. La celda siguiente
grafica cada valor p del motor contra el de referencia en ejes logarítmicos, con la diagonal donde serían idénticos.

````python
# Valores p del motor contra la implementación de referencia, prueba z y prueba de Fisher
values = list(identity.values())
z_engine, chi_scipy = [v[0] for v in values], [v[1] for v in values]
fisher_engine, fisher_scipy = [v[2] for v in values], [v[3] for v in values]
fig, ax = plt.subplots(figsize=(5.8, 4.6))
ax.loglog(chi_scipy, z_engine, "o", label="prueba z contra chi-cuadrado")
ax.loglog(fisher_scipy, fisher_engine, "s", label="prueba de Fisher contra SciPy")
ax.plot([1e-130, 1], [1e-130, 1], "k--", linewidth=1)
ax.set_xlabel("valor p de la referencia (SciPy)")
ax.set_ylabel("valor p del motor")
ax.legend(fontsize=8)
fig.tight_layout()
plt.show()
````

Todos los puntos quedan sobre la diagonal, a lo largo de 128 órdenes de magnitud: los valores p de la prueba z igualan a
los del chi-cuadrado de Pearson desde 2,857e-01 para el grupo asiático hasta 1,401e-128 para el grupo Black, y los
valores p de Fisher del motor igualan a los de SciPy, como 2,998e-120 para el grupo Black. El estadístico z al cuadrado
iguala al chi-cuadrado (581,966 para el grupo Black) hasta los dígitos impresos. La concordancia es una identidad entre
dos fórmulas y no una estimación independiente de la misma cantidad, así que verifica la implementación y no dice si la
prueba es la adecuada.

### Las métricas de tasa de selección contra fairlearn 0.15.0

fairlearn define la diferencia de paridad demográfica como la mayor menos la menor tasa de selección entre los grupos, y su razón como la menor sobre la mayor. El motor enuncia ambas contra un grupo de referencia elegido. Cuando el grupo de referencia tiene la tasa más alta, como White en el atributo raza de HMDA versionado, la mayor diferencia absoluta del motor es igual a la diferencia de fairlearn y la menor razón del motor es igual a la razón de fairlearn. La celda siguiente convierte los conteos versionados de los seis grupos de raza listados en una decisión por solicitante (aprobada o negada: los conteos fijan las decisiones salvo su orden, y las métricas no dependen del orden), corre sobre ellas el `MetricFrame` de fairlearn y sus dos funciones de paridad, e imprime el valor del motor junto a cada una con la diferencia absoluta. La tolerancia es $10^{-12}$: ambos lados dividen los mismos enteros, así que solo puede diferir el orden de las operaciones de coma flotante.

Las diferencias de probabilidades igualadas y de igualdad de oportunidad de fairlearn necesitan etiquetas verdaderas además de las decisiones, y el documento C21 versionado contiene conteos de aprobaciones por grupo y ningún resultado, así que esas dos no se reproducen.

````python
# fairlearn 0.15.0 sobre los conteos de HMDA versionados: tasa de selección por grupo, diferencia y razón de paridad
names = [g["group"] for g in race["groups"]]
ones = [np.r_[np.ones(groups[n][0]), np.zeros(groups[n][1] - groups[n][0])] for n in names]
decision = np.concatenate(ones).astype(int)
membership = np.concatenate([np.repeat(n, groups[n][1]) for n in names])
engine_rates = metrics.group_selection_rates(decision, membership)
frame = MetricFrame(metrics=selection_rate, y_true=decision, y_pred=decision, sensitive_features=membership)
print(f"{'group':44s} {'tasa del motor':>12s} {'tasa de fairlearn':>15s} {'|diferencia|':>13s}")
for name in names:
    gap = abs(engine_rates[name].rate - frame.by_group[name])
    print(f"{name:44s} {engine_rates[name].rate:12.6f} {frame.by_group[name]:15.6f} {gap:13.2e}")
highest = max(engine_rates, key=lambda g: engine_rates[g].rate)
print(
    "grupo de referencia del atributo versionado:", race["reference_group"], "| grupo de mayor tasa:", highest
)
differences = metrics.demographic_parity_difference(engine_rates, highest)
engine_difference = max(abs(v) for v in differences.values())
engine_ratio = min(metrics.demographic_parity_ratio(engine_rates, highest).values())
fl_difference = demographic_parity_difference(decision, decision, sensitive_features=membership)
fl_ratio = demographic_parity_ratio(decision, decision, sensitive_features=membership)
print(f"diferencia de paridad: motor {engine_difference:.12f} | fairlearn {fl_difference:.12f}", end=" ")
print(f"| |diferencia| {abs(engine_difference - fl_difference):.2e}")
print(f"razón de paridad: motor {engine_ratio:.12f} | fairlearn {fl_ratio:.12f}", end=" ")
print(f"| |diferencia| {abs(engine_ratio - fl_ratio):.2e}")
within = all(abs(engine_rates[n].rate - frame.by_group[n]) < 1e-12 for n in names)
within = within and abs(engine_difference - fl_difference) < 1e-12 and abs(engine_ratio - fl_ratio) < 1e-12
print("dentro de la tolerancia de 1e-12:", YN[bool(within)])
````

El paquete y el motor coinciden hasta los dígitos impresos, lo que una figura muestra como dos barras iguales. La celda
siguiente dibuja la tasa de selección de cada grupo y la diferencia y la razón de paridad desde ambas fuentes.

````python
# Tasas de selección por grupo, y diferencia y razón de paridad, motor contra fairlearn
engine_values = [engine_rates[n].rate for n in names]
fl_values = [frame.by_group[n] for n in names]
position = np.arange(len(names))
fig, (left, right) = plt.subplots(1, 2, figsize=(10.5, 3.8), gridspec_kw={"width_ratios": [2, 1]})
left.bar(position - 0.2, engine_values, width=0.4, label="motor")
left.bar(position + 0.2, fl_values, width=0.4, label="fairlearn")
left.set_xticks(position)
left.set_xticklabels([n[:14] for n in names], fontsize=6, rotation=20)
left.set_ylabel("tasa de selección")
left.legend(fontsize=8)
right.bar([-0.2, 0.8], [engine_difference, engine_ratio], width=0.4, label="motor")
right.bar([0.2, 1.2], [fl_difference, fl_ratio], width=0.4, label="fairlearn")
right.set_xticks([0, 1])
right.set_xticklabels(["diferencia de paridad", "razón de paridad"], fontsize=8)
right.legend(fontsize=8)
fig.tight_layout()
plt.show()
````

Las barras pareadas son indistinguibles en los seis grupos, con diferencia de 0 en cada tasa, y las dos barras de la
diferencia de paridad (0,280754) y de la razón de paridad (0,637966) también coinciden, hasta unas pocas unidades en el
decimal dieciséis. White es el grupo de mayor tasa, así que la mayor diferencia y la menor razón del motor contra él son
exactamente las definiciones que usa fairlearn. Lo que esto no cubre son las diferencias de probabilidades igualadas y de

### Lo que se corrió con los paquetes, y lo que no pudo correrse

El propio conjunto de pruebas del motor compara con MAPIE y fairlearn bajo su extra `ml` (`mapie>=1.5,<2`, `fairlearn>=0.15,<0.16`). El entorno de este notebook fija `mapie==1.5.0`, `fairlearn==0.15.0` y `scikit-learn==1.9.1` (la versión de la cadena de cálculo, que ambos necesitan), así que las dos comparaciones de arriba corren los paquetes mismos. Lo que sigue sin correrse: las diferencias de probabilidades igualadas y de igualdad de oportunidad de fairlearn, porque el documento C21 versionado contiene conteos de aprobaciones por grupo y ningún resultado con etiquetas verdaderas; y NannyML y SHAP, que no están en el entorno: la fórmula de CBPE del motor es la que da la documentación de NannyML (notebook 04), y la igualdad de los números no se probó aquí.

### La evidencia, dibujada

La figura muestra la grilla del Método: para cada tamaño de calibración, la proporción de los 99 valores de $\alpha$ para los que la llamada `higher` de NumPy y la regla ingenua en flotantes devuelven un estadístico de orden distinto del del motor; la proporción de pares respondidos en que el paquete MAPIE difiere del motor; y la proporción de los 99 valores para los que MAPIE rechaza un par que el motor responde.

````python
# Para cada tamaño de calibración, la proporción de valores de alfa en que cada regla difiere del motor
sizes = list(range(5, 400, 5))
alphas = np.arange(1, 100) / 100
share_numpy, share_naive, share_mapie, share_refused = [], [], [], []
for n in sizes:
    s = np.sort(rng.random(n))
    engine = np.array([conformal.split_conformal_quantile(s, float(a)) for a in alphas])
    numpy_values = np.array([numpy_rule(s, a) for a in alphas])
    naive_values = np.array([naive_rule(s, a) for a in alphas])
    mapie_list = [mapie_quantile(s, float(a)) for a in alphas]
    mapie_values = np.array([np.nan if v is None else v for v in mapie_list])
    finite = np.isfinite(engine)
    answered = np.array([v is not None for v in mapie_list])
    both = finite & answered
    share_numpy.append(float(np.mean(numpy_values[finite] != engine[finite])))
    share_naive.append(float(np.mean(naive_values[finite] != engine[finite])))
    share_mapie.append(float(np.mean(mapie_values[both] != engine[both])) if both.any() else float("nan"))
    share_refused.append(float(np.mean(finite & ~answered)))
means = [float(np.nanmean(v)) for v in (share_numpy, share_naive, share_mapie, share_refused)]
print("n:", sizes[0], "...", sizes[-1], "| alfa:", alphas[0], "...", alphas[-1])
for label, mean in zip(
    (
        "llamada higher de NumPy",
        "regla ingenua de flotantes",
        "paquete MAPIE, pares respondidos",
        "MAPIE se niega, el motor responde",
    ),
    means,
):
    print(f"{label:34s} proporción media sobre los tamaños {mean:.4f}")
````

El recorrido contó pares; la celda siguiente dibuja la misma comparación en función del tamaño de calibración. Grafica,
para cada tamaño, la proporción de valores de alfa en que cada regla devuelve otro estadístico de orden que el motor, y
la proporción en que MAPIE rechaza un par que el motor responde.

````python
# Proporción de valores de alfa en que cada regla devuelve otro estadístico de orden que el motor, según n
fig, ax = plt.subplots(figsize=(7.5, 3.8))
ax.plot(sizes, share_numpy, label="llamada higher de NumPy")
ax.plot(sizes, share_naive, label="regla ingenua de flotantes")
ax.plot(sizes, share_mapie, label="paquete MAPIE, pares respondidos")
ax.plot(sizes, share_refused, linestyle="--", label="MAPIE se niega, el motor responde")
ax.set_xlabel("tamaño de calibración n")
ax.set_ylabel("proporción de valores de alfa")
ax.legend(fontsize=8)
fig.tight_layout()
plt.show()
````

La llamada `higher` de NumPy difiere del motor en cerca de un valor de alfa de cada dos en todo tamaño (19.525 de los
39.105 pares del recorrido), lo que confirma que no es la regla que aplica el paquete MAPIE. El paquete MAPIE y la regla
ingenua de flotantes se mantienen cerca de cero y se mueven juntos, siendo los 227 pares en que MAPIE difiere aquellos en
que devuelve el puntaje de la regla ingenua. Los rechazos de MAPIE se concentran en los tamaños pequeños, donde n es
menor que 1/alfa o 1/(1-alfa).
